### Part 1: Big Data Fundamentals & Architecture

#### The 5 V's of Big Data

| V            | Meaning                                      | Example                                                         |
| ------------ | -------------------------------------------- | --------------------------------------------------------------- |
| **Volume**   | Massive amount of data                       | Petabytes of log data, social media posts                       |
| **Velocity** | Speed at which data is generated & processed | Real-time clickstreams, IoT sensor feeds                        |
| **Variety**  | Different types and formats of data          | Structured (SQL), semi-structured (JSON), unstructured (images) |
| **Veracity** | Accuracy and trustworthiness of data         | Noisy sensor data, incomplete records, duplicates               |
| **Value**    | Business insights derived from data          | Customer churn prediction, fraud detection                      |

#### Horizontal vs Vertical Scaling

| Aspect          | Vertical (Scale Up)             | Horizontal (Scale Out)       |
| --------------- | ------------------------------- | ---------------------------- |
| Method          | Add more CPU/RAM to one machine | Add more machines            |
| Cost            | Expensive (bigger hardware)     | Cheaper (commodity hardware) |
| Limit           | Physical hardware limit         | Virtually unlimited          |
| Fault Tolerance | Single point of failure         | Redundancy across nodes      |
| Big Data?       | Not suitable                    | Preferred approach           |

#### Data Architecture Patterns

| Pattern            | Description                                           | Use Case                       |
| ------------------ | ----------------------------------------------------- | ------------------------------ |
| **Data Lake**      | Store raw data in any format (schema-on-read)         | Exploration, ML, cheap storage |
| **Data Warehouse** | Store structured, cleaned data (schema-on-write)      | BI, reporting, analytics       |
| **Data Lakehouse** | Lake + warehouse capabilities (ACID, SQL, governance) | Modern unified platform        |
| **Data Mesh**      | Decentralized data ownership by domain teams          | Large organizations, autonomy  |
| **Data Fabric**    | AI-driven data management across all environments     | Multi-cloud, auto-discovery    |

#### Batch vs Stream Processing

| Aspect     | Batch Processing         | Stream Processing                     |
| ---------- | ------------------------ | ------------------------------------- |
| Latency    | Minutes to hours         | Milliseconds to seconds               |
| Data       | Bounded (finite dataset) | Unbounded (continuous)                |
| Tools      | Spark, MapReduce, Hive   | Kafka Streams, Flink, Spark Streaming |
| Use Case   | ETL, reporting, backfill | Fraud detection, real-time dashboards |
| Complexity | Simpler to debug/test    | Harder (state, ordering, late data)   |

#### Schema-on-Read vs Schema-on-Write

| Aspect              | Schema-on-Read        | Schema-on-Write            |
| ------------------- | --------------------- | -------------------------- |
| When schema applied | At query time         | At write/load time         |
| Where               | Data lake             | Data warehouse             |
| Flexibility         | High (any format)     | Low (strict schema)        |
| Data quality        | Lower (raw data)      | Higher (validated)         |
| Tools               | Spark, Presto, Athena | RDBMS, Snowflake, Redshift |

#### ETL vs ELT

| Aspect      | ETL                            | ELT                            |
| ----------- | ------------------------------ | ------------------------------ |
| Transform   | Before loading                 | After loading                  |
| Where       | ETL tool (Informatica, Talend) | In warehouse/lake (SQL, Spark) |
| Storage     | Warehouse only                 | Lake + warehouse               |
| Modern?     | Traditional (on-prem)          | Modern (cloud-native)          |
| Scalability | Limited by ETL tool            | Scales with cloud compute      |

### Part 2: Hadoop Ecosystem

#### Hadoop Overview

Apache Hadoop is an open-source framework for **distributed storage and processing** of large datasets on clusters of commodity hardware. The core ecosystem:

![Hadoop Ecosystem](https://media.geeksforgeeks.org/wp-content/cdn-uploads/HadoopEcosystem-min.png)

*Source: [GeeksforGeeks — Hadoop Ecosystem](https://www.geeksforgeeks.org/hadoop-ecosystem/)*

#### HDFS Architecture

| Component              | Role                      | Details                                                                                         |
| ---------------------- | ------------------------- | ----------------------------------------------------------------------------------------------- |
| **NameNode**           | Master — metadata manager | Stores file→block mapping, permissions, replication info. Single point of failure (use HA mode) |
| **DataNode**           | Worker — data storage     | Stores actual data blocks, sends heartbeats to NameNode                                         |
| **Secondary NameNode** | Checkpoint helper         | Merges edit logs with fsimage periodically (NOT a hot standby)                                  |
| **Block**              | Storage unit              | Default 128 MB; each file split into blocks                                                     |
| **Replication**        | Fault tolerance           | Default factor = 3 (rack-aware placement)                                                       |

##### HDFS Read/Write Flow

> **WRITE:** Client → NameNode (get block locations) → Client → DataNode1 → DataNode2 → DataNode3 (pipeline replication) → Pipeline ACK → Client confirms
>
> **READ:** Client → NameNode (get block locations) → Client → nearest DataNode (rack-aware)

#### YARN Architecture

| Component             | Role                                                |
| --------------------- | --------------------------------------------------- |
| **ResourceManager**   | Global resource scheduler — allocates containers    |
| **NodeManager**       | Per-node agent — manages containers on that machine |
| **ApplicationMaster** | Per-app coordinator — requests resources from RM    |
| **Container**         | Allocation unit — CPU + memory combination          |

#### MapReduce Flow

**Input** → **Split** → **Map Phase** → **Shuffle & Sort** → **Reduce Phase** → **Output**

| Map Phase | Shuffle & Sort | Reduce Phase |
| --- | --- | --- |
| Produces key-value pairs | Groups by key | Aggregation |
| Parallel execution | Network I/O | Parallel execution |

#### Common HDFS Commands

| Command                           | Description              |
| --------------------------------- | ------------------------ |
| `hdfs dfs -ls /`                  | List root directory      |
| `hdfs dfs -put local.txt /data/`  | Upload file to HDFS      |
| `hdfs dfs -get /data/file.txt .`  | Download file from HDFS  |
| `hdfs dfs -cat /data/file.txt`    | Print file contents      |
| `hdfs dfs -mkdir /data/new_dir`   | Create directory         |
| `hdfs dfs -rm -r /data/old_dir`   | Remove directory         |
| `hdfs dfs -du -h /data/`          | Check directory size     |
| `hdfs fsck /data/ -files -blocks` | File system health check |


#### Hadoop Ecosystem — Interview Questions

| #   | Question                                                | Keywords                  | Answer                                                                                                                                                                                                                                                                             |
| --- | ------------------------------------------------------- | ------------------------- | ---------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| 1   | What is Hadoop? Core components?                        | Hadoop, ecosystem         | • Open-source framework for distributed storage+processing<br>• Core: HDFS (storage), MapReduce (processing), YARN (resource mgmt)<br>• Handles petabytes on commodity hardware<br>• Ecosystem: Hive, Pig, HBase, Sqoop, Flume, ZooKeeper                                          |
| 2   | What is HDFS? How does it work?                         | HDFS, architecture        | • Hadoop Distributed File System for large-scale storage<br>• Files split into blocks (128 MB default), replicated across DataNodes<br>• NameNode: stores metadata (file→block mapping)<br>• Designed for batch processing, write-once-read-many                                   |
| 3   | NameNode vs DataNode?                                   | NameNode, DataNode        | • NameNode: master — metadata, file→block map, permissions<br>• DataNode: worker — stores actual data blocks<br>• NameNode is single point of failure (use HA mode)<br>• DataNodes send heartbeats and block reports                                                               |

Is it a backup?         | Secondary NN              | • NOT a backup/standby for NameNode<br>• Periodically merges edit logs with fsimage (checkpointing)<br>• Reduces NameNode startup time<br>• For HA, use Standby NameNode (not secondary)                                                                                           |
| 5   | HDFS block size — why 128 MB? Why not smaller?          | Block size, 128MB         | • Large blocks reduce metadata overhead on NameNode<br>• Minimize number of disk seeks<br>• Better network throughput (fewer connections)<br>• Small blocks → too many blocks → NameNode memory pressure                                                                           |
| 6   | HDFS replication factor — how does rack awareness work? | Replication, rack         | • Default replication = 3: same rack, different rack, different rack<br>• Rack-aware: first replica on local rack, others on remote racks<br>• Survives individual node AND entire rack failure<br>• Configurable per file: `hdfs dfs -setrep -w 2 /path`                          |
| 7   | HDFS read and write flow?                               | HDFS, read, write         | • Write: client → NameNode (locations) → pipeline: DN1→DN2→DN3<br>• Pipeline replication with ACK propagation<br>• Read: client → NameNode (locations) → nearest DataNode (rack-aware)<br>• Data locality: process data where it resides                                           |
| 8   | What is HDFS Federation?                                | Federation, NameNode      | • Multiple NameNodes each managing a portion of namespace<br>• Scales NameNode horizontally (no single bottleneck)<br>• Each NameNode has its own namespace volume<br>• DataNodes are shared across all NameNodes                                                                  |
| 9   | HDFS High Availability — how implemented?               | HA, standby               | • Active NameNode + Standby NameNode (hot standby)<br>• Shared edit log via JournalNodes (quorum-based)<br>• ZooKeeper handles automatic failover (ZKFC)<br>• Eliminates NameNode as single point of failure                                                                       |
| 10  | What is YARN? Architecture?                             | YARN, resource            | • Yet Another Resource Negotiator — cluster resource manager<br>• ResourceManager: global scheduler, allocates containers<br>• NodeManager: per-node agent, manages containers<br>• ApplicationMaster: per-application coordinator                                                 |
| 11  | YARN schedulers — Fair vs Capacity?                     | Scheduler, fair, capacity | • Fair: equal share of resources to all running apps<br>• Capacity: guaranteed capacity per queue (multi-tenant)<br>• FIFO: oldest-first (rarely used in production)<br>• Capacity Scheduler is default in most Hadoop distributions                                               |
| 12  | Explain MapReduce paradigm.                             | MapReduce, paradigm       | • Programming model for parallel processing<br>• Map phase: process input key-value pairs → intermediate pairs<br>• Shuffle & Sort: group by key across network<br>• Reduce phase: aggregate values per key → final output                                                         |
| 13  | Why is MapReduce slow compared to Spark?                | MapReduce, slow           | • Writes intermediate results to disk after every stage<br>• No in-memory caching between iterations<br>• Rigid two-phase model (map→reduce only)<br>• Spark uses in-memory computation and DAG optimization                                                                       |
| 14  | What is a Combiner in MapReduce?                        | Combiner, optimization    | • Mini-Reducer that runs on Mapper's output locally<br>• Reduces data transferred during shuffle phase<br>• Same interface as Reducer (but runs per-mapper)<br>• Example: word count — combine local counts before shuffle                                                         |
| 15  | InputFormat and OutputFormat in MapReduce?              | InputFormat, output       | • InputFormat: determines how input is split and read (TextInputFormat, SequenceFileInputFormat)<br>• OutputFormat: determines how output is written (TextOutputFormat)<br>• Custom InputFormat for non-standard data sources<br>• Controls RecordReader and RecordWriter behavior |
| 16  | What is speculative execution in Hadoop?                | Speculative, straggler    | • Hadoop launches duplicate of slow-running tasks<br>• First finished task's output is accepted; other is killed<br>• Handles hardware degradation and data skew<br>• Can be disabled: `mapreduce.map.speculative=false`                                                           |
| 17  | What is ZooKeeper? Use cases in Hadoop?                 | ZooKeeper, coordination   | • Distributed coordination service for config, naming, sync<br>• Maintains configuration information reliably<br>• Use cases: NameNode HA failover, HBase master election<br>• Uses Zab protocol for consensus (like Paxos/Raft)                                                   |
| 18  | What is Sqoop? How does it work?                        | Sqoop, RDBMS              | • Tool for transferring data between RDBMS and HDFS<br>• Import: RDBMS → HDFS/Hive (uses MapReduce for parallel transfer)<br>• Export: HDFS → RDBMS<br>• Connects via JDBC, splits by primary key for parallelism                                                                  |
| 19  | Sqoop import — incremental modes?                       | Sqoop, incremental        | • Full import: entire table every time<br>• Incremental append: only new rows (--incremental append --check-column id)<br>• Incremental lastmodified: rows modified since last import<br>• Incremental avoids re-transferring existing data                                        |
| 20  | What is Flume? Architecture?                            | Flume, log ingestion      | • Service for collecting and moving large log/event data<br>• Sources → Channels → Sinks architecture<br>• Source: where data comes from (syslog, HTTP, Kafka)<br>• Channel: buffer (memory or file); Sink: destination (HDFS, HBase)                                              |
| 21  | Data locality in Hadoop — why important?                | Data locality             | • Process data where it's stored (avoid network transfer)<br>• YARN schedules tasks on nodes with data blocks<br>• Three levels: node-local → rack-local → off-rack<br>• Critical for performance in data-intensive workloads                                                      |

### Part 3: Apache Hive

#### Overview

Apache Hive is a **SQL-based data warehouse** built on top of Hadoop. It translates SQL queries (HiveQL) into MapReduce/Tez/Spark jobs for processing data stored in HDFS or cloud storage.

#### Architecture

![Hive Architecture](https://cwiki.apache.org/confluence/download/attachments/27362072/system_architecture.png)

*Source: [Apache Hive Wiki — Architecture](https://cwiki.apache.org/confluence/display/Hive/Design)*

#### Key Concepts

| Concept            | Description                                                                                             |
| ------------------ | ------------------------------------------------------------------------------------------------------- |
| **Metastore**      | Central schema repository — stores table schemas, partitions, column types (backed by RDBMS like MySQL) |
| **SerDe**          | Serializer/Deserializer — defines how to read/write data formats (JSON, CSV, Avro, ORC)                 |
| **Internal Table** | Hive manages both metadata AND data; DROP deletes both                                                  |
| **External Table** | Hive manages only metadata; data lives in external location; DROP keeps data                            |
| **Partitioning**   | Subdirectories by column values (e.g., `/year=2024/month=01/`); enables partition pruning               |
| **Bucketing**      | Hashes data into fixed number of files by column; improves join and sampling performance                |
| **ORC Format**     | Optimized Row Columnar — Hive's preferred format with compression and predicate pushdown                |
| **Tez**            | DAG-based execution engine replacing MapReduce (10x faster for Hive queries)                            |

#### Partitioning vs Bucketing

| Aspect      | Partitioning                  | Bucketing                           |
| ----------- | ----------------------------- | ----------------------------------- |
| Storage     | Subdirectories                | Fixed number of files               |
| Key         | Column values (date, region)  | Hash of column                      |
| Pruning     | Partition pruning (skip dirs) | Not prunable directly               |
| Performance | Great for filter queries      | Great for joins/sampling            |
| Cardinality | Low cardinality columns       | High cardinality columns            |
| Syntax      | `PARTITIONED BY (col)`        | `CLUSTERED BY (col) INTO n BUCKETS` |

#### Internal vs External Tables

| Aspect        | Internal (Managed)      | External                 |
| ------------- | ----------------------- | ------------------------ |
| Data owned by | Hive                    | User / external system   |
| DROP behavior | Deletes metadata + data | Deletes metadata only    |
| Use case      | Hive-only data          | Shared data across tools |
| Location      | Default warehouse dir   | Any HDFS/S3 location     |

### Part 4: Apache HBase

#### Overview

Apache HBase is a **NoSQL, column-family, wide-column store** built on top of HDFS. It provides **random, real-time read/write access** to Big Data — filling the gap HDFS has (batch-only, no random access).

#### Data Model

![HBase Data Model](https://www.tutorialspoint.com/hbase/images/table.jpg)

*Source: [TutorialsPoint — HBase Architecture](https://www.tutorialspoint.com/hbase/hbase_architecture.htm)*

**Key Points:** Each row key maps to column families; columns within a family are stored together; cells can be empty (sparse storage).

#### Architecture

| Component                 | Role                                                               |
| ------------------------- | ------------------------------------------------------------------ |
| **HMaster**               | Cluster coordinator — assigns regions, handles DDL, load balancing |
| **RegionServer**          | Worker — serves reads/writes for assigned regions                  |
| **Region**                | Subset of a table's rows (contiguous range of row keys)            |
| **WAL (Write-Ahead Log)** | Durability — logs writes before applying to MemStore               |
| **MemStore**              | In-memory write buffer per column family                           |
| **HFile**                 | On-disk sorted file (stored in HDFS)                               |
| **ZooKeeper**             | Cluster coordination, RegionServer health, master election         |

#### Read/Write Path

> **WRITE:** Client → RegionServer → WAL (durability) → MemStore (memory) → MemStore flush → HFile (disk/HDFS)
>
> **READ:** Client → BlockCache (if cached) → MemStore → HFile
>
> Bloom filters speed up HFile lookups

#### Compaction

| Type      | Description                                                                    |
| --------- | ------------------------------------------------------------------------------ |
| **Minor** | Merges several small HFiles into fewer larger ones (fast, no deletions)        |
| **Major** | Merges ALL HFiles for a region into one, applies deletions and TTL (expensive) |

### Part 5: Apache Cassandra

#### Overview

Apache Cassandra is a **distributed, peer-to-peer, wide-column NoSQL database** designed for **high availability and massive write throughput**. It sacrifices strong consistency for availability (AP in CAP theorem).

#### Architecture — Peer-to-Peer (No Master)

![Cassandra Architecture](https://cassandra.apache.org/_/_images/diagrams/apache-cassandra-diagrams-01.jpg)

*Source: [Apache Cassandra Official](https://cassandra.apache.org/)*

#### Data Model

| Concept            | Description                                                                |
| ------------------ | -------------------------------------------------------------------------- |
| **Keyspace**       | Top-level namespace (like database in RDBMS), defines replication strategy |
| **Table**          | Collection of rows, defined by CREATE TABLE                                |
| **Partition Key**  | Determines which node stores the data (hashed by partitioner)              |
| **Clustering Key** | Determines sort order within a partition                                   |
| **Primary Key**    | = Partition Key + Clustering Key(s)                                        |
| **Column**         | Name-value pair within a row                                               |

#### Consistency Levels

| Level            | Meaning                                                            |
| ---------------- | ------------------------------------------------------------------ |
| **ONE**          | Only 1 replica must respond (fastest, least consistent)            |
| **QUORUM**       | Majority of replicas must respond (balances consistency + speed)   |
| **ALL**          | All replicas must respond (strongest consistency, highest latency) |
| **LOCAL_QUORUM** | Quorum within local data center (multi-DC deployments)             |

#### Write Path

> **WRITE:** Client → Coordinator Node → Commit Log (durability) → MemTable (memory) → MemTable flush → SSTable (immutable sorted file on disk)
>
> SSTables compacted periodically (merge + remove tombstones)

#### Read Path

> **READ:** Client → Coordinator → Check MemTable → Check SSTables
>
> - Bloom filter: skip SSTables that don't have the key
> - Partition index + compression offsets → locate data
> - Merge results from MemTable + SSTables → return

### Part 6: Apache Spark — Core Concepts

#### What is Apache Spark?

Apache Spark is a **distributed data processing engine** designed for large-scale analytics and ML. Unlike MapReduce (disk-based), Spark performs **in-memory computation** using a **DAG execution model**, making it 10-100x faster for iterative and interactive workloads.

#### Spark Architecture

![Spark Architecture](https://spark.apache.org/docs/latest/img/cluster-overview.png)

*Source: [Apache Spark Official Documentation](https://spark.apache.org/docs/latest/cluster-overview.html)*

##### Component Roles

| Component           | Role                                 | Key Details                                         |
| ------------------- | ------------------------------------ | --------------------------------------------------- |
| **Driver**          | Brain — creates DAG, schedules tasks | Runs main(), creates SparkSession, collects results |
| **Executor**        | Muscle — runs tasks, stores data     | JVM process on worker node, reports to driver       |
| **Cluster Manager** | Resource allocator                   | YARN, Mesos, Kubernetes, Standalone                 |
| **SparkSession**    | Unified entry point                  | Replaces SQLContext + HiveContext + SparkContext    |
| **SparkContext**    | Low-level RDD entry point            | Accessible as `spark.sparkContext`                  |

#### Spark APIs (RDD → DataFrame → Dataset)

| Feature      | RDD                        | DataFrame             | Dataset (Scala/Java)  |
| ------------ | -------------------------- | --------------------- | --------------------- |
| Abstraction  | Low-level objects          | Named columns (table) | Typed objects + table |
| Schema       | No schema                  | Schema-aware          | Schema + type-safe    |
| Optimization | None (manual)              | Catalyst + Tungsten   | Catalyst + Tungsten   |
| Language     | Scala, Java, Python        | All                   | Scala, Java only      |
| Performance  | Slowest                    | Fast (optimized)      | Fast + type-safe      |
| Best for     | Custom logic, unstructured | 90% of workloads      | Type-safe structured  |


#### DAG — Directed Acyclic Graph

When an **action** is triggered (e.g., `count()`, `collect()`), Spark builds a DAG of transformations and breaks it into **stages** and **tasks**:

**Transformations** (lazy) → **Action** (eager) triggers DAG execution:

> `read` → `filter` → `map` → `groupBy` → `count`
>
> (narrow transforms) &nbsp;&nbsp;&nbsp;&nbsp;&nbsp; (wide = shuffle)

- Stage boundary = shuffle (wide transformation)
- Each stage → N tasks (one per partition)
- Tasks sent to executors for parallel execution

#### Narrow vs Wide Transformations

| Type       | Data Movement                                  | Stage Boundary | Examples                                          |
| ---------- | ---------------------------------------------- | -------------- | ------------------------------------------------- |
| **Narrow** | No shuffle — each partition → one partition    | No             | map, filter, flatMap, withColumn, select          |
| **Wide**   | Shuffle — data redistributed across partitions | Yes            | groupBy, reduceByKey, join, distinct, repartition |

#### Lazy Evaluation

Spark **defers execution** of transformations until an action is called:
- Allows Catalyst to optimize the entire plan before execution
- Avoids unnecessary computations
- Transformations: build the plan (lazy)
- Actions: trigger execution (eager)

| Transformations (Lazy)    | Actions (Eager)              |
| ------------------------- | ---------------------------- |
| select, filter, map, join | count, collect, show, write  |
| withColumn, groupBy, agg  | first, take, reduce, foreach |
| union, distinct, cache    | saveAsTable, toPandas        |

#### Catalyst Optimizer + Tungsten Engine

![Catalyst Optimizer Pipeline](https://www.databricks.com/wp-content/uploads/2018/05/Catalyst-Optimizer-diagram.png)

*Source: [Databricks — Catalyst Optimizer](https://www.databricks.com/glossary/catalyst-optimizer)*

> **Catalyst** = decides WHAT & HOW (query planning) — Brain of Spark SQL
>
> **Tungsten** = executes FAST (off-heap memory, WSCG, cache-aware) — Muscle

##### Catalyst Optimizations
- **Predicate pushdown**: push filters closer to data source
- **Column pruning**: read only needed columns
- **Constant folding**: evaluate constant expressions at compile time
- **Join reordering**: optimize join order based on statistics

##### Tungsten Optimizations
- **Off-heap memory**: bypasses JVM garbage collection
- **Whole-Stage Code Generation (WSCG)**: compiles query stages into optimized JVM bytecode
- **Cache-aware computation**: organizes data for CPU cache efficiency
- **Binary processing**: operates on binary data directly, avoiding serialization overhead

**Dynamic boundary:** Execution can borrow from Storage (evicts cached data) and vice versa.

| Config                          | Default | Purpose                                        |
| ------------------------------- | ------- | ---------------------------------------------- |

#### Cache vs Persist

| Method              | Description                                               |
| ------------------- | --------------------------------------------------------- |
| `df.cache()`        | Shortcut for `persist(MEMORY_ONLY)`                       |
| `df.persist(level)` | Choose storage level (MEMORY_ONLY, MEMORY_AND_DISK, etc.) |
| `df.unpersist()`    | Remove from cache                                         |

**Storage Levels:**
- `MEMORY_ONLY` — fastest, loses data if not enough memory
- `MEMORY_AND_DISK` — spills to disk if memory runs out
- `DISK_ONLY` — only on disk (slow but safe)
- `MEMORY_ONLY_2` — with 2x replication (fault-tolerant)

#### Adaptive Query Execution (AQE) — Spark 3.0+

AQE dynamically optimizes queries at runtime based on actual statistics:

| Feature               | Description                                                        |
| --------------------- | ------------------------------------------------------------------ |
| Dynamic join strategy | Switches sort-merge to broadcast if input is smaller than expected |
| Skew join handling    | Splits skewed partitions into smaller tasks                        |
| Partition coalescing  | Reduces empty/small shuffle partitions                             |
| Enable                | `spark.sql.adaptive.enabled=true` (default in Spark 3.2+)          |

| **Cluster** | On cluster node         | Production, scheduled jobs |
| **Local**   | Single JVM (`local[*]`) | Development, testing       |


````python
# ──────────────────────────────────────────────
# Spark Core — exploring the execution model
# ──────────────────────────────────────────────

# 1. RDD basics
rdd = sc.parallelize([1, 2, 3, 4, 5, 6, 7, 8], 4)  # 4 partitions
print(f'Partitions: {rdd.getNumPartitions()}')
print(f'Sum: {rdd.reduce(lambda a, b: a + b)}')

# 2. Transformations vs Actions
# Transformations are LAZY (build DAG)
filtered = rdd.filter(lambda x: x > 3)      # lazy
mapped = filtered.map(lambda x: x * 10)      # lazy
# Action triggers execution
print(f'Result: {mapped.collect()}')           # [40, 50, 60, 70, 80]

# 3. View execution plan
df = spark.range(1000)
result = df.filter('id > 500').groupBy((F.col('id') % 10).alias('bucket')).count()
result.explain(True)  # Shows Catalyst optimization plan
````

````python
# ──────────────────────────────────────────────
# Accumulators and Broadcast Variables
# ──────────────────────────────────────────────

# Accumulator — aggregation variable shared across tasks
error_count = sc.accumulator(0)

def process_row(x):
    global error_count
    if x < 0:
        error_count.add(1)
    return abs(x)

data = sc.parallelize([1, -2, 3, -4, 5])
result = data.map(process_row).collect()
print(f'Processed: {result}')
print(f'Errors found: {error_count.value}')  # 2

# Broadcast Variable — read-only variable sent once to all executors
lookup = {'A': 'Alice', 'B': 'Bob', 'C': 'Carol'}
broadcast_lookup = sc.broadcast(lookup)

codes = sc.parallelize(['A', 'B', 'C', 'A'])
names = codes.map(lambda c: broadcast_lookup.value.get(c, 'Unknown'))
print(f'Names: {names.collect()}')  # ['Alice', 'Bob', 'Carol', 'Alice']
````

````python
# ──────────────────────────────────────────────
# PySpark DataFrame operations
# ──────────────────────────────────────────────

# Create sample data
emp_data = [
    (1, 'Alice',   'Sales',       65000),
    (2, 'Bob',     'Engineering', 85000),
    (3, 'Carol',   'Sales',       70000),
    (4, 'Dave',    'Engineering', 90000),
    (5, 'Eve',     'HR',          55000),
    (6, 'Frank',   'Engineering', 75000),
]
emp = spark.createDataFrame(emp_data, ['id', 'name', 'dept', 'salary'])

# Filter + Select
emp.filter(F.col('salary') > 70000).select('name', 'dept', 'salary').show()

# GroupBy + Aggregations
emp.groupBy('dept').agg(
    F.count('*').alias('count'),
    F.avg('salary').alias('avg_salary'),
    F.max('salary').alias('max_salary')
).orderBy('avg_salary', ascending=False).show()

# Window functions — rank within department
w = Window.partitionBy('dept').orderBy(F.col('salary').desc())
emp.withColumn('rank', F.rank().over(w)) \
   .withColumn('running_total', F.sum('salary').over(w)) \
   .show()
````

````python
# ──────────────────────────────────────────────
# Joins and SparkSQL
# ──────────────────────────────────────────────

# Department lookup table
dept_data = [('Sales', 'NYC'), ('Engineering', 'SF'), ('HR', 'Chicago')]
dept = spark.createDataFrame(dept_data, ['dept', 'location'])

# Inner join
emp.join(dept, 'dept', 'inner').show()

# Broadcast join (small table optimization)
emp.join(F.broadcast(dept), 'dept', 'inner').explain()  # Check plan

# SparkSQL — register views and query
emp.createOrReplaceTempView('employees')
dept.createOrReplaceTempView('departments')

spark.sql('''
    SELECT e.name, e.dept, e.salary, d.location,
           RANK() OVER (PARTITION BY e.dept ORDER BY e.salary DESC) as rank
    FROM employees e
    JOIN departments d ON e.dept = d.dept
    WHERE e.salary > 60000
    ORDER BY e.dept, rank
''').show()
````

````python
# ──────────────────────────────────────────────
# Complex types and UDFs
# ──────────────────────────────────────────────

# Array type operations
data = [(1, ['Python', 'Spark', 'SQL']), (2, ['Java', 'Spark']), (3, ['Python'])]
df_arr = spark.createDataFrame(data, ['id', 'skills'])
df_arr.select('id', F.explode('skills').alias('skill')).show()
df_arr.select('id', F.size('skills').alias('num_skills')).show()

# Struct (nested) type
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
schema = StructType([
    StructField('name', StringType()),
    StructField('address', StructType([
        StructField('city', StringType()),
        StructField('zip', StringType())
    ]))
])
nested_data = [('Alice', ('NYC', '10001')), ('Bob', ('SF', '94102'))]
df_nested = spark.createDataFrame(nested_data, schema)
df_nested.select('name', 'address.city', 'address.zip').show()

# UDF vs built-in function
from pyspark.sql.functions import udf
@udf(StringType())
def upper_custom(s):
    return s.upper() if s else None

# UDF (slower — avoid when possible)
emp.select('name', upper_custom('name').alias('upper_udf')).show(3)
# Built-in (faster — preferred)
emp.select('name', F.upper('name').alias('upper_builtin')).show(3)
````

#### Watermarks

Watermarks handle **late-arriving data** in event-time processing:

```python
df = spark.readStream.format('kafka').load()
df.withWatermark('event_time', '10 minutes') # tolerate 10 min late data
 .groupBy(F.window('event_time', '5 minutes'), 'user_id')
 .count()
```

- Data arriving within watermark threshold → included in results
- Data arriving after threshold → dropped
- Enables Spark to clean up old state

### Part 9: Delta Lake & Modern Data Lakes

#### What is Delta Lake?

Delta Lake is an **open-source storage layer** that brings **ACID transactions, schema enforcement, and time travel** to data lakes. It sits on top of Parquet and adds a **transaction log (_delta_log)** for reliability.

#### Delta Lake vs Traditional Data Lake

| Feature             | Raw Parquet/CSV Lake   | Delta Lake                |
| ------------------- | ---------------------- | ------------------------- |
| ACID Transactions   |                        | Atomic commits            |
| Schema Enforcement  | Silent corruption      | Rejects bad schemas       |
| Schema Evolution    | Manual                 | mergeSchema option        |
| Time Travel         |                        | VERSION AS OF / TIMESTAMP |
| UPDATE/DELETE/MERGE | Rewrite entire dataset | Native DML support        |
| Data Skipping       |                        | Min/max stats per file    |
| Concurrent Writes   | Corruption risk        | Optimistic concurrency    |
| Streaming + Batch   | Separate pipelines     | Unified on same table     |

#### Transaction Log (_delta_log)

![Delta Lake Medallion Architecture](https://www.databricks.com/wp-content/uploads/2020/09/delta-lake-medallion-model-scaled.jpg)

![Delta Transaction Log](https://www.databricks.com/wp-content/uploads/2019/08/image7.png)

*Source: [Databricks — Unpacking the Transaction Log](https://www.databricks.com/blog/2019/08/21/diving-into-delta-lake-unpacking-the-transaction-log.html)*

Each commit is a JSON file recording file-level actions (add/remove). Checkpoints are Parquet summaries for faster log replay. This enables ACID, time travel, and concurrent access.

#### Key Operations

| Operation      | SQL                                                                 | PySpark                             | Purpose                           |
| -------------- | ------------------------------------------------------------------- | ----------------------------------- | --------------------------------- |
| **MERGE INTO** | `MERGE INTO target USING source ON condition WHEN MATCHED THEN ...` | `DeltaTable.merge()`                | Upsert (insert+update)            |
| **UPDATE**     | `UPDATE table SET col=val WHERE ...`                                | `deltaTable.update(condition, set)` | Modify rows                       |
| **DELETE**     | `DELETE FROM table WHERE ...`                                       | `deltaTable.delete(condition)`      | Remove rows                       |
| **OPTIMIZE**   | `OPTIMIZE table`                                                    | —                                   | Compact small files               |
| **ZORDER**     | `OPTIMIZE table ZORDER BY (col)`                                    | —                                   | Co-locate data for faster queries |
| **VACUUM**     | `VACUUM table RETAIN 168 HOURS`                                     | —                                   | Remove old files                  |

#### Schema Enforcement vs Evolution

| Feature  | Enforcement                | Evolution                        |
| -------- | -------------------------- | -------------------------------- |
| Purpose  | Reject incompatible writes | Allow new columns                |
| Default  | Always ON                  | Must enable                      |
| Enable   | Automatic                  | `.option('mergeSchema', 'true')` |
| Use case | Prevent corruption         | Evolving data sources            |

#### Time Travel

```sql
-- Query by version number
SELECT * FROM my_table VERSION AS OF 5;

-- Query by timestamp
SELECT * FROM my_table TIMESTAMP AS OF '2025-01-01T12:00:00';

-- PySpark
spark.read.format('delta').option('versionAsOf', 5).load('/path')
spark.read.format('delta').option('timestampAsOf', '2025-01-01').load('/path')

-- View history
DESCRIBE HISTORY my_table;
```

### Part 11: Apache Kafka

#### Overview

Apache Kafka is a **distributed event streaming platform** for high-throughput, fault-tolerant, real-time data pipelines and event-driven architectures.

#### Architecture

![Kafka Cluster Architecture](https://www.tutorialspoint.com/apache_kafka/images/cluster_architecture.jpg)

*Source: [TutorialsPoint — Apache Kafka](https://www.tutorialspoint.com/apache_kafka/apache_kafka_cluster_architecture.htm)*

#### Core Concepts

| Concept                    | Description                                                         |
| -------------------------- | ------------------------------------------------------------------- |
| **Topic**                  | Named stream of messages (like a table)                             |
| **Partition**              | Ordered, immutable sequence within a topic; enables parallelism     |
| **Offset**                 | Position of a message within a partition (monotonically increasing) |
| **Broker**                 | Kafka server that stores data and serves clients                    |
| **Producer**               | Publishes messages to topics                                        |
| **Consumer**               | Reads messages from topics                                          |
| **Consumer Group**         | Group of consumers sharing partition load                           |
| **Replication Factor**     | Number of copies per partition (typically 3)                        |
| **Leader/Follower**        | One leader handles reads/writes; followers replicate                |
| **ISR (In-Sync Replicas)** | Followers fully caught up with leader                               |

#### Kafka Ecosystem

| Component           | Purpose                                                                    |
| ------------------- | -------------------------------------------------------------------------- |
| **Kafka Connect**   | Pre-built connectors for data ingestion/egestion (JDBC, S3, ElasticSearch) |
| **Kafka Streams**   | Stream processing library (Java) for transformations                       |
| **Schema Registry** | Centralized schema management (Avro, JSON Schema, Protobuf)                |
| **ksqlDB**          | SQL engine for stream processing on Kafka                                  |
| **KRaft**           | Kafka Raft — replaces ZooKeeper for metadata management (Kafka 3.3+)       |

#### Overview

Apache Airflow is a **workflow orchestration platform** for scheduling, monitoring, and managing data pipelines as code (DAGs — Directed Acyclic Graphs).

#### Key Concepts

| Concept        | Description                                                             |
| -------------- | ----------------------------------------------------------------------- |
| **DAG**        | Directed Acyclic Graph — defines task dependencies and execution order  |
| **Task**       | Unit of work within a DAG (Python function or operator)                 |
| **Operator**   | Template for a task (BashOperator, PythonOperator, SparkSubmitOperator) |
| **Sensor**     | Special operator that waits for condition (file exists, time reached)   |

| **Hook**       | Interface to external systems (database, S3, Spark)                     |
| **Connection** | Stored credentials for external systems                                 |
| **Variable**   | Global key-value config accessible across DAGs                          |

| **Celery**     | Distributed workers via Redis/RabbitMQ | Production (common)     |
| **Kubernetes** | Each task runs in its own K8s pod      | Cloud-native, isolation |

### Part 15: Data Pipeline Design & Orchestration

#### Pipeline Design Principles

| Principle           | Description                                             |
| ------------------- | ------------------------------------------------------- |
| **Idempotency**     | Re-running produces same result (no duplicates)         |
| **Reproducibility** | Same input → same output every time                     |
| **Observability**   | Monitor, log, alert on pipeline health                  |
| **Data Quality**    | Validate data at each stage                             |
| **Fault Tolerance** | Handle failures gracefully (retries, DLQ)               |
| **Data Contracts**  | Agreement between producers and consumers on schema/SLA |

### Part 16: Performance Tuning & Optimization

#### Spark Performance Tuning Checklist

| Area              | Check                        | Action                                            |
| ----------------- | ---------------------------- | ------------------------------------------------- |
| **Shuffle**       | Too many shuffle partitions? | Tune `spark.sql.shuffle.partitions` (default 200) |
| **Skew**          | Straggler tasks?             | Salting, AQE, broadcast join                      |
| **Join**          | Large shuffle reads?         | Broadcast small tables                            |
| **Cache**         | Recomputing same data?       | cache()/persist() reused DataFrames               |
| **UDF**           | Slow Python UDFs?            | Replace with built-in functions or Pandas UDFs    |
| **Serialization** | High serialization time?     | Use Kryo serialization                            |
| **GC**            | High GC time?                | Tune memory, reduce object creation               |
| **Files**         | Too many small files?        | OPTIMIZE/coalesce, avoid too many partitions      |


#### Performance Tuning — Interview Questions

| #   | Question                                       | Keywords             | Answer                                                                                                                                                                                                                                                                    |
| --- | ---------------------------------------------- | -------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| 1   | How to identify bottlenecks in Spark UI?       | Spark UI, bottleneck | • Stages tab: check for skewed tasks (one task much longer)<br>• Shuffle read/write: high values indicate expensive shuffles<br>• GC time: high GC in Executors tab = memory pressure<br>• Task duration distribution: look for stragglers                                |
| 2   | Data skew — how to detect and fix?             | Skew, detect         | • Detect: Spark UI shows one partition much larger; task times vary 10x+<br>• Fix: salting (random prefix on key), AQE (auto-splits skewed partitions)<br>• Fix: broadcast join for small-large table joins<br>• Fix: repartition with different key                      |
| 3   | Salting technique — step by step?              | Salting, steps       | • Add random salt (0 to N) as prefix to skewed key: `concat(rand_int, '-', key)`<br>• Explode other table with same salt range (cross join with salt)<br>• Join on salted key → distributes evenly<br>• After join: remove salt prefix, re-aggregate if needed            |
| 4   | Spark shuffle partitions — how to tune?        | Shuffle, partitions  | • Default: 200 (`spark.sql.shuffle.partitions`)<br>• Too few: large partitions → OOM or slow tasks<br>• Too many: overhead per small partition<br>• Rule: target ~128 MB per partition; AQE auto-coalesces                                                                |
| 5   | Broadcast join threshold?                      | Broadcast, threshold | • Default: 10 MB (`spark.sql.autoBroadcastJoinThreshold`)<br>• Increase for larger lookup tables: `.config('...', '100m')`<br>• Manual hint: `F.broadcast(df)`<br>• AQE can auto-switch to broadcast at runtime                                                           |
| 6   | When to use cache/persist? When is it harmful? | Cache, harmful       | • Good: DataFrame reused multiple times in same job<br>• Bad: single-use DataFrame (wastes memory, causes eviction)<br>• Bad: very large DataFrame (spills to disk, slow)<br>• Always unpersist after use                                                                 |
| 7   | UDF performance — why slow with Python UDFs?   | UDF, slow            | • Data serialized from JVM to Python and back (per row)<br>• Breaks Catalyst optimization (black box)<br>• Fix: use built-in functions whenever possible<br>• Fix: Pandas UDFs (vectorized, Apache Arrow, 10-100x faster)                                                 |
| 8   | Kryo serialization — when and why?             | Kryo, serialization  | • Kryo is 10x faster and more compact than Java serialization<br>• Enable: `spark.serializer=org.apache.spark.serializer.KryoSerializer`<br>• Register custom classes for best performance<br>• Used for: shuffle data, cache serialization                               |
| 9   | Small files problem in Spark?                  | Small files, compact | • Too many small files → excessive metadata, slow reads<br>• Caused by: many partitions, streaming micro-batches<br>• Fix: coalesce(n) before write, OPTIMIZE (Delta)<br>• Target: 128-256 MB per file                                                                    |
| 10  | Executor memory tuning?                        | Executor, memory     | • `spark.executor.memory`: heap size (try 8-16g per executor)<br>• `spark.executor.memoryOverhead`: 10% + additional for off-heap<br>• `spark.memory.fraction`: 0.6 default (execution+storage)<br>• Balance: fewer large executors vs. many small executors              |
| 11  | Number of executors vs cores per executor?     | Executors, cores     | • Cores per executor: 4-5 recommended (HDFS I/O degrades above 5)<br>• Too many cores: contention; too few: under-utilization<br>• Total cores = executors × cores_per_executor<br>• Leave 1 core per node for OS/YARN                                                    |
| 12  | How to reduce GC overhead?                     | GC, overhead         | • Reduce long-lived object creation (avoid collect on large data)<br>• Use off-heap storage (Tungsten, persist with DISK_ONLY)<br>• Tune G1GC: `-XX:+UseG1GC -XX:InitiatingHeapOccupancyPercent=35`<br>• Monitor GC time in Executors tab of Spark UI                     |
| 13  | Dynamic allocation benefits and config?        | Dynamic, allocation  | • Auto-scale executors based on workload<br>• `spark.dynamicAllocation.enabled=true`<br>• Requires: external shuffle service<br>• Saves cost when workload varies (scale down when idle)                                                                                  |
| 14  | Partition pruning — how to leverage?           | Partition, pruning   | • Query must include partition column in WHERE clause<br>• Spark skips reading entire partitions that don't match<br>• Write data partitioned by frequently filtered columns<br>• Check: explain() should show PartitionFilters                                           |

| 17  | File format impact on performance?             | Format, performance  | • Parquet/ORC: columnar, compressed, predicate pushdown (fast)<br>• CSV/JSON: row-based, no columnar optimization (slow)<br>• Always convert raw CSV/JSON to Parquet/Delta for analytics<br>• Delta: adds ACID + data skipping on top of Parquet                          |
| 18  | How to handle OOM errors?                      | OOM, memory          | • Increase executor memory or number of executors<br>• Repartition to create more/smaller partitions<br>• Avoid collect() on large datasets (bring to driver)<br>• Reduce broadcast variables if too large for executors                                                  |

| 20  | What is data locality in Spark?                | Data locality        | • Process data on the node where it resides<br>• Levels: PROCESS_LOCAL → NODE_LOCAL → RACK_LOCAL → ANY<br>• Spark prefers local but may schedule remotely under pressure<br>• Locality mostly relevant for HDFS; less for cloud storage                                   |
| 21  | How to profile Spark jobs?                     | Profile, metrics     | • Spark UI: stages, tasks, shuffle, cache metrics<br>• Event log: `spark.eventLog.enabled=true`<br>• Spark History Server: analyze completed jobs<br>• Metrics: Prometheus/Graphite sink for monitoring                                                                   |
| 22  | Cost vs performance tradeoffs?                 | Cost, performance    | • More executors = faster but more expensive<br>• Spot/preemptible instances: 60-90% cheaper, risk of interruption<br>• Cache frequently used data (saves recomputation cost)<br>• Optimize file format and partitioning before scaling hardware                          |
| 23  | Tungsten off-heap memory?                      | Tungsten, off-heap   | • Stores data outside JVM heap (avoids GC overhead)<br>• Managed by Spark directly (not JVM garbage collector)<br>• Enable via: `spark.memory.offHeap.enabled=true`<br>• Useful for: large memory workloads, reducing GC pauses                                           |

### Part 19: Apache Flink

#### Flink Architecture

| Component | Description |
| --- | --- |
| **JobManager** | Master: receives jobs, schedules tasks, manages checkpoints |
| **TaskManager** | Worker: runs tasks in slots (parallel pipelines) |
| **Slot** | Fixed portion of a TaskManager's resources (CPU + memory) |
| **Checkpoint** | Consistent snapshot of distributed state for fault tolerance |
| **Savepoint** | Manual checkpoint — used for upgrades, migration, A/B testing |
| **State Backend** | Where state is stored: HashMapStateBackend (heap) or EmbeddedRocksDB |

#### Flink vs Spark Streaming

| Aspect | Apache Flink | Spark Structured Streaming |
| --- | --- | --- |
| **Processing** | True event-at-a-time | Micro-batch (or continuous experimental) |
| **Latency** | Milliseconds | Seconds (micro-batch interval) |
| **State** | First-class managed state | State via grouping/aggregation |
| **Exactly-once** | Native (Chandy-Lamport checkpointing) | Via idempotent sinks + checkpoints |
| **Windows** | Flexible (tumbling, sliding, session, global) | Tumbling, sliding, session |
| **Batch** | Treats batch as bounded stream | Native batch mode |
| **Ecosystem** | Growing | Very mature (ML, SQL, Graph) |

### Part 20: Data Modeling

#### Normalization Forms

| Form | Rule | Example Violation |
| --- | --- | --- |
| **1NF** | Atomic values, no repeating groups | Column with comma-separated values |
| **2NF** | 1NF + no partial dependencies on composite key | Non-key column depends on part of composite key |
| **3NF** | 2NF + no transitive dependencies | Column A → B → C (C depends on A transitively) |
| **BCNF** | Every determinant is a candidate key | Non-candidate key determines another column |

#### Modeling Methodologies

| Methodology | Key Concepts | Best For |
| --- | --- | --- |
| **Kimball** | Star schema, conformed dimensions, bus matrix | Department-level analytics, bottom-up |
| **Inmon** | Enterprise data warehouse, 3NF, then data marts | Enterprise-wide, top-down |
| **Data Vault 2.0** | Hubs (business keys), Links (relationships), Satellites (attributes) | Agile, auditable, historical DW |
| **Activity Schema** | Single wide table with activity_id, entity, timestamp, features | Event-driven analytics, ML feature engineering |
| **One Big Table** | Fully denormalized single table | Simple analytics, small teams |

| Feature | Delta Lake | Apache Iceberg | Apache Hudi |

#### Snowflake Architecture

| Layer | Description |
| --- | --- |
| **Cloud Services** | Authentication, metadata, query optimization, security |

| Concept | Description |
| --- | --- |

| **Zero-copy clone** | Instant clone of table/schema/database (no data duplication) |
| **Snowpipe** | Continuous auto-ingestion from cloud storage |
| **Data Sharing** | Share live data across accounts without copying |
| **Snowpark** | DataFrame API for Python/Java/Scala in Snowflake |
