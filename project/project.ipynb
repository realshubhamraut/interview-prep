| 4 | What are the trade-offs of having two backends vs a monolith? | trade-off, monolith | • **Pros:** independent scaling, tech-appropriate frameworks, isolated failures, separate deployments<br>• **Cons:** network latency between services, data consistency complexity, operational overhead<br>• Inter-service communication adds ~2-5ms per HTTP call<br>• Mitigated by keeping services on same Docker network |

| 31 | Why does the system use 5 different databases? Is this over-engineering? | polyglot persistence | • **PostgreSQL:** relational data (users, classrooms, curriculum) — ACID transactions needed<br>• **Qdrant:** vector embeddings for semantic search — specialized vector index (HNSW)<br>• **Redis:** caching, sessions — sub-millisecond reads, TTL expiry<br>• **MongoDB:** proctoring frames — high-write, schema-flexible documents<br>• **Cassandra:** time-series analytics — append-only, high throughput<br>• Each chosen for its strength — using one DB for all would be worse at everything |
| 32 | What would break if you replaced all databases with just PostgreSQL? | trade-off, database | • **Vector search:** PostgreSQL pgvector exists but is 10-50x slower than Qdrant for ANN search<br>• **Caching:** PostgreSQL as cache would add ~5-10ms per read vs Redis's <1ms<br>• **Proctoring writes:** PostgreSQL row-level locking would bottleneck at 30fps frame writes<br>• **Time-series:** no native time-series partitioning, window queries would be slow<br>• Feasible but would require significant performance compromises |
| 33 | How does the system handle data consistency across services? | consistency, distributed | • **Eventual consistency** for async flows: Kafka events processed with at-least-once delivery<br>• **Request-scoped consistency** for sync flows: single database transaction per REST request<br>• **No distributed transactions:** services own their data, communicate via events<br>• **Idempotency keys** on Kafka consumers prevent duplicate processing<br>• Acceptable because most flows are user-facing with natural delays |

| 36 | How would you implement A/B testing for different AI agent versions? | A/B testing, experiment | • Use feature flags in environment config: `TUTOR_AGENT_VERSION=v2`<br>• Orchestrator routes based on user's experiment group (stored in user profile)<br>• Both agent versions deployed simultaneously as separate classes<br>• Metrics: track response quality, latency, user satisfaction per version<br>• Kafka events log which version served each request for analysis |
| 37 | Scenario: LLM API rate limits are hit. | 51 | What databases does EnsureStudy use and what role does each play? | polyglot, databases | • **PostgreSQL:** primary relational store — users, classrooms, curriculum, assessments (39 tables)<br>• **Qdrant:** vector database — document embeddings for RAG semantic search<br>• **Redis:** caching layer — ABCR results, session data, web content cache<br>• **MongoDB:** proctoring data — high-frequency frame captures, schema-flexible<br>• **Cassandra:** time-series analytics — student activity tracking, append-only writes |
| 52 | How many PostgreSQL tables exist and how are they grouped? | PostgreSQL, schema | • 39 tables across model files: `user.py` (9 models), `curriculum.py` (16 models), `classroom.py` (3 models), `meeting.py` (5 models), others (6)<br>• Groups: User/Auth → Classroom/Enrollment → Curriculum/Topics → Assessment/Questions → Meeting/Recording → Documents/Notes<br>• Uses SQLAlchemy ORM with Flask-SQLAlchemy extension |

| 55 | What is the Entity Relationship overview? | ER, relationships, diagram | • User ←(1:N)→ Enrollment ←(N:1)→ Classroom<br>• Classroom ←(1:N)→ Syllabus ←(1:N)→ Topic<br>• User ←(1:N)→ Curriculum ←(1:N)→ DailyGoal ←(N:1)→ Topic<br>• User ←(1:N)→ AssessmentAttempt ←(N:1)→ Assessment<br>• Classroom ←(1:N)→ Document ←(1:N)→ DocumentChunk (in Qdrant) |
| 56 | How does SQLAlchemy ORM handle database migrations? | migrations, Alembic | • Flask-Migrate wraps Alembic for migration management<br>• `flask db migrate`: auto-generates migration from model changes<br>• `flask db upgrade`: applies pending migrations<br>• Migration files stored in `migrations/versions/`<br>• Auto-detection compares current models to database state<br>• Production: migrations run as pre-deployment step |
| 57 | What database indexes are defined and why? | indexes, performance | • `users.email`: unique index for login lookup<br>• `enrollments(user_id, classroom_id)`: composite for membership checks<br>• `documents(classroom_id, status)`: filter documents by classroom and processing status<br>• `assessment_attempts(user_id, created_at)`: query user's recent attempts<br>• Missing: no full-text search index (used Qdrant instead)<br>• All foreign keys automatically indexed by PostgreSQL |
| 58 | How does Qdrant store and retrieve document embeddings? | Qdrant, vectors, HNSW | • Collection `documents`: stores 768-dimensional vectors (all-mpnet-base-v2)<br>• HNSW index: approximate nearest neighbor search with ~95% recall<br>• Each vector has payload: `{classroom_id, document_id, chunk_text, chunk_index, source_page}`<br>• Retrieval: embed query → search top-k similar vectors → filter by classroom_id → return chunks |
| 59 | What is the Qdrant collection schema? | Qdrant, schema, collection | • Collection name: `ensurestudy_documents`<br>• Vector size: 768 (matching all-mpnet-base-v2 output)<br>• Distance metric: Cosine similarity<br>• Payload indexes: `classroom_id` (keyword), `document_id` (keyword), `source_type` (keyword)<br>• Settings: `on_disk=True` for datasets > RAM, `hnsw_config(m=16, ef_construct=100)` |
| 60 | How does Qdrant filtering work during RAG queries? | Qdrant, filtering, RAG | • Filter format: `{"must": [{"key": "classroom_id", "match": {"value": "cls_123"}}]}`<br>• Applied BEFORE vector search — narrows search space for performance<br>• Supports: `must`, `should`, `must_not` conditions<br>• Can filter by multiple fields simultaneously<br>• Payload indexes make filtering efficient (not full scan) |
| 61 | Why was all-mpnet-base-v2 chosen as the embedding model? | embedding, model selection | • 768 dimensions — good balance of quality vs storage/compute<br>• Trained on 1B+ sentence pairs — strong semantic understanding<br>• Top performer on MTEB benchmark for semantic similarity at its size class<br>• Runs locally on CPU in ~50ms per sentence — no API dependency<br>• Trade-off: larger models (e5-large) are better but 3x slower<br>• Consistency: same model for ingestion AND retrieval (critical) |
| 62 | What is the embedding consistency issue mentioned in the docs? | embedding, consistency, bug | • If embedding model is changed, ALL existing vectors become incompatible<br>• New queries embedded with new model won't match old document vectors<br>• Solution: model version stored in collection metadata<br>• On model change: must re-embed entire corpus (expensive for large datasets)<br>• Mitigation: model name is a configuration constant, changes require explicit migration |
| 63 | How does Redis caching work in the system? | Redis, caching, TTL | • **ABCR cache** (5min TTL): caches query classification results to avoid repeated LLM calls<br>• **Session cache** (30min TTL): stores conversation history for tutor agent<br>• **Web content cache** (1hr TTL): caches scraped web pages from research agent<br>• **Response cache** (variable TTL): caches RAG retrieval results for identical queries<br>• Key format: `cache_type:user_id:hash(query)` |

| 111 | Show the LangGraph workflow definition for the Orchestrator. | LangGraph, workflow | • Graph: `classify_intent → select_agent → execute_agent → synthesize_response → END`<br>• Conditional edge after `classify_intent`: routes based on intent type<br>• Error edge: any node failure → `handle_error` → END with error message<br>• Compiled with `graph.compile()` for optimized execution<br>• StateGraph with OrchestratorState as state schema |
| 112 | What is a LangGraph state machine and why is it used for agents? | LangGraph, state machine | • LangGraph provides directed graph execution where nodes are functions and edges are transitions<br>• State (TypedDict) flows through nodes, each node can read/modify state<br>• Benefits: explicit control flow, easy debugging, built-in persistence, retry support<br>• vs LangChain chains: supports cycles, conditional branching, human-in-the-loop<br>• Every agent except ModerationAgent uses LangGraph |

| 115 | Describe the Agent Tools System. | tools, function calling | • Agents have access to tools: database queries, web search, embedding generation<br>• Tools defined as Python functions decorated with `@tool`<br>• LLM decides when to call tools based on query analysis<br>• Tool results injected into agent's context for answer generation<br>• Examples: `search_documents(query, classroom_id)`, `fetch_web_content(url)`, `get_student_performance(user_id)` |
| 116 | Explain the agent lifecycle and initialization. | lifecycle, init, singleton | • Agents initialized at FastAPI startup via `@app.on_event("startup")`<br>• Stored as singletons: one TutorAgent instance serves all requests<br>• LLM clients created once, connection pools shared across requests<br>• LangGraph compiled once at startup, executed per-request with fresh state<br>• Shutdown: graceful cleanup of LLM connections and cache |
| 117 | Why are agents singletons instead of per-request instances? | singleton, per-request, trade-off | • **Singleton pros:** avoid repeated LLM client initialization (~200ms), share compiled LangGraph<br>• **Singleton cons:** must be thread-safe, shared state risks<br>• **Per-request pros:** complete isolation, no shared state bugs<br>• **Decision:** singleton with per-request STATE (TypedDict is new each request)<br>• Agent instance is stateless — all state lives in LangGraph state dict |

**Key Concepts:** Retrieval-Augmented Generation · Embedding Strategy · all-mpnet-base-v2 · Qdrant · HNSW Index · Semantic Chunking · Document Ingestion · Retrieval Pipeline · Scoring & Ranking · Specialized RAG Variants

| 261 | What is RAG and why does EnsureStudy use it? | RAG, overview | • Retrieval-Augmented Generation: retrieve relevant documents first, then generate answer using them as context<br>• Without RAG: LLM answers from training data only (may be outdated/wrong for course-specific content)<br>• With RAG: LLM answers grounded in uploaded course materials (accurate, citable)<br>• Reduces hallucination by providing factual evidence in the prompt |
| 262 | Describe the end-to-end RAG architecture. | architecture, pipeline | • Offline: documents uploaded → chunked → embedded → indexed in Qdrant<br>• Online: query → embedded → Qdrant similarity search → top-k chunks retrieved → injected into LLM prompt → answer generated<br>• Two phases: ingestion (offline, batch) and retrieval (online, real-time)<br>• Components: EmbeddingService, QdrantService, ChunkingService, LLM client |
| 263 | Why was all-mpnet-base-v2 chosen as the embedding model? | embedding, model | • 768 dimensions — good balance of quality vs storage/compute<br>• Top performer on MTEB benchmark at its size class<br>• Trained on 1B+ sentence pairs for strong semantic understanding<br>• Runs locally on CPU (~50ms per sentence) — no external API dependency<br>• Trade-off: larger models (e5-large) are better but 3x slower |
| 264 | What is HNSW and how does Qdrant use it? | HNSW, vector index | • Hierarchical Navigable Small World — approximate nearest neighbor algorithm<br>• Creates layered graph of vectors: top layers sparse (fast navigation), bottom layers dense (precise)<br>• Parameters: `m=16` (connections per node), `ef_construct=100` (build quality)<br>• Search: ~95% recall at 10-20ms for 1M vectors<br>• Trade-off: higher m/ef = better recall but more memory and slower build |
| 265 | How does semantic chunking work in the document pipeline? | chunking, semantic | • Documents split into 512-token chunks with 50-token overlap<br>• Overlap prevents: losing context at chunk boundaries (sentence split mid-idea)<br>• Chunking strategy: split on paragraph boundaries when possible<br>• Fallback: if paragraph > 512 tokens, split on sentence boundaries<br>• Metadata per chunk: source document, page number, chunk index |
| 266 | Why 512 tokens with 50-token overlap? Why not different sizes? | chunk size, trade-off | • Too small (128): fragments context, retrieval returns incomplete ideas<br>• Too large (1024): retrieval returns too much irrelevant text with the relevant part<br>• 512: fits well in embedding model's context window (max 512 for mpnet-base)<br>• 50-token overlap: ~10% redundancy, ensures key sentences at boundaries aren't lost<br>• Tested internally: 512/50 gave best retrieval F1 score |
| 267 | What are the 7 stages of the document ingestion pipeline? | ingestion, stages | • Stage 1: Validation (file type, size, format)<br>• Stage 2: Preprocessing (text extraction, cleaning)<br>• Stage 3: Text extraction/OCR (for scanned documents)<br>• Stage 4: Semantic chunking (split into 512-token chunks)<br>• Stage 5: Embedding generation (all-mpnet-base-v2)<br>• Stage 6: Qdrant indexing (store vectors + payload)<br>• Stage 7: Completion notification (update DB status, notify user) |
| 268 | How does the retrieval pipeline score and rank results? | scoring, ranking | • Qdrant returns top-k with cosine similarity scores (0.0 to 1.0)<br>• Score threshold: chunks below 0.3 excluded (not relevant enough)<br>• Re-ranking: vector score (70%) + keyword overlap (20%) + recency (10%)<br>• Diversity: max 2 chunks from same document to avoid dominance<br>• Final: top 5 chunks selected for LLM context window |

| 564 | How does the LSTM temporal predictor work? | LSTM, temporal, predictor | • Input: sequence of 30 frame features (1-second window at 30fps)<br>• Output: cheat probability considering temporal patterns<br>• Why LSTM: catches patterns LightGBM misses (e.g., repeated glancing = suspicious, single glance = normal)<br>• Architecture: 2-layer LSTM → Dense → Sigmoid<br>• Complements LightGBM: static + temporal for robust detection |

| 644 | How is temperature configured per use case? | temperature, tuning | • Tutor generation: 0.3 (factual, accurate, not creative)<br>• Research synthesis: 0.7 (some creativity for combining sources)<br>• Question generation: 0.1 (deterministic, reproducible)<br>• ABCR classification: 0.0 (fully deterministic route selection)<br>• Moderation: 0.0 (safety decisions must be consistent)<br>• Follow-up suggestions: 0.8 (creative variety in suggestions) |
| 645 | How does token budget management work? | token, budget, management | • Total prompt budget: ~8000 tokens (Gemini generation prompt)<br>• System prompt: ~500 tokens (fixed)<br>• RAG context: ~2000 tokens (5 chunks × ~400 tokens each)<br>• Conversation history: ~4000 tokens (last N messages, truncated if needed)<br>• User query: ~200 tokens<br>• Instructions: ~300 tokens<br>• Budget tracking: tiktoken estimates, truncate largest section if over |

| 660 | How would you implement LLM output guardrails? | guardrails, safety, output | • Content filter: regex check for obviously harmful patterns in output<br>• Length guard: reject responses < 50 chars (too short) or > 5000 chars (too long)<br>• Format guard: verify expected structure (JSON parseable, markdown valid)<br>• Topic guard: verify response stays on-topic (embedding similarity with query)<br>• PII guard: regex scan for personal information leakage<br>• All guards: post-processing before response reaches student |

| 761 | How is the Singleton pattern used in the system? | Singleton, pattern | • Embedding model: single instance loaded at startup, shared across requests<br>• LLM clients: one GeminiClient, one GroqClient shared globally<br>• Agent instances: Orchestrator, TutorAgent, etc. — one per process<br>• Implementation: module-level instantiation or `@lru_cache` on factory function<br>• Purpose: avoid re-loading expensive resources (model weights, connections) |
| 762 | How is the Factory pattern used? | Factory, pattern | • Application Factory: `create_app()` in Flask for configurable app instances<br>• Agent Factory: Orchestrator selects and instantiates appropriate agent per query<br>• LLM Factory: `create_llm_client(provider)` returns Gemini or Groq client<br>• Benefits: loose coupling, easy testing with mock factories<br>• Configuration-driven: factory behavior changes based on environment |

| 764 | How is the Observer pattern used via Kafka? | Observer, Kafka, pattern | • Producer = Subject: publishes events (assessment completed, document uploaded)<br>• Consumer = Observer: subscribes to events, reacts accordingly<br>• Loose coupling: producer doesn't know who consumes events<br>• Multiple observers: single assessment event triggers LearningAgent + RevisionAgent + Analytics<br>• Decoupled: adding new observer requires zero changes to producer |
| 765 | How is the Decorator pattern used? | Decorator, pattern, Python | • `@token_required`: wraps route handler with JWT authentication<br>• `@role_required("teacher")`: adds role-based access control<br>• `@cache_result(ttl=300)`: wraps function with Redis caching<br>• `@validate_request(Schema)`: adds input validation before handler<br>• `@retry(max_retries=3)`: adds retry logic to external API calls<br>• Python decorators: clean, composable, reusable middleware |
| 766 | How is the State Machine pattern used with LangGraph? | State Machine, LangGraph | • Each agent: LangGraph `StateGraph` defines nodes and edges<br>• State: TypedDict passed between nodes, modified immutably<br>• Conditional edges: `if state["is_safe"]` → continue, else → reject<br>• Pattern: explicit states, explicit transitions, no implicit control flow<br>• Benefits: debuggable (inspect state at each node), testable (mock state transitions) |
