### Part 1: NoSQL Overview & CAP Theorem

#### SQL vs NoSQL

| Feature | SQL (Relational) | NoSQL |
| --- | --- | --- |
| **Data Model** | Tables with rows/columns | Documents, Key-Value, Wide-Column, Graph |
| **Schema** | Fixed schema (DDL) | Flexible / schema-less |
| **Scaling** | Vertical (scale up) | Horizontal (scale out) |
| **ACID** | Full ACID compliance | Tunable consistency (BASE) |
| **Query Language** | SQL (standardized) | Database-specific APIs |
| **Joins** | Native JOIN support | Usually no joins (denormalization) |
| **Best For** | Complex queries, transactions | High throughput, flexible data |

#### CAP Theorem

> **You can only guarantee 2 out of 3**: Consistency, Availability, Partition Tolerance

$$\text{Pick 2 of: } C \text{ (Consistency)} + A \text{ (Availability)} + P \text{ (Partition Tolerance)}$$

| Database | CAP | Trade-off |
| --- | --- | --- |
| **MongoDB** | CP | Consistent reads, may sacrifice availability during partition |
| **Cassandra** | AP | Always available, eventual consistency |
| **Redis** | CP | Single-master consistency |
| **DynamoDB** | AP | Eventual consistency (tunable) |

#### BASE vs ACID

| Property | ACID | BASE |
| --- | --- | --- |
| | **A**tomicity, **C**onsistency, **I**solation, **D**urability | **B**asically **A**vailable, **S**oft state, **E**ventual consistency |
| **Priority** | Data correctness | Availability & performance |
| **Use Case** | Banking, transactions | Social media, analytics |

#### NoSQL Database Types

| Type | Model | Example | Best For |
| --- | --- | --- | --- |
| **Document** | JSON-like documents | MongoDB, CouchDB | Content management, catalogs |
| **Key-Value** | Simple key → value pairs | Redis, DynamoDB | Caching, sessions |
| **Wide-Column** | Column families | Cassandra, HBase | Time-series, IoT |
| **Graph** | Nodes + edges | Neo4j, Amazon Neptune | Social networks, recommendations |

### Part 2: MongoDB — Document Database

#### What is MongoDB?

MongoDB is a **document-oriented NoSQL database** that stores data in flexible, JSON-like **BSON** (Binary JSON) documents.

#### Core Concepts

| SQL Term | MongoDB Term | Description |
| --- | --- | --- |
| **Database** | Database | Container for collections |
| **Table** | Collection | Group of documents (no fixed schema) |
| **Row** | Document | A single BSON record (JSON-like) |
| **Column** | Field | A key-value pair within a document |
| **Primary Key** | `_id` | Auto-generated `ObjectId` (12 bytes) |
| **JOIN** | `$lookup` | Aggregation stage for joining |
| **INDEX** | Index | B-tree indexes (same concept) |

#### BSON Data Types

| Type | Example | Notes |
| --- | --- | --- |
| **String** | `"hello"` | UTF-8 |
| **Int32 / Int64** | `42` | Number types |
| **Double** | `3.14` | 64-bit float |
| **Boolean** | `true` | true/false |
| **ObjectId** | `ObjectId("...")` | 12-byte unique ID |
| **Date** | `ISODate("2024-01-01")` | UTC datetime |
| **Array** | `[1, 2, 3]` | Ordered list |
| **Embedded Doc** | `{name: {first: "John"}}` | Nested document |
| **Null** | `null` | Absent value |

#### Document Design — Embedding vs Referencing

| Strategy | When to Use | Example |
| --- | --- | --- |
| **Embedding** (denormalize) | 1:1 or 1:few relationships, data accessed together | User + address in same doc |
| **Referencing** (normalize) | 1:many or many:many, data updated independently | User doc + separate Orders collection |

**Rule of thumb**: If data is always read together → embed. If data is shared/updated independently → reference.

### Part 3: MongoDB — CRUD Operations

#### CRUD Command Reference

| Operation | Command | SQL Equivalent |
| --- | --- | --- |
| **Create** | `insertOne()`, `insertMany()` | `INSERT INTO` |
| **Read** | `find()`, `findOne()` | `SELECT` |

#### Query Operators

| Category | Operator | Meaning | Example |
| --- | --- | --- | --- |
| **Comparison** | `$eq` | Equal | `{age: {$eq: 25}}` |
| | `$gt`, `$gte` | Greater than (or equal) | `{age: {$gt: 18}}` |
| | `$lt`, `$lte` | Less than (or equal) | `{price: {$lt: 100}}` |
| | `$ne` | Not equal | `{status: {$ne: "deleted"}}` |
| | `$in` | In array | `{color: {$in: ["red","blue"]}}` |
| **Logical** | `$and` | AND | `{$and: [{a: 1}, {b: 2}]}` |
| | `$or` | OR | `{$or: [{a: 1}, {b: 2}]}` |
| | `$not` | NOT | `{age: {$not: {$gt: 25}}}` |
| **Element** | `$exists` | Field exists | `{email: {$exists: true}}` |
| | `$type` | Field type | `{age: {$type: "int"}}` |
| **Array** | `$all` | Contains all | `{tags: {$all: ["a","b"]}}` |
| | `$size` | Array length | `{tags: {$size: 3}}` |
| | `$elemMatch` | Element match | `{scores: {$elemMatch: {$gt: 90}}}` |


````python
# ──────────────────────────────────────────────
# MongoDB — CRUD Operations (using pymongo)
# ──────────────────────────────────────────────
try:
    from pymongo import MongoClient
    
    # Connect (default localhost:27017)
    client = MongoClient('mongodb://localhost:27017/', serverSelectionTimeoutMS=2000)
    client.server_info()  # Test connection
    
    db = client['interview_prep']
    collection = db['users']
    
    # ── CREATE ──
    collection.insert_one({
        'name': 'Alice', 'age': 30, 'email': 'alice@example.com',
        'skills': ['Python', 'MongoDB'], 'address': {'city': 'NYC', 'state': 'NY'}
    })
    collection.insert_many([
        {'name': 'Bob', 'age': 25, 'skills': ['Java', 'Redis']},
        {'name': 'Charlie', 'age': 35, 'skills': ['Python', 'Cassandra']},
    ])
    
    # ── READ ──
    print("All users:", list(collection.find({}, {'_id': 0, 'name': 1, 'age': 1})))
    print("Age > 28:", list(collection.find({'age': {'$gt': 28}}, {'_id': 0, 'name': 1})))
    
    # ── UPDATE ──
    collection.update_one({'name': 'Alice'}, {'$set': {'age': 31}, '$push': {'skills': 'Redis'}})
    
    # ── DELETE ──
    collection.delete_many({'age': {'$lt': 26}})
    
    print("After ops:", list(collection.find({}, {'_id': 0})))
    collection.drop()
    client.close()
    
except Exception as e:
    print(f"MongoDB not available: {e}")
    print("Demo output (simulated):")
    print("All users: [{'name': 'Alice', 'age': 30}, {'name': 'Bob', 'age': 25}, {'name': 'Charlie', 'age': 35}]")
    print("Age > 28: [{'name': 'Alice'}, {'name': 'Charlie'}]")
````

### Part 4: MongoDB — Aggregation Pipeline

The aggregation pipeline processes documents through a sequence of **stages** (like Unix pipes).

#### Key Aggregation Stages

| Stage | Purpose | SQL Equivalent |
| --- | --- | --- |
| `$match` | Filter documents | `WHERE` |
| `$group` | Group + aggregate | `GROUP BY` |
| `$project` | Reshape / select fields | `SELECT` |
| `$sort` | Order results | `ORDER BY` |
| `$limit` / `$skip` | Pagination | `LIMIT` / `OFFSET` |
| `$unwind` | Flatten arrays | Unnest |
| `$lookup` | Join collections | `LEFT JOIN` |
| `$addFields` | Add computed fields | `SELECT col, expr AS alias` |
| `$count` | Count documents | `COUNT(*)` |
| `$bucket` | Range-based grouping | `CASE WHEN` |
| `$facet` | Multiple pipelines | Sub-queries |

#### Aggregation Expressions

| Expression | Meaning | Example |
| --- | --- | --- |
| `$sum` | Sum values | `{$sum: "$price"}` |
| `$avg` | Average | `{$avg: "$rating"}` |
| `$min` / `$max` | Min / Max | `{$min: "$created_at"}` |
| `$first` / `$last` | First / Last in group | `{$first: "$name"}` |
| `$push` | Collect into array | `{$push: "$item"}` |
| `$addToSet` | Unique values array | `{$addToSet: "$category"}` |


````python
# ──────────────────────────────────────────────
# MongoDB Aggregation Pipeline — Examples
# ──────────────────────────────────────────────
print("MongoDB Aggregation Pipeline Examples (simulated)")
print()

# Example 1: Sales by category
pipeline_1 = '''
db.orders.aggregate([
    { $match: { status: "completed" } },
    { $group: {
        _id: "$category",
        total_sales: { $sum: "$amount" },
        avg_order: { $avg: "$amount" },
        count: { $sum: 1 }
    }},
    { $sort: { total_sales: -1 } },
    { $limit: 5 }
])
'''
print("Example 1 — Sales by Category:")
print(pipeline_1)

# Example 2: $lookup (JOIN)
pipeline_2 = '''
db.orders.aggregate([
    { $lookup: {
        from: "users",
        localField: "user_id",
        foreignField: "_id",
        as: "user_info"
    }},
    { $unwind: "$user_info" },
    { $project: {
        order_id: 1,
        amount: 1,
        "user_info.name": 1,
        "user_info.email": 1
    }}
])
'''
print("Example 2 — $lookup JOIN:")
print(pipeline_2)
````

### Part 5: MongoDB — Indexing & Performance

#### Index Types

| Index Type | Syntax | Use Case |
| --- | --- | --- |
| **Single Field** | `createIndex({field: 1})` | Simple queries on one field |
| **Compound** | `createIndex({a: 1, b: -1})` | Queries on multiple fields |
| **Multikey** | Auto on array fields | Querying array elements |
| **Text** | `createIndex({field: "text"})` | Full-text search |
| **Geospatial** | `createIndex({loc: "2dsphere"})` | Location queries |
| **Hashed** | `createIndex({field: "hashed"})` | Hash-based sharding |
| **TTL** | `createIndex({date: 1}, {expireAfterSeconds: 3600})` | Auto-expire documents |
| **Unique** | `createIndex({email: 1}, {unique: true})` | Prevent duplicates |
| **Partial** | `createIndex({...}, {partialFilterExpression: {...}})` | Index subset of docs |

#### Index Optimization — `explain()`

```
db.users.find({age: {$gt: 25}}).explain("executionStats")

Key metrics to check:
- totalDocsExamined vs nReturned → ratio close to 1 = good
- stage: "COLLSCAN" → BAD (full scan, needs index)
- stage: "IXSCAN" → GOOD (using index)
```

#### ESR Rule (Equality, Sort, Range)

Build compound indexes in this order for optimal performance:

$$\text{Index order: } \underbrace{\text{Equality}}_{ = } \rightarrow \underbrace{\text{Sort}}_{\text{ORDER BY}} \rightarrow \underbrace{\text{Range}}_{>, <, \geq}$$

### Part 6: MongoDB — Sharding & Replication

#### Replication (Replica Set)

```
Primary ──write──→ Secondary 1
   │                Secondary 2
   └── Elections if primary fails
```

| Concept | Description |
| --- | --- |
| **Primary** | Receives all writes |
| **Secondary** | Replicates from primary (async), can serve reads |
| **Arbiter** | Votes in elections but holds no data |
| **Failover** | Automatic via election (majority vote) |
| **Read Preference** | `primary`, `secondary`, `nearest` |
| **Write Concern** | `w: 1` (primary ack), `w: "majority"` (majority ack) |

#### Sharding (Horizontal Partitioning)

```
Client → mongos (router) → Config Servers (metadata)
                         → Shard 1 (replica set)
                         → Shard 2 (replica set)
                         → Shard 3 (replica set)
```

| Shard Key Strategy | Pros | Cons |
| --- | --- | --- |
| **Ranged** | Good for range queries | Hot spots if sequential |
| **Hashed** | Even distribution | No range queries |
| **Compound** | Balanced | More complex to design |

**Choosing a good shard key:**
- High cardinality (many unique values)
- Even write distribution
- Commonly used in queries
- Avoid monotonically increasing keys (e.g., timestamps alone)

### Part 8: Cassandra — Wide-Column Store

#### What is Apache Cassandra?

Cassandra is a **distributed, wide-column NoSQL database** designed for **high availability** and **linear scalability** with no single point of failure.

#### Key Characteristics

| Feature | Detail |
| --- | --- |
| **Architecture** | Peer-to-peer (no master/slave) |
| **CAP** | AP — Available + Partition Tolerant |
| **Consistency** | Tunable (ONE, QUORUM, ALL) |
| **Scaling** | Linear horizontal scaling |
| **Query Language** | CQL (Cassandra Query Language) |
| **Data Model** | Partitioned row store (wide columns) |
| **Replication** | Configurable replication factor |
| **Written in** | Java |

#### SQL vs CQL Comparison

| SQL Concept | CQL Equivalent | Key Difference |
| --- | --- | --- |
| Database | Keyspace | Includes replication config |
| Table | Table | Must have partition key |
| Row | Row (partition) | Rows can be very wide |
| Column | Column | Static or clustering |
| SELECT * | SELECT * | No JOINs, no subqueries |
| WHERE (any col) | WHERE (partition key required) | Must include partition key |
| GROUP BY | GROUP BY (limited) | Only on clustering columns |

#### Data Model Hierarchy

```
Cluster → Keyspace → Table → Partition → Row → Column

Partition Key — determines which node stores the data
Clustering Key — determines sort order within a partition
```

$$\text{Token} = \text{hash}(\text{partition\_key}) \rightarrow \text{determines node placement}$$

### Part 9: Cassandra — Data Modeling & CQL

#### Primary Key Design (CRITICAL)

```
PRIMARY KEY ((partition_key), clustering_col1, clustering_col2)
              ↑ determines node     ↑ determines sort order
```

| Key Type | Purpose | Example |
| --- | --- | --- |
| **Partition Key** | Determines data distribution | `user_id` |
| **Composite Partition** | Multiple cols for partition | `(country, city)` |
| **Clustering Column** | Sort within partition | `created_at DESC` |
| **Static Column** | One value per partition | `user_name STATIC` |

#### CQL Data Types

| Type | Description | Example |
| --- | --- | --- |
| `text` / `varchar` | UTF-8 string | `'hello'` |
| `int` / `bigint` | Integer (32/64 bit) | `42` |
| `float` / `double` | Floating point | `3.14` |
| `boolean` | True/False | `true` |
| `timestamp` | Date + time | `'2024-01-01T00:00:00'` |
| `uuid` / `timeuuid` | Unique identifiers | `uuid()` |
| `blob` | Binary data | Binary |
| `list<T>` | Ordered collection | `['a', 'b']` |
| `set<T>` | Unique collection | `{'a', 'b'}` |
| `map<K,V>` | Key-value pairs | `{'key': 'val'}` |
| `tuple<T1,T2>` | Fixed-length tuple | `(1, 'hello')` |
| `counter` | Distributed counter | `UPDATE ... SET count = count + 1` |

#### CQL Commands Reference

| Operation | CQL | Notes |
| --- | --- | --- |
| **Create Keyspace** | `CREATE KEYSPACE ks WITH replication = {...}` | Set replication strategy |
| **Create Table** | `CREATE TABLE t (id int PRIMARY KEY, ...)` | Must define primary key |
| **Insert** | `INSERT INTO t (id, name) VALUES (1, 'Alice')` | Upsert by default! |
| **Select** | `SELECT * FROM t WHERE id = 1` | Must include partition key |
| **Update** | `UPDATE t SET name = 'Bob' WHERE id = 1` | Same as INSERT (upsert) |
| **Delete** | `DELETE FROM t WHERE id = 1` | Writes a tombstone |
| **TTL** | `INSERT ... USING TTL 3600` | Auto-expire in 1 hour |
| **Batch** | `BEGIN BATCH ... APPLY BATCH` | Atomic within partition |

### Part 10: Cassandra — Architecture & Consistency

#### Ring Architecture

```
      Node A                   
    (Token: 0-25)         
   ↗             ↘             
Node D            Node B       
(Token: 76-100)   (Token: 26-50)
   ↖             ↙             
      Node C                   
    (Token: 51-75)        
```

Each node owns a **token range**. Data is placed on nodes based on the hash of the partition key.

#### Consistency Levels

| Level | Reads | Writes | Use Case |
| --- | --- | --- | --- |
| **ONE** | 1 replica | 1 replica | Fastest, least consistent |
| **QUORUM** | majority | majority | Balance speed + consistency |
| **LOCAL_QUORUM** | majority in local DC | majority in local DC | Multi-DC deployments |
| **ALL** | all replicas | all replicas | Strongest, slowest |
| **EACH_QUORUM** | quorum per DC | quorum per DC | Global consistency |

#### Strong Consistency Formula

$$R + W > N \implies \text{Strong Consistency}$$

Where: $R$ = read consistency, $W$ = write consistency, $N$ = replication factor.

Example: $N=3$, `QUORUM` = 2 → $R(2) + W(2) = 4 > 3$ ✅ Strong consistency

#### Key Architecture Concepts

| Concept | Description |
| --- | --- |
| **Gossip Protocol** | Nodes share state info every second |
| **Snitch** | Determines node locations (data center, rack) |
| **Partitioner** | Murmur3 hash function for token assignment |
| **Compaction** | Merges SSTables (Sized-Tiered, Leveled, Time-Window) |
| **Memtable** | In-memory write buffer → flushed to SSTable |
| **SSTable** | Sorted String Table — immutable on-disk files |
| **Commit Log** | Write-ahead log for durability |
| **Bloom Filter** | Probabilistic check: "Is this key in this SSTable?" |
| **Tombstone** | Marker for deleted data (cleaned up during compaction) |
| **Repair** | Anti-entropy process to sync replicas |

#### Write Path

$$\text{Client} \xrightarrow{\text{write}} \text{Commit Log} \rightarrow \text{Memtable} \xrightarrow{\text{flush}} \text{SSTable}$$

#### Read Path

$$\text{Client} \xrightarrow{\text{read}} \text{Bloom Filter} \rightarrow \text{Partition Index} \rightarrow \text{SSTable / Memtable}$$

### Part 14: Redis — Caching Patterns & Persistence

#### Caching Strategies

| Pattern | How It Works | Pros | Cons |
| --- | --- | --- | --- |
| **Cache-Aside (Lazy)** | App checks cache → miss → read DB → write cache | Simple, app controls | Cold start, stale data |
| **Write-Through** | App writes cache + DB simultaneously | Always consistent | Write latency |
| **Write-Behind (Back)** | App writes cache, async write to DB | Fast writes | Data loss risk |
| **Read-Through** | Cache library reads DB on miss | Simple for app | Libraries needed |
| **Refresh-Ahead** | Pre-refresh before TTL expires | No cold reads | Complexity, wasted refreshes |

#### Cache-Aside Pattern (Most Common)

```python
def get_user(user_id):
    # 1. Check cache
    cached = redis.get(f"user:{user_id}")
    if cached:
        return json.loads(cached)   # Cache HIT
    
    # 2. Cache MISS → read from database
    user = db.query("SELECT * FROM users WHERE id = ?", user_id)
    
    # 3. Store in cache with TTL
    redis.setex(f"user:{user_id}", 3600, json.dumps(user))
    return user
```

#### Eviction Policies

| Policy | Behavior | Use Case |
| --- | --- | --- |
| `noeviction` | Return error when full | Critical data |
| `allkeys-lru` | Evict least recently used | General caching (recommended) |
| `allkeys-lfu` | Evict least frequently used | Frequency-based access |
| `volatile-lru` | LRU among keys with TTL | Mixed persistent + cache |
| `volatile-ttl` | Evict nearest TTL | Time-sensitive data |
| `allkeys-random` | Random eviction | When all keys equal priority |

#### Persistence Options

| Option | Mechanism | Pros | Cons |
| --- | --- | --- | --- |
| **RDB** (Snapshot) | Point-in-time snapshots (`BGSAVE`) | Compact, fast recovery | Data loss between snapshots |
| **AOF** (Append-Only) | Log every write operation | Minimal data loss | Larger files, slower recovery |
| **RDB + AOF** | Both combined | Best durability | More disk usage |
| **None** | Pure in-memory | Fastest | All data lost on restart |

#### Cache Invalidation (The Hard Problem!)

```
"There are only two hard things in CS: cache invalidation and naming things."
    — Phil Karlton
```

| Strategy | When to Use |
| --- | --- |
| **TTL-based** | Set expiration, accept brief staleness |
| **Event-driven** | Invalidate on write/update events |
| **Version-based** | Key includes version: `user:1:v3` |
| **Pub/Sub** | Broadcast invalidation to all app instances |

### Part 17: NoSQL Comparison — When to Use What

#### Head-to-Head Comparison

| Feature | MongoDB | Cassandra | Redis |
| --- | --- | --- | --- |
| **Type** | Document | Wide-Column | Key-Value / Data Structures |
| **CAP** | CP | AP | CP |
| **Data Model** | JSON/BSON documents | Partitioned rows | Keys → data structures |
| **Query Language** | MQL + Aggregation | CQL | Commands |
| **Scaling** | Horizontal (sharding) | Linear horizontal | Cluster (hash slots) |
| **Consistency** | Strong (configurable) | Tunable | Strong (single master) |
| **Speed** | ms-level | ms-level | Sub-ms (in-memory) |
| **Persistence** | Disk-based | Disk-based | In-memory + optional persist |
| **Best For** | General purpose, flexible queries | Write-heavy, time-series | Caching, real-time, sessions |
| **Joins** | $lookup (aggregation) | No joins | No joins |
| **Schema** | Flexible | Column-family defined | Schema-less |
| **Max Doc/Row Size** | 16 MB | 2 GB (partition) | 512 MB (value) |
