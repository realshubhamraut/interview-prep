### SQL Terminology & Key Concepts

Understanding these foundational terms is essential before writing any SQL. A **relational database** stores data in tables (relations) with rows (tuples) and columns (attributes). Tables are linked through keys.

---

#### Core Database Terms

| Term | Meaning | Example |
|------|---------|----------|
| **Database** | Organized collection of structured data | `CREATE DATABASE company;` |
| **Schema** | Logical namespace grouping tables/views | `CREATE SCHEMA hr;` — PG supports multiple schemas |
| **Table (Relation)** | Collection of rows and columns | `CREATE TABLE employees (...);` |
| **Row (Record/Tuple)** | Single entry in a table | One employee's complete data |
| **Column (Field/Attribute)** | A specific property with a data type | `name VARCHAR(100)` |
| **Primary Key (PK)** | Unique + NOT NULL identifier for each row | `id SERIAL PRIMARY KEY` |
| **Foreign Key (FK)** | Column referencing PK of another table | `dept_id INT REFERENCES departments(id)` |
| **Candidate Key** | Column(s) that could be PK (unique + not null) | email, phone if both unique |
| **Composite Key** | PK made of multiple columns | `PRIMARY KEY (student_id, course_id)` |
| **Index** | Data structure for faster SELECT queries | `CREATE INDEX idx_name ON emp(name);` |
| **View** | Virtual table from a saved query | `CREATE VIEW active_emp AS SELECT ...` |
| **Materialized View** | Cached view stored on disk (PG) | Needs `REFRESH` to update |
| **Stored Procedure** | Reusable block of SQL logic | `CREATE PROCEDURE process_order(...)` |
| **Function** | Returns a value; usable inside SELECT | `CREATE FUNCTION get_tax(...) RETURNS ...` |
| **Trigger** | Auto-executes on INSERT/UPDATE/DELETE | Audit logging, timestamp updates |
| **Transaction** | Group of operations — all succeed or all fail | `BEGIN; ... COMMIT;` or `ROLLBACK;` |
| **Sequence** | Auto-incrementing number generator (PG) | `CREATE SEQUENCE emp_id_seq;` |
| **Constraint** | Rule enforced on column data | `NOT NULL`, `UNIQUE`, `CHECK`, `FK` |
| **Normalization** | Organizing data to reduce redundancy | 1NF → 2NF → 3NF → BCNF |
| **Denormalization** | Intentionally adding redundancy for read speed | Used in reporting/OLAP systems |
| **ACID** | Atomicity, Consistency, Isolation, Durability | Properties every transaction must satisfy |
| **MVCC** | Multi-Version Concurrency Control | How PostgreSQL handles concurrent reads/writes |

---

#### SQL Command Categories

SQL commands are grouped into 5 categories based on their purpose:

| Category | Full Name | Commands | Purpose |
|----------|-----------|----------|----------|
| **DDL** | Data Definition Language | `CREATE, ALTER, DROP, TRUNCATE, RENAME` | Define/modify database structure |
| **DML** | Data Manipulation Language | `INSERT, UPDATE, DELETE, MERGE/UPSERT` | Add/modify/remove data |
| **DQL** | Data Query Language | `SELECT` | Retrieve/query data |
| **DCL** | Data Control Language | `GRANT, REVOKE` | Manage user permissions |
| **TCL** | Transaction Control Language | `BEGIN, COMMIT, ROLLBACK, SAVEPOINT` | Manage transactions |

### Data Types Reference

Choosing the right data type affects storage, performance, and data integrity. Here's a comparison of MySQL and PostgreSQL types:

---

#### Numeric Types

| Type | MySQL | PostgreSQL | Range / Notes |
|------|-------|------------|---------------|
| Small integer | `SMALLINT` | `SMALLINT` | -32,768 to 32,767 |
| Integer | `INT` | `INTEGER` | -2.1B to 2.1B |
| Big integer | `BIGINT` | `BIGINT` | Very large numbers |
| Auto-increment | `INT AUTO_INCREMENT` | `SERIAL` / `BIGSERIAL` | PG: `SERIAL` = auto-increment |
| Exact decimal | `DECIMAL(p,s)` | `NUMERIC(p,s)` | For money: `NUMERIC(10,2)` |
| Floating point | `FLOAT`, `DOUBLE` | `REAL`, `DOUBLE PRECISION` | Approximate — avoid for money |

#### String Types

| Type | MySQL | PostgreSQL | Notes |
|------|-------|------------|-------|
| Fixed-length | `CHAR(n)` | `CHAR(n)` | Padded with spaces |
| Variable-length | `VARCHAR(n)` | `VARCHAR(n)` | Most common for strings |
| Unlimited text | `TEXT` | `TEXT` | No length limit |
| Enum | `ENUM('a','b','c')` | `CREATE TYPE ... AS ENUM(...)` | PG uses custom types |

#### Date/Time Types

| Type | MySQL | PostgreSQL | Notes |
|------|-------|------------|-------|
| Date only | `DATE` | `DATE` | `YYYY-MM-DD` |
| Time only | `TIME` | `TIME` | `HH:MI:SS` |
| Date + time | `DATETIME` | `TIMESTAMP` | Without timezone |
| With timezone | `TIMESTAMP` | `TIMESTAMPTZ` | **Always use this for timestamps** |
| Duration | — | `INTERVAL` | `'3 days'`, `'2 hours'` |

#### Special Types (PostgreSQL)

| Type | Purpose | Example |
|------|---------|----------|
| `BOOLEAN` | True/False/NULL | `is_active BOOLEAN DEFAULT TRUE` |
| `UUID` | Universally unique ID | `gen_random_uuid()` (PG 13+) |
| `JSONB` | Binary JSON (indexable) | `metadata JSONB DEFAULT '{}'` |
| `INTEGER[]` | Array of integers | `scores INTEGER[]` → `'{90,85,95}'` |
| `TEXT[]` | Array of strings | `tags TEXT[]` |
| `INET` | IP address | `ip_addr INET` |
| `DATERANGE` | Date range | `[2024-01-01, 2024-12-31)` |
| `BYTEA` | Binary data | For files, images |
| `TSVECTOR` | Full-text search | Pre-processed searchable text |

### Built-in Functions — Quick Reference

SQL provides hundreds of built-in functions. These are the most important ones grouped by category.

---

#### Aggregate Functions

Aggregate functions compute a **single result** from a set of rows. They ignore NULLs (except `COUNT(*)`).

| Function | What it does | Example | Notes |
|----------|-------------|---------|-------|
| `COUNT(*)` | Count all rows | `SELECT COUNT(*) FROM emp;` | Includes NULLs |
| `COUNT(col)` | Count non-NULL values | `SELECT COUNT(email) FROM emp;` | Excludes NULLs |
| `COUNT(DISTINCT col)` | Count unique values | `SELECT COUNT(DISTINCT dept) FROM emp;` | |
| `SUM(col)` | Sum of values | `SELECT SUM(salary) FROM emp;` | Ignores NULLs |
| `AVG(col)` | Average value | `SELECT AVG(salary) FROM emp;` | Ignores NULLs |
| `MIN(col)` | Minimum value | `SELECT MIN(hire_date) FROM emp;` | Works on dates/strings |
| `MAX(col)` | Maximum value | `SELECT MAX(salary) FROM emp;` | Works on dates/strings |
| `STRING_AGG(col, sep)` | Concatenate strings (PG) | `STRING_AGG(name, ', ')` | MySQL: `GROUP_CONCAT` |
| `ARRAY_AGG(col)` | Collect into array (PG) | `ARRAY_AGG(name ORDER BY name)` | |

---

#### String Functions

| Function | What it does | Example | Output |
|----------|-------------|---------|--------|
| `LENGTH(s)` / `CHAR_LENGTH(s)` | String length | `LENGTH('hello')` | `5` |
| `UPPER(s)` / `LOWER(s)` | Change case | `UPPER('hello')` | `HELLO` |
| `INITCAP(s)` | Capitalize words (PG) | `INITCAP('hello world')` | `Hello World` |
| `TRIM(s)` | Remove whitespace | `TRIM('  hi  ')` | `hi` |
| `SUBSTRING(s, pos, len)` | Extract substring | `SUBSTRING('hello', 2, 3)` | `ell` |
| `LEFT(s,n)` / `RIGHT(s,n)` | First/last n chars | `LEFT('hello', 3)` | `hel` |
| `REPLACE(s, old, new)` | Replace occurrences | `REPLACE('hello','l','r')` | `herro` |
| `CONCAT(a, b)` or `\|\|` | Join strings | `'hi' \|\| ' ' \|\| 'there'` | `hi there` |
| `POSITION(sub IN s)` | Find position | `POSITION('ll' IN 'hello')` | `3` |
| `LPAD(s, len, fill)` | Left-pad | `LPAD('42', 5, '0')` | `00042` |
| `SPLIT_PART(s, sep, n)` | Split & get nth (PG) | `SPLIT_PART('a,b,c', ',', 2)` | `b` |
| `REVERSE(s)` | Reverse string | `REVERSE('hello')` | `olleh` |
| `REGEXP_REPLACE(s,pat,rep)` | Regex replace | `REGEXP_REPLACE('a1b2','[0-9]','','g')` | `ab` |

---

#### Numeric Functions

| Function | What it does | Example | Output |
|----------|-------------|---------|--------|
| `ROUND(n, d)` | Round to d decimals | `ROUND(3.14159, 2)` | `3.14` |
| `CEIL(n)` | Round up | `CEIL(4.2)` | `5` |
| `FLOOR(n)` | Round down | `FLOOR(4.8)` | `4` |
| `TRUNC(n, d)` | Truncate (PG) | `TRUNC(3.789, 1)` | `3.7` |
| `ABS(n)` | Absolute value | `ABS(-5)` | `5` |
| `MOD(a, b)` | Remainder | `MOD(17, 5)` | `2` |
| `POWER(a, b)` | Exponent | `POWER(2, 10)` | `1024` |
| `SQRT(n)` | Square root | `SQRT(16)` | `4` |
| `RANDOM()` | Random 0..1 (PG) | `SELECT RANDOM()` | `0.73...` |
| `GREATEST(a,b,c)` | Max of values | `GREATEST(10, 20, 5)` | `20` |
| `LEAST(a,b,c)` | Min of values | `LEAST(10, 20, 5)` | `5` |

---

#### Date/Time Functions

| Function | What it does | Example |
|----------|-------------|----------|
| `NOW()` / `CURRENT_TIMESTAMP` | Current date+time | `SELECT NOW()` |
| `CURRENT_DATE` | Today's date | `SELECT CURRENT_DATE` |
| `EXTRACT(part FROM dt)` | Extract component | `EXTRACT(YEAR FROM NOW())` → `2024` |
| `DATE_PART('part', dt)` | Same (PG) | `DATE_PART('month', NOW())` → `1` |
| `AGE(d1, d2)` | Interval between (PG) | `AGE(NOW(), '1999-12-31')` |
| `DATE_TRUNC('part', dt)` | Truncate precision (PG) | `DATE_TRUNC('month', NOW())` → 1st of month |
| `TO_CHAR(dt, fmt)` | Format as string (PG) | `TO_CHAR(NOW(), 'DD/MM/YYYY')` |
| `TO_DATE(s, fmt)` | Parse to date (PG) | `TO_DATE('15/01/2024', 'DD/MM/YYYY')` |
| `INTERVAL` | Add/subtract time | `NOW() + INTERVAL '7 days'` |
| `GENERATE_SERIES(...)` | Generate sequence (PG) | `GENERATE_SERIES(1, 10)` |
| `DATEDIFF(d1, d2)` | Days between (MySQL) | `DATEDIFF(CURDATE(), hire_date)` |
| `DATE_ADD(d, INTERVAL)` | Add time (MySQL) | `DATE_ADD(NOW(), INTERVAL 6 MONTH)` |

---

#### Conditional & Conversion Functions

| Function | What it does | Example |
|----------|-------------|----------|
| `CASE WHEN ... THEN ... END` | Conditional logic | `CASE WHEN sal>50000 THEN 'High' ELSE 'Low' END` |
| `COALESCE(a, b, c)` | First non-NULL | `COALESCE(phone, email, 'N/A')` |
| `NULLIF(a, b)` | NULL if a = b | `NULLIF(score, 0)` — prevents div by zero |
| `CAST(x AS type)` | Type conversion | `CAST('42' AS INTEGER)` |
| `::type` | PG cast shorthand | `'42'::INTEGER`, `NOW()::DATE` |
| `IFNULL(a, b)` | MySQL: if a is NULL, use b | `IFNULL(phone, 'No phone')` |
| `IF(cond, true, false)` | MySQL inline if | `IF(salary>50000, 'High', 'Low')` |

### Interview Knowledge Tables

---

#### Query Execution Order

SQL does **NOT** execute in the order you write it. Understanding this is critical for debugging.

| Step | Clause | What happens | Why it matters |
|------|--------|-------------|----------------|
| 1 | `FROM / JOIN` | Tables are loaded and joined | Tables must exist |
| 2 | `WHERE` | Rows filtered (before grouping) | Can't use aliases or aggregates here |
| 3 | `GROUP BY` | Rows grouped | Non-aggregated columns must be in GROUP BY |
| 4 | `HAVING` | Groups filtered (after grouping) | Can use aggregates like `HAVING COUNT(*) > 5` |
| 5 | `SELECT` | Columns/expressions computed | Aliases created here |
| 6 | `DISTINCT` | Duplicate rows removed | Applied after SELECT |
| 7 | `ORDER BY` | Results sorted | Can use aliases (SELECT ran first) |
| 8 | `LIMIT / OFFSET` | Results paginated | Last step |

---

#### JOINs Comparison

| JOIN Type | Returns | NULL behavior | Use case |
|-----------|---------|---------------|----------|
| `INNER JOIN` | Only matching rows from both | No NULLs from join | Default join; most common |
| `LEFT JOIN` | All left + matching right | Right side NULLs if no match | Find all items, even unmatched |
| `RIGHT JOIN` | All right + matching left | Left side NULLs if no match | Rarely used (rewrite as LEFT) |
| `FULL OUTER JOIN` | All from both sides | Either side can be NULL | Find all unmatched from both |
| `CROSS JOIN` | Cartesian product | No NULLs; n × m rows | Generate combinations |
| `SELF JOIN` | Table with itself | Depends on join type | Hierarchies, row comparisons |
| `LATERAL JOIN` | Subquery references outer | PG only | Correlated subquery in FROM |

**Key pattern:** `LEFT JOIN + WHERE right.id IS NULL` = **anti-join** (rows with NO match)

---

#### Normalization

Normalization reduces data redundancy by organizing tables and relationships.

| Form | Rule | Violation Example | Fix |
|------|------|-------------------|-----|
| **1NF** | Atomic values, no repeating groups | `skills='Python,Java'` | Separate rows or table |
| **2NF** | 1NF + no partial dependency | Non-key depends on PART of composite PK | Move to own table |
| **3NF** | 2NF + no transitive dependency | `zip → city` (city depends on zip, not PK) | Separate address table |
| **BCNF** | Every determinant is a candidate key | Stricter 3NF | Decompose further |

---

#### ACID & Isolation Levels

| Property | Meaning | How PG ensures it |
|----------|---------|-------------------|
| **Atomicity** | All succeed or all fail | `ROLLBACK` undoes everything |
| **Consistency** | Data satisfies all constraints | Constraints, checks, triggers |
| **Isolation** | Concurrent txns don't interfere | MVCC + isolation levels |
| **Durability** | Committed data survives crashes | Write-Ahead Log (WAL) |

| Isolation Level | Dirty Read | Non-Repeatable | Phantom Read | PG Default? |
|----------------|------------|----------------|--------------|-------------|
| `READ UNCOMMITTED` | ⚠️ Yes | ⚠️ Yes | ⚠️ Yes | No |
| `READ COMMITTED` | ✅ No | ⚠️ Yes | ⚠️ Yes | ✅ **Yes** |
| `REPEATABLE READ` | ✅ No | ✅ No | ⚠️ Yes | No |
| `SERIALIZABLE` | ✅ No | ✅ No | ✅ No | No (slowest) |

---

#### DELETE vs TRUNCATE vs DROP

| Feature | `DELETE` | `TRUNCATE` | `DROP` |
|---------|----------|------------|--------|
| Removes | Specific rows | All rows | Entire table (structure + data) |
| WHERE clause | ✅ Yes | ❌ No | ❌ No |
| Rollback | ✅ Yes | ⚠️ Depends on DB | ❌ No |
| Triggers fire | ✅ Yes | ❌ No | ❌ No |
| Resets auto-increment | ❌ No | ✅ Yes (with RESTART) | N/A |
| Speed | Slow (row-by-row) | Fast (deallocates pages) | Instant |
| Logged | Fully logged | Minimally logged | N/A |

---

#### Window Functions Overview

Window functions perform calculations across a **set of rows** without collapsing them (unlike GROUP BY).

| Function | What it does | Example |
|----------|-------------|----------|
| `ROW_NUMBER()` | Unique sequential number | `ROW_NUMBER() OVER (ORDER BY sal DESC)` |
| `RANK()` | Rank with gaps on ties | Salaries 90k,80k,80k,70k → Rank: 1,2,2,**4** |
| `DENSE_RANK()` | Rank without gaps | Same → Dense: 1,2,2,**3** |
| `NTILE(n)` | Divide into n buckets | `NTILE(4)` → quartiles 1,2,3,4 |
| `LAG(col, n)` | Previous row's value | Compare current vs previous row |
| `LEAD(col, n)` | Next row's value | Compare current vs next row |
| `FIRST_VALUE(col)` | First in window | Top earner per department |
| `LAST_VALUE(col)` | Last in window | Needs explicit frame clause |
| `SUM() OVER(...)` | Running/cumulative sum | `SUM(sales) OVER (ORDER BY date)` |
| `AVG() OVER(...)` | Moving average | `AVG(sales) OVER (ROWS 6 PRECEDING)` |
| `PERCENT_RANK()` | Relative rank (0 to 1) | Percentile position |

---

#### Common Interview Patterns

| # | Pattern | SQL Approach |
|---|---------|-------------|
| 1 | **Find duplicates** | `GROUP BY col HAVING COUNT(*) > 1` |
| 2 | **Delete duplicates** | CTE + `ROW_NUMBER()`, delete where rn > 1 |
| 3 | **Nth highest salary** | `DENSE_RANK()` or `DISTINCT + LIMIT 1 OFFSET n-1` |
| 4 | **Running total** | `SUM(col) OVER (ORDER BY date)` |
| 5 | **YoY growth** | `LAG(revenue, 1) OVER (ORDER BY year)` |
| 6 | **Gaps in sequence** | `GENERATE_SERIES + LEFT JOIN` |
| 7 | **Pivot (rows→cols)** | `CASE WHEN + SUM()` or `crosstab()` |
| 8 | **Recursive hierarchy** | `WITH RECURSIVE cte AS (base UNION ALL recursive)` |
| 9 | **Top N per group** | `ROW_NUMBER() OVER (PARTITION BY ... ORDER BY ...)` |
| 10 | **Consecutive days** | `date - ROW_NUMBER()` = same group ID |
| 11 | **Anti-join** | `LEFT JOIN ... WHERE t2.id IS NULL` or `NOT EXISTS` |
| 12 | **Cumulative %** | `SUM() OVER(ORDER BY) / SUM() OVER()` |
| 13 | **Moving average** | `AVG() OVER (ORDER BY date ROWS n PRECEDING)` |
| 14 | **EXISTS vs IN** | `EXISTS` faster for large subqueries |
| 15 | **Self-join employee→manager** | `JOIN employees m ON e.manager_id = m.id` |

### Database Basics & DDL (Data Definition Language)

DDL commands define and modify the **structure** of database objects — tables, schemas, indexes. These changes are auto-committed (cannot be rolled back in most databases).

---

#### CREATE TABLE

Tables are the foundation of a relational database. Each column has a name, data type, and optional constraints.


```sql
-- MySQL syntax (from course material)
CREATE TABLE IF NOT EXISTS students (
    student_id INT AUTO_INCREMENT PRIMARY KEY,
    first_name VARCHAR(50) NOT NULL,
    last_name VARCHAR(50) NOT NULL,
    date_of_birth DATE,
    email VARCHAR(100) UNIQUE,
    phone VARCHAR(15),
    gender ENUM('Male', 'Female', 'Other'),
    enrollment_date DATE DEFAULT (CURRENT_DATE),
    is_active BOOLEAN DEFAULT TRUE,
    cgpa DECIMAL(3,2),
    address TEXT,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    updated_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP ON UPDATE CURRENT_TIMESTAMP
);

-- PostgreSQL equivalent
CREATE TABLE employees (
    id          SERIAL PRIMARY KEY,
    first_name  VARCHAR(50) NOT NULL,
    last_name   VARCHAR(50) NOT NULL,
    email       VARCHAR(100) UNIQUE NOT NULL,
    salary      NUMERIC(10, 2) DEFAULT 30000.00,
    dept_id     INTEGER REFERENCES departments(id),
    hire_date   DATE DEFAULT CURRENT_DATE,
    is_active   BOOLEAN DEFAULT TRUE,
    created_at  TIMESTAMPTZ DEFAULT NOW()
);

-- Composite primary key (many-to-many junction table)
CREATE TABLE enrollments (
    student_id  INT REFERENCES students(student_id) ON DELETE CASCADE,
    course_id   INT REFERENCES courses(course_id) ON DELETE CASCADE,
    grade       VARCHAR(2),
    enrolled_at DATE DEFAULT CURRENT_DATE,
    PRIMARY KEY (student_id, course_id)  -- composite PK
);
```


---

#### ALTER TABLE

Modify existing table structure without dropping and recreating.


```sql
-- Add column
ALTER TABLE students ADD COLUMN nationality VARCHAR(50) DEFAULT 'Indian';

-- Drop column
ALTER TABLE students DROP COLUMN nationality;

-- Rename column (MySQL)
ALTER TABLE students CHANGE COLUMN nationality country VARCHAR(50);
-- Rename column (PostgreSQL)
ALTER TABLE students RENAME COLUMN nationality TO country;

-- Change data type
ALTER TABLE employees ALTER COLUMN salary TYPE NUMERIC(12, 2);
-- MySQL: ALTER TABLE employees MODIFY COLUMN salary DECIMAL(12,2);

-- Add constraint
ALTER TABLE students ADD CONSTRAINT chk_cgpa CHECK (cgpa >= 0 AND cgpa <= 10);

-- Drop constraint
ALTER TABLE students DROP CONSTRAINT chk_cgpa;

-- Rename table
ALTER TABLE employees RENAME TO staff;
```


---

#### DROP, TRUNCATE, DELETE — Differences


```sql
-- DROP: removes entire table structure + data (IRREVERSIBLE)
DROP TABLE IF EXISTS temp_table;
DROP TABLE employees CASCADE;  -- also removes dependent FK references

-- TRUNCATE: removes all data, keeps structure (faster than DELETE)
TRUNCATE TABLE employees;
TRUNCATE TABLE employees RESTART IDENTITY CASCADE;  -- reset SERIAL

-- DELETE: removes specific rows, fully logged, can be rolled back
DELETE FROM employees WHERE is_active = FALSE;

-- DESCRIBE table structure
-- MySQL: DESCRIBE students;  or  SHOW CREATE TABLE students;
-- PostgreSQL: \d students  (in psql)
```

### Data Manipulation — INSERT, UPDATE, DELETE

DML commands modify the **data** inside tables. Unlike DDL, DML operations can be rolled back within a transaction.

---

#### INSERT


```sql
-- Single row
INSERT INTO departments (department_name, hod_name, office_location, budget)
VALUES ('Computer Science', 'Dr. Rajesh Kumar', 'Block A', 5000000.00);

-- Multiple rows
INSERT INTO students (first_name, last_name, email, cgpa, gender) VALUES
    ('Rahul', 'Sharma', 'rahul@example.com', 8.5, 'Male'),
    ('Priya', 'Patel', 'priya@example.com', 9.2, 'Female'),
    ('Amit', 'Kumar', 'amit@example.com', 7.8, 'Male');

-- Insert from SELECT (copy data between tables)
INSERT INTO emp_archive
SELECT * FROM employees WHERE is_active = FALSE;

-- UPSERT: Insert or Update on conflict (PostgreSQL)
INSERT INTO employees (email, first_name, salary)
VALUES ('rahul@example.com', 'Rahul', 80000)
ON CONFLICT (email) DO UPDATE SET salary = EXCLUDED.salary;

-- MySQL equivalent: ON DUPLICATE KEY UPDATE
INSERT INTO employees (email, first_name, salary)
VALUES ('rahul@example.com', 'Rahul', 80000)
ON DUPLICATE KEY UPDATE salary = VALUES(salary);

-- RETURNING clause (PostgreSQL) — get inserted data back
INSERT INTO employees (first_name, last_name, email)
VALUES ('New', 'Employee', 'new@example.com')
RETURNING id, first_name, email;
```


---

#### UPDATE


```sql
-- Simple update
UPDATE students SET phone = '9999999999' WHERE student_id = 1;

-- Update multiple columns
UPDATE students SET cgpa = 9.0, address = 'Mumbai' WHERE student_id = 1;

-- Update with calculation
UPDATE students SET cgpa = cgpa + 0.5 WHERE cgpa < 8.0;

-- Update with JOIN (PostgreSQL)
UPDATE employees e SET salary = salary + d.budget * 0.01
FROM departments d WHERE e.dept_id = d.id AND d.name = 'Sales';

-- Update with JOIN (MySQL)
UPDATE courses c JOIN departments d ON c.department_id = d.department_id
SET c.is_active = FALSE
WHERE d.department_name = 'Database Management';

-- Update with subquery
UPDATE students SET cgpa = (
    SELECT AVG(CASE
        WHEN e.grade='A+' THEN 10 WHEN e.grade='A' THEN 9
        WHEN e.grade='B+' THEN 8 WHEN e.grade='B' THEN 7 ELSE 5
    END) FROM enrollments e WHERE e.student_id = students.student_id
) WHERE student_id IN (SELECT student_id FROM enrollments);
```


---

#### DELETE


```sql
-- Delete specific rows
DELETE FROM students WHERE first_name = 'Test';

-- Delete with JOIN (MySQL)
DELETE c FROM courses c
JOIN departments d ON c.department_id = d.department_id
WHERE c.course_code = 'TEST101';

-- Delete with subquery
DELETE FROM enrollments
WHERE grade = 'F' AND student_id IN (
    SELECT student_id FROM students WHERE cgpa > 8.0
);

-- Delete with RETURNING (PostgreSQL)
DELETE FROM employees WHERE id = 10 RETURNING *;
```


<a id="select"></a>

---

### SELECT & Filtering

The `SELECT` statement is the most used SQL command. It retrieves data from tables with filtering, sorting, and pagination.

---

#### Basic SELECT


```sql
-- All columns
SELECT * FROM students;

-- Specific columns
SELECT first_name, last_name, email, cgpa FROM students;

-- Aliases
SELECT first_name AS "First Name", salary * 12 AS annual_salary FROM employees;

-- DISTINCT
SELECT DISTINCT gender FROM students;

-- DISTINCT ON (PostgreSQL only) — first row per group
SELECT DISTINCT ON (dept_id) * FROM employees
ORDER BY dept_id, salary DESC;
```


---

#### WHERE Clause & Operators

Filter rows using conditions. `WHERE` is applied **before** grouping.


```sql
-- Comparison
SELECT * FROM students WHERE cgpa > 8.5;
SELECT * FROM students WHERE date_of_birth >= '2000-01-01';

-- BETWEEN (inclusive both ends)
SELECT * FROM students WHERE cgpa BETWEEN 8.0 AND 9.0;

-- IN / NOT IN
SELECT * FROM students WHERE student_id IN (1, 3, 5);

-- LIKE pattern matching
-- % = any characters, _ = exactly one character
SELECT * FROM students WHERE first_name LIKE 'A%';      -- starts with A
SELECT * FROM students WHERE email LIKE '%@gmail.com';   -- ends with @gmail.com
SELECT * FROM students WHERE last_name LIKE 'S_____';    -- S + exactly 5 chars

-- ILIKE (case-insensitive, PostgreSQL only)
SELECT * FROM employees WHERE first_name ILIKE 'john%';

-- NULL checks (never use = NULL)
SELECT * FROM students WHERE phone IS NULL;
SELECT * FROM students WHERE phone IS NOT NULL;

-- AND / OR / NOT
SELECT * FROM students
WHERE (dept_id = 1 OR dept_id = 2) AND salary > 50000;

-- EXISTS (check if subquery returns any rows)
SELECT * FROM departments d
WHERE EXISTS (SELECT 1 FROM employees e WHERE e.dept_id = d.id);
```


---

#### ORDER BY, LIMIT, OFFSET


```sql
-- Sort ascending/descending
SELECT * FROM students ORDER BY cgpa DESC;
SELECT * FROM students ORDER BY last_name ASC, first_name ASC;

-- NULLS FIRST / NULLS LAST (PostgreSQL)
SELECT * FROM employees ORDER BY phone NULLS LAST;

-- Pagination with LIMIT/OFFSET
SELECT * FROM students ORDER BY student_id LIMIT 10;           -- first 10
SELECT * FROM students ORDER BY student_id LIMIT 10 OFFSET 20; -- rows 21-30

-- SQL standard: FETCH
SELECT * FROM employees ORDER BY id FETCH FIRST 10 ROWS ONLY;
```


<a id="aggregates"></a>

---

### Aggregate Functions & GROUP BY

Aggregate functions compute a **single result** from multiple rows. `GROUP BY` groups rows before aggregation; `HAVING` filters the resulting groups.

---


```sql
-- Basic aggregates
SELECT
    COUNT(*) AS total_students,
    AVG(cgpa) AS average_cgpa,
    MIN(cgpa) AS lowest,
    MAX(cgpa) AS highest
FROM students;

-- COUNT non-NULL vs COUNT(*)
SELECT COUNT(date_of_birth) AS has_dob FROM students; -- excludes NULLs
SELECT COUNT(*) AS total FROM students;               -- includes NULLs

-- GROUP BY
SELECT gender, COUNT(*) AS count, AVG(cgpa) AS avg_cgpa
FROM students GROUP BY gender;

-- GROUP BY with JOIN
SELECT d.department_name, COUNT(c.course_id) AS courses, SUM(c.credits) AS total_credits
FROM departments d LEFT JOIN courses c ON d.department_id = c.department_id
GROUP BY d.department_name ORDER BY courses DESC;
```


---

#### HAVING — Filter Groups

`WHERE` filters **rows** before grouping. `HAVING` filters **groups** after aggregation.


```sql
-- Departments with more than 1 course
SELECT department_id, COUNT(*) AS course_count
FROM courses GROUP BY department_id HAVING COUNT(*) > 1;

-- STRING_AGG / GROUP_CONCAT
-- PostgreSQL:
SELECT dept_id, STRING_AGG(first_name, ', ' ORDER BY first_name) AS members
FROM employees GROUP BY dept_id;
-- MySQL:
SELECT dept_id, GROUP_CONCAT(first_name ORDER BY first_name SEPARATOR ', ') AS members
FROM employees GROUP BY dept_id;

-- ROLLUP (subtotals + grand total)
SELECT dept_id, EXTRACT(YEAR FROM hire_date) AS yr, COUNT(*)
FROM employees GROUP BY ROLLUP(dept_id, EXTRACT(YEAR FROM hire_date));

-- Conditional aggregation with CASE
SELECT
    COUNT(*) AS total,
    COUNT(CASE WHEN gender='Male' THEN 1 END) AS male_count,
    COUNT(CASE WHEN gender='Female' THEN 1 END) AS female_count,
    AVG(CASE WHEN gender='Male' THEN cgpa END) AS male_avg
FROM students;
```


<a id="joins"></a>

---

### JOINs — All Types

JOINs combine rows from two or more tables based on related columns (usually FK→PK). The type of JOIN determines what happens with unmatched rows.

---

#### INNER JOIN

Returns only rows with a match in **both** tables.


```sql
SELECT s.first_name, s.last_name, c.course_name, e.grade
FROM students s
INNER JOIN enrollments e ON s.student_id = e.student_id
INNER JOIN courses c ON e.course_id = c.course_id;
```


---

#### LEFT JOIN (LEFT OUTER JOIN)

All rows from left + matching right. NULLs where no match on right.


```sql
-- All students, even those not enrolled in any course
SELECT s.first_name, c.course_name, e.grade
FROM students s
LEFT JOIN enrollments e ON s.student_id = e.student_id
LEFT JOIN courses c ON e.course_id = c.course_id;

-- Anti-join: find students NOT enrolled anywhere
SELECT s.first_name FROM students s
LEFT JOIN enrollments e ON s.student_id = e.student_id
WHERE e.student_id IS NULL;
```


---

#### RIGHT JOIN, FULL OUTER JOIN, CROSS JOIN


```sql
-- RIGHT JOIN: all courses, even those with no enrollments
SELECT c.course_name, e.student_id
FROM enrollments e RIGHT JOIN courses c ON e.course_id = c.course_id;

-- FULL OUTER JOIN (PostgreSQL; MySQL uses UNION trick)
SELECT e.first_name, d.name FROM employees e
FULL OUTER JOIN departments d ON e.dept_id = d.id;

-- MySQL FULL JOIN workaround:
SELECT s.first_name, c.course_name
FROM students s LEFT JOIN enrollments e ON s.student_id = e.student_id
LEFT JOIN courses c ON e.course_id = c.course_id
UNION
SELECT s.first_name, c.course_name
FROM students s RIGHT JOIN enrollments e ON s.student_id = e.student_id
RIGHT JOIN courses c ON e.course_id = c.course_id
WHERE s.student_id IS NULL;

-- CROSS JOIN: every combination (cartesian product)
SELECT s.first_name, c.course_name
FROM students s CROSS JOIN courses c LIMIT 10;
```


---

#### SELF JOIN

A table joined with itself. Common for hierarchical data (employee→manager) and row comparisons.


```sql
-- Employees and their managers
SELECT
    CONCAT(e.first_name, ' ', e.last_name) AS employee,
    CONCAT(m.first_name, ' ', m.last_name) AS manager
FROM employees e
LEFT JOIN employees m ON e.manager_id = m.employee_id;

-- Find employees earning more than their manager
SELECT e.first_name AS emp, e.salary AS emp_sal,
       m.first_name AS mgr, m.salary AS mgr_sal
FROM employees e JOIN employees m ON e.manager_id = m.id
WHERE e.salary > m.salary;
```


<a id="subqueries"></a>

---

### Subqueries & CTEs

**Subqueries** are queries nested inside other queries. **CTEs** (Common Table Expressions) use `WITH` to create named temporary result sets, making complex queries readable.

---

#### Subqueries


```sql
-- Scalar subquery (single value)
SELECT first_name, cgpa FROM students
WHERE cgpa > (SELECT AVG(cgpa) FROM students);

-- IN subquery
SELECT course_code, course_name FROM courses
WHERE course_id IN (SELECT course_id FROM enrollments WHERE student_id = 1);

-- Derived table (subquery in FROM)
SELECT gender, average_cgpa FROM (
    SELECT gender, AVG(cgpa) AS average_cgpa FROM students GROUP BY gender
) sub WHERE average_cgpa > 8.0;

-- EXISTS (correlated subquery — usually faster than IN)
SELECT s.first_name FROM students s
WHERE EXISTS (
    SELECT 1 FROM enrollments e
    WHERE e.student_id = s.student_id AND e.grade = 'A+'
);

-- Subquery in SELECT
SELECT d.department_name,
    (SELECT COUNT(*) FROM courses c WHERE c.department_id = d.department_id) AS course_count
FROM departments d;
```


---

#### CTEs (WITH clause)

CTEs break complex queries into named, readable steps. Each CTE is like a temporary view for that query.


```sql
-- Basic CTE
WITH StudentGrades AS (
    SELECT s.first_name, s.last_name, e.grade, c.course_name
    FROM students s
    JOIN enrollments e ON s.student_id = e.student_id
    JOIN courses c ON e.course_id = c.course_id
)
SELECT * FROM StudentGrades WHERE grade = 'A';

-- Multiple CTEs
WITH
DeptCourses AS (
    SELECT department_id, COUNT(*) AS course_count
    FROM courses GROUP BY department_id
),
TopDepts AS (
    SELECT * FROM DeptCourses WHERE course_count > 1
)
SELECT d.department_name, td.course_count
FROM TopDepts td JOIN departments d ON td.department_id = d.department_id;
```


---

#### Recursive CTEs

Used for **hierarchical / tree** data like org charts, category trees, and sequences.


```sql
-- Employee hierarchy (org chart)
WITH RECURSIVE EmployeeHierarchy AS (
    -- Base: top-level managers (no manager)
    SELECT employee_id, first_name, manager_id, 0 AS level,
        CONCAT(first_name, ' ', last_name) AS path
    FROM employees WHERE manager_id IS NULL
    
    UNION ALL
    
    -- Recursive: each level of reports
    SELECT e.employee_id, e.first_name, e.manager_id, eh.level + 1,
        CONCAT(eh.path, ' > ', e.first_name, ' ', e.last_name)
    FROM employees e
    JOIN EmployeeHierarchy eh ON e.manager_id = eh.employee_id
)
SELECT * FROM EmployeeHierarchy ORDER BY level, first_name;

-- Generate date series (MySQL, since no GENERATE_SERIES)
WITH RECURSIVE DateSeries AS (
    SELECT CURDATE() AS dt
    UNION ALL
    SELECT dt + INTERVAL 1 DAY FROM DateSeries WHERE dt < CURDATE() + INTERVAL 6 DAY
)
SELECT dt, DAYNAME(dt) AS day_of_week FROM DateSeries;
```


<a id="window"></a>

---

### Window Functions

Window functions perform calculations across a **set of rows** related to the current row, **without collapsing** rows into groups. They use the `OVER()` clause.

**Syntax:** `function() OVER (PARTITION BY ... ORDER BY ... frame_clause)`

---

#### Ranking Functions


```sql
-- ROW_NUMBER, RANK, DENSE_RANK: from course sales table
SELECT product_name, category, amount,
    ROW_NUMBER() OVER(ORDER BY amount DESC) AS row_num,
    RANK()       OVER(ORDER BY amount DESC) AS rank,
    DENSE_RANK() OVER(ORDER BY amount DESC) AS dense_rank
FROM sales;

-- Top N per group (INTERVIEW FAVORITE)
WITH ranked AS (
    SELECT *, DENSE_RANK() OVER (
        PARTITION BY dept_id ORDER BY salary DESC
    ) AS rn
    FROM employees
)
SELECT * FROM ranked WHERE rn <= 3;

-- NTILE: divide into quartiles
SELECT first_name, salary, NTILE(4) OVER (ORDER BY salary) AS quartile
FROM employees;
```


---

#### LAG, LEAD — Compare with Previous/Next Row


```sql
-- Previous and next values
SELECT product_name, sale_date, amount,
    LAG(amount, 1)  OVER (ORDER BY sale_date) AS prev_amount,
    LEAD(amount, 1) OVER (ORDER BY sale_date) AS next_amount,
    amount - LAG(amount, 1) OVER (ORDER BY sale_date) AS diff
FROM sales;

-- Year-over-year revenue growth
WITH yearly AS (
    SELECT EXTRACT(YEAR FROM order_date) AS yr, SUM(amount) AS revenue
    FROM orders GROUP BY 1
)
SELECT yr, revenue,
    LAG(revenue) OVER (ORDER BY yr) AS prev_year,
    ROUND((revenue - LAG(revenue) OVER (ORDER BY yr))
        / LAG(revenue) OVER (ORDER BY yr) * 100, 2) AS growth_pct
FROM yearly;
```


---

#### Running Totals & Moving Averages


```sql
-- Cumulative/running total
SELECT sale_date, amount,
    SUM(amount) OVER (ORDER BY sale_date) AS running_total
FROM sales;

-- Per-category running total
SELECT category, sale_date, amount,
    SUM(amount) OVER (PARTITION BY category ORDER BY sale_date) AS cat_running
FROM sales;

-- 3-period moving average
SELECT sale_date, amount,
    AVG(amount) OVER (ORDER BY sale_date ROWS BETWEEN 1 PRECEDING AND 1 FOLLOWING) AS moving_avg
FROM sales;

-- Percentage of total
SELECT first_name, salary,
    ROUND(salary * 100.0 / SUM(salary) OVER (), 2) AS pct_of_total,
    ROUND(salary * 100.0 / SUM(salary) OVER (PARTITION BY dept_id), 2) AS pct_of_dept
FROM employees;
```


<a id="sets"></a>

---

### Set Operations

Set operations combine results from **two or more** SELECT statements. Both queries must have the same number of columns with compatible types.

---


```sql
-- UNION: combine + remove duplicates
SELECT email FROM employees UNION SELECT email FROM contractors;

-- UNION ALL: combine + keep duplicates (faster, no sort needed)
SELECT email FROM employees UNION ALL SELECT email FROM contractors;

-- INTERSECT: rows in BOTH
SELECT email FROM employees INTERSECT SELECT email FROM newsletter;

-- EXCEPT: rows in first but NOT second
SELECT email FROM employees EXCEPT SELECT email FROM terminated;

-- Note: MySQL uses UNION but not INTERSECT/EXCEPT until MySQL 8.0.31+
```


<a id="views"></a>

---

### Views & Materialized Views

A **view** is a saved SQL query that acts as a virtual table — it doesn't store data itself. A **materialized view** (PostgreSQL) stores results on disk for faster reads but must be manually refreshed.

---


```sql
-- Create view
CREATE OR REPLACE VIEW student_course_view AS
SELECT s.student_id, s.first_name, s.last_name, c.course_name, e.grade
FROM students s
JOIN enrollments e ON s.student_id = e.student_id
JOIN courses c ON e.course_id = c.course_id;

-- Use like a regular table
SELECT * FROM student_course_view WHERE grade = 'A';

-- Updatable view with CHECK OPTION
CREATE OR REPLACE VIEW active_students AS
SELECT student_id, first_name, last_name, email, cgpa
FROM students WHERE is_active = TRUE
WITH CHECK OPTION;  -- prevents inserts/updates that would make rows invisible

-- Update through view
UPDATE active_students SET phone = '8888888888' WHERE student_id = 2;

-- View with calculations
CREATE OR REPLACE VIEW dept_summary AS
SELECT d.name, COUNT(e.id) AS emp_count, ROUND(AVG(e.salary),2) AS avg_salary
FROM departments d LEFT JOIN employees e ON d.id = e.dept_id
GROUP BY d.name;

-- Drop view
DROP VIEW IF EXISTS student_course_view;

-- Materialized View (PostgreSQL only)
CREATE MATERIALIZED VIEW mv_monthly_sales AS
SELECT DATE_TRUNC('month', order_date) AS month,
    SUM(amount) AS total, COUNT(*) AS order_count
FROM orders GROUP BY 1 ORDER BY 1;

-- Refresh (required after data changes)
REFRESH MATERIALIZED VIEW mv_monthly_sales;
REFRESH MATERIALIZED VIEW CONCURRENTLY mv_monthly_sales;
-- CONCURRENTLY: doesn't lock reads (needs UNIQUE index)
```


<a id="indexes"></a>

---

### Indexes & Performance

An **index** is a data structure that speeds up data retrieval at the cost of extra storage and slower writes. Think of it like a book index — instead of reading every page, you jump directly to the relevant section.

**Types of indexes:**
- **B-Tree** (default): Best for equality `=` and range `<, >, BETWEEN` queries
- **Hash**: Only equality `=`; rarely better than B-Tree
- **GIN** (Generalized Inverted): For JSONB, arrays, full-text search
- **GiST** (Generalized Search Tree): Geometric data, range types
- **BRIN** (Block Range): Physically sorted data like timestamps — very compact

---


```sql
-- B-Tree (default)
CREATE INDEX idx_emp_salary ON employees(salary);

-- Composite index (leftmost prefix rule!)
CREATE INDEX idx_emp_dept_sal ON employees(dept_id, salary);
-- Helps: WHERE dept_id = 1
-- Helps: WHERE dept_id = 1 AND salary > 50000
-- Does NOT help: WHERE salary > 50000 (no leftmost column)

-- Unique index
CREATE UNIQUE INDEX idx_emp_email ON employees(email);

-- Partial index (only index some rows)
CREATE INDEX idx_active ON employees(dept_id) WHERE is_active = TRUE;

-- Expression index
CREATE INDEX idx_lower_name ON employees(LOWER(last_name));

-- GIN index (JSONB, arrays, full-text)
CREATE INDEX idx_meta ON products USING GIN(metadata);

-- BRIN index (timestamps, naturally ordered)
CREATE INDEX idx_created ON orders USING BRIN(created_at);

-- Covering index with INCLUDE (PG 11+)
CREATE INDEX idx_dept_incl ON employees(dept_id) INCLUDE(first_name, salary);

-- Fulltext index (MySQL)
ALTER TABLE courses ADD FULLTEXT INDEX idx_ft(course_name, description);
SELECT * FROM courses WHERE MATCH(course_name, description)
AGAINST('database' IN NATURAL LANGUAGE MODE);

-- Drop index
DROP INDEX IF EXISTS idx_emp_salary;

-- Check index usage (PostgreSQL)
SELECT indexrelname, idx_scan, idx_tup_read
FROM pg_stat_user_indexes WHERE schemaname = 'public'
ORDER BY idx_scan DESC;

-- Show indexes (MySQL)
SHOW INDEX FROM students;
```


<a id="constraints"></a>

---

### Constraints & Data Integrity

Constraints are rules that enforce data integrity at the database level. They prevent invalid data from being inserted or updated.

| Constraint | Purpose | Example |
|-----------|---------|----------|
| `NOT NULL` | Column cannot be NULL | `name VARCHAR(50) NOT NULL` |
| `UNIQUE` | No duplicate values allowed | `email VARCHAR(100) UNIQUE` |
| `PRIMARY KEY` | NOT NULL + UNIQUE identifier | `id SERIAL PRIMARY KEY` |
| `FOREIGN KEY` | References another table's PK | `REFERENCES departments(id)` |
| `CHECK` | Custom validation rule | `CHECK (salary >= 0)` |
| `DEFAULT` | Auto value if not specified | `DEFAULT CURRENT_DATE` |
| `EXCLUDE` | Prevent overlapping ranges (PG) | Used with range types |

---


```sql
-- Foreign key with cascade actions
CREATE TABLE orders (
    id SERIAL PRIMARY KEY,
    employee_id INTEGER REFERENCES employees(id)
        ON DELETE SET NULL    -- set FK to NULL if parent deleted
        ON UPDATE CASCADE,    -- update FK if parent PK changes
    amount NUMERIC(10, 2) NOT NULL CHECK (amount > 0)
);

-- FK action options:
-- CASCADE: auto delete/update child rows
-- SET NULL: set FK to NULL
-- SET DEFAULT: set FK to default value
-- RESTRICT: prevent parent delete if children exist (immediate)
-- NO ACTION: same as RESTRICT but deferred (default)

-- Named constraint
ALTER TABLE students ADD CONSTRAINT chk_cgpa CHECK (cgpa >= 0 AND cgpa <= 10);

-- Composite unique
ALTER TABLE enrollments ADD CONSTRAINT uq_student_course UNIQUE (student_id, course_id);

-- Exclusion constraint: prevent overlapping bookings (PostgreSQL)
CREATE EXTENSION IF NOT EXISTS btree_gist;
ALTER TABLE room_bookings ADD CONSTRAINT no_overlap
EXCLUDE USING GIST (room_id WITH =, tsrange(start_time, end_time) WITH &&);
```


<a id="procedures"></a>

---

### Stored Procedures & Functions

**Functions** return a value and can be used in SELECT. **Procedures** (PG 11+) can contain transaction control (COMMIT/ROLLBACK) but cannot be used in queries.

| Feature | Function | Procedure |
|---------|----------|----------|
| Returns value | ✅ Yes | ❌ No (OUT params only) |
| Use in SELECT | ✅ Yes | ❌ No |
| Transaction control | ❌ No | ✅ Yes |
| Call syntax | `SELECT func()` | `CALL proc()` |

---


```sql
-- MySQL: Stored procedure with parameters
DELIMITER //
CREATE PROCEDURE GetStudentsByGrade(IN grade_param VARCHAR(2))
BEGIN
    SELECT s.first_name, s.last_name, c.course_name, e.grade
    FROM students s
    JOIN enrollments e ON s.student_id = e.student_id
    JOIN courses c ON e.course_id = c.course_id
    WHERE e.grade = grade_param;
END //
DELIMITER ;
CALL GetStudentsByGrade('A');

-- MySQL: Procedure with OUT parameters
DELIMITER //
CREATE PROCEDURE GetDepartmentStats(
    IN dept_id INT, OUT course_count INT, OUT avg_credits DECIMAL(5,2)
)
BEGIN
    SELECT COUNT(*), AVG(credits) INTO course_count, avg_credits
    FROM courses WHERE department_id = dept_id;
END //
DELIMITER ;
CALL GetDepartmentStats(1, @count, @avg);
SELECT @count, @avg;

-- PostgreSQL: Function
CREATE OR REPLACE FUNCTION get_salary_grade(sal NUMERIC) RETURNS TEXT AS $$
BEGIN
    IF sal >= 100000 THEN RETURN 'A';
    ELSIF sal >= 70000 THEN RETURN 'B';
    ELSIF sal >= 40000 THEN RETURN 'C';
    ELSE RETURN 'D';
    END IF;
END;
$$ LANGUAGE plpgsql;

SELECT first_name, salary, get_salary_grade(salary) AS grade FROM employees;

-- PostgreSQL: Function returning table
CREATE OR REPLACE FUNCTION get_dept_employees(dept TEXT)
RETURNS TABLE(name TEXT, sal NUMERIC) AS $$
BEGIN
    RETURN QUERY SELECT e.first_name::TEXT, e.salary
    FROM employees e JOIN departments d ON e.dept_id = d.id
    WHERE d.name = dept;
END; $$ LANGUAGE plpgsql;

SELECT * FROM get_dept_employees('Engineering');

-- MySQL: Custom function
DELIMITER //
CREATE FUNCTION CalculateAge(birth_date DATE) RETURNS INT DETERMINISTIC
BEGIN
    RETURN TIMESTAMPDIFF(YEAR, birth_date, CURDATE());
END //
DELIMITER ;
SELECT first_name, CalculateAge(date_of_birth) AS age FROM students;
```


<a id="triggers"></a>

---

### Triggers

Triggers automatically execute a function **BEFORE** or **AFTER** an INSERT, UPDATE, or DELETE. Common uses: audit logging, data validation, auto-updating timestamps.

---


```sql
-- MySQL: Audit trail triggers
CREATE TABLE student_audit (
    audit_id INT AUTO_INCREMENT PRIMARY KEY,
    student_id INT, action_type VARCHAR(10),
    action_time TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    old_data JSON, new_data JSON
);

DELIMITER //
CREATE TRIGGER after_student_insert AFTER INSERT ON students
FOR EACH ROW BEGIN
    INSERT INTO student_audit (student_id, action_type, new_data)
    VALUES (NEW.student_id, 'INSERT',
        JSON_OBJECT('first_name',NEW.first_name,'email',NEW.email,'cgpa',NEW.cgpa));
END //

CREATE TRIGGER after_student_update AFTER UPDATE ON students
FOR EACH ROW BEGIN
    INSERT INTO student_audit (student_id, action_type, old_data, new_data)
    VALUES (NEW.student_id, 'UPDATE',
        JSON_OBJECT('first_name',OLD.first_name,'cgpa',OLD.cgpa),
        JSON_OBJECT('first_name',NEW.first_name,'cgpa',NEW.cgpa));
END //

-- Validation trigger
CREATE TRIGGER before_enrollment_insert BEFORE INSERT ON enrollments
FOR EACH ROW BEGIN
    DECLARE cnt INT;
    SELECT COUNT(*) INTO cnt FROM enrollments WHERE student_id = NEW.student_id;
    IF cnt >= 5 THEN
        SIGNAL SQLSTATE '45000'
        SET MESSAGE_TEXT = 'Student cannot enroll in more than 5 courses';
    END IF;
END //
DELIMITER ;

-- PostgreSQL: Audit trigger
CREATE OR REPLACE FUNCTION log_changes() RETURNS TRIGGER AS $$
BEGIN
    IF TG_OP = 'INSERT' THEN
        INSERT INTO audit_log(tbl, action, new_data)
        VALUES (TG_TABLE_NAME, 'INSERT', row_to_json(NEW)::JSONB);
        RETURN NEW;
    ELSIF TG_OP = 'UPDATE' THEN
        INSERT INTO audit_log(tbl, action, old_data, new_data)
        VALUES (TG_TABLE_NAME, 'UPDATE',
            row_to_json(OLD)::JSONB, row_to_json(NEW)::JSONB);
        RETURN NEW;
    ELSIF TG_OP = 'DELETE' THEN
        INSERT INTO audit_log(tbl, action, old_data)
        VALUES (TG_TABLE_NAME, 'DELETE', row_to_json(OLD)::JSONB);
        RETURN OLD;
    END IF;
END; $$ LANGUAGE plpgsql;

CREATE TRIGGER trg_audit AFTER INSERT OR UPDATE OR DELETE ON employees
FOR EACH ROW EXECUTE FUNCTION log_changes();
```


<a id="transactions"></a>

---

### Transactions & Concurrency

A **transaction** groups multiple SQL statements into a single atomic unit — either ALL succeed or ALL fail. PostgreSQL uses MVCC for concurrent access.

---


```sql
-- Basic transaction
BEGIN;  -- or START TRANSACTION (MySQL)
    UPDATE accounts SET balance = balance - 1000 WHERE id = 1;
    UPDATE accounts SET balance = balance + 1000 WHERE id = 2;
COMMIT;  -- save all changes
-- ROLLBACK;  -- undo all changes

-- Savepoints (partial rollback)
BEGIN;
    INSERT INTO employees VALUES (...);  -- this stays
    SAVEPOINT sp1;
    INSERT INTO employees VALUES (...);  -- this might fail
    ROLLBACK TO sp1;  -- undo only second insert
COMMIT;  -- first insert is saved

-- MySQL: Transaction in procedure
DELIMITER //
CREATE PROCEDURE TransferStudent(IN sid INT, IN from_cid INT, IN to_cid INT)
BEGIN
    DECLARE EXIT HANDLER FOR SQLEXCEPTION BEGIN ROLLBACK; END;
    START TRANSACTION;
    DELETE FROM enrollments WHERE student_id=sid AND course_id=from_cid;
    INSERT INTO enrollments (student_id, course_id) VALUES (sid, to_cid);
    COMMIT;
    SELECT 'Transfer successful' AS message;
END //
DELIMITER ;

-- Set isolation level
BEGIN ISOLATION LEVEL SERIALIZABLE;
    SELECT * FROM accounts WHERE id = 1;
    UPDATE accounts SET balance = balance - 100 WHERE id = 1;
COMMIT;

-- Row-level locking
SELECT * FROM accounts WHERE id = 1 FOR UPDATE;  -- locks until COMMIT

-- SKIP LOCKED (job queue pattern)
SELECT * FROM jobs WHERE status = 'pending'
ORDER BY id LIMIT 1 FOR UPDATE SKIP LOCKED;
```

### Advanced PostgreSQL — Arrays, JSON, Full-Text Search

PostgreSQL extends standard SQL with powerful data types and operators not found in most other databases.

---

#### Arrays

PostgreSQL supports **array columns** — store multiple values in a single field. Arrays are 1-indexed.


```sql
CREATE TABLE students_pg (
    id SERIAL PRIMARY KEY, name TEXT,
    scores INTEGER[]
);
INSERT INTO students_pg (name, scores) VALUES
    ('Alice', '{90, 85, 95}'), ('Bob', ARRAY[75, 80, 70]);

-- Access (1-indexed!)
SELECT name, scores[1] AS first_score FROM students_pg;

-- Array functions
SELECT name, array_length(scores,1) AS n,
    array_to_string(scores, ', ') AS s FROM students_pg;

-- ANY / ALL
SELECT * FROM students_pg WHERE 90 = ANY(scores);
SELECT * FROM students_pg WHERE 70 < ALL(scores);

-- Unnest: expand to rows
SELECT name, unnest(scores) AS score FROM students_pg;

-- Aggregate into array
SELECT dept_id, ARRAY_AGG(first_name ORDER BY first_name) FROM employees GROUP BY dept_id;
```


---

#### JSONB

JSONB stores JSON in a binary format that supports indexing and efficient querying. Always prefer JSONB over JSON.


```sql
CREATE TABLE products (id SERIAL PRIMARY KEY, name TEXT, metadata JSONB DEFAULT '{}');

INSERT INTO products (name, metadata) VALUES
('Laptop', '{"brand":"Dell","specs":{"ram":16,"ssd":512},"tags":["electronics"]}');

-- Access: -> returns JSON, ->> returns TEXT
SELECT metadata->>'brand' AS brand,
    (metadata->'specs'->>'ram')::INT AS ram
FROM products;

-- Containment @> and key existence ?
SELECT * FROM products WHERE metadata @> '{"brand":"Dell"}';
SELECT * FROM products WHERE metadata ? 'brand';

-- Update JSON
UPDATE products SET metadata = metadata || '{"color":"silver"}' WHERE id=1;
UPDATE products SET metadata = metadata - 'color' WHERE id=1; -- remove key

-- MySQL JSON equivalent
UPDATE students SET preferences =
    '{"theme":"dark","notifications":true,"language":"en"}'
WHERE student_id = 1;
SELECT JSON_EXTRACT(preferences, '$.theme') AS theme FROM students;

-- GIN index for JSONB
CREATE INDEX idx_meta ON products USING GIN(metadata);

-- Build JSON from rows
SELECT jsonb_agg(jsonb_build_object('name',first_name,'salary',salary))
FROM employees WHERE dept_id = 1;
```


---

#### Full-Text Search

PostgreSQL has built-in full-text search — no need for ElasticSearch for basic use cases.


```sql
-- Basic search
SELECT * FROM articles
WHERE to_tsvector('english', title || ' ' || body)
    @@ to_tsquery('english', 'python & database');

-- Dedicated column + GIN index for performance
ALTER TABLE articles ADD COLUMN search_vec TSVECTOR;
UPDATE articles SET search_vec = to_tsvector('english', title || ' ' || COALESCE(body,''));
CREATE INDEX idx_search ON articles USING GIN(search_vec);

-- Search with ranking
SELECT title, ts_rank(search_vec, to_tsquery('python & web')) AS rank
FROM articles WHERE search_vec @@ to_tsquery('python & web')
ORDER BY rank DESC;

-- Auto-update trigger
CREATE TRIGGER trg_search BEFORE INSERT OR UPDATE ON articles
FOR EACH ROW EXECUTE FUNCTION
    tsvector_update_trigger(search_vec, 'pg_catalog.english', title, body);
```

### Query Optimization & EXPLAIN

Understanding query execution plans is crucial for writing performant SQL. PostgreSQL's `EXPLAIN` shows how the query planner will execute your query.

---


```sql
-- See query plan
EXPLAIN SELECT * FROM employees WHERE salary > 50000;

-- Actually run and show timing
EXPLAIN ANALYZE SELECT * FROM employees WHERE dept_id = 1;

-- MySQL equivalent
EXPLAIN SELECT s.first_name, c.course_name, e.grade
FROM students s
JOIN enrollments e ON s.student_id = e.student_id
JOIN courses c ON e.course_id = c.course_id
WHERE s.cgpa > 8.0;

-- Read the plan:
-- Seq Scan       = full table scan (no index)
-- Index Scan     = using index
-- Index Only Scan = covering index (fastest)
-- Bitmap Scan    = combining multiple indexes
-- Hash/Merge/Nested Loop = join strategies

-- Update statistics
ANALYZE employees;
VACUUM employees;
VACUUM FULL employees;  -- aggressive; locks table
-- MySQL: OPTIMIZE TABLE students, enrollments;

-- Table size
SELECT pg_size_pretty(pg_total_relation_size('employees'));

-- Find slow queries (PostgreSQL)
CREATE EXTENSION IF NOT EXISTS pg_stat_statements;
SELECT query, calls, mean_exec_time
FROM pg_stat_statements ORDER BY total_exec_time DESC LIMIT 10;

-- Active connections
SELECT pid, state, query FROM pg_stat_activity WHERE state = 'active';
-- MySQL: SHOW PROCESSLIST;

-- Kill query
SELECT pg_cancel_backend(pid);     -- graceful
SELECT pg_terminate_backend(pid);  -- forceful
```

### Practical Case Studies

Real-world SQL scenarios from the CDAC coursework. These queries demonstrate how to combine multiple concepts.

#### Student Transcript Generation


```sql
SELECT s.student_id,
    CONCAT(s.first_name, ' ', s.last_name) AS student_name,
    c.course_code, c.course_name, c.credits, e.grade,
    CASE
        WHEN e.grade='A+' THEN 10.0 WHEN e.grade='A' THEN 9.0
        WHEN e.grade='A-' THEN 8.5 WHEN e.grade='B+' THEN 8.0
        WHEN e.grade='B' THEN 7.0 WHEN e.grade='B-' THEN 6.0
        ELSE 5.0
    END AS grade_points,
    c.credits * CASE
        WHEN e.grade='A+' THEN 10.0 WHEN e.grade='A' THEN 9.0
        WHEN e.grade='A-' THEN 8.5 WHEN e.grade='B+' THEN 8.0
        WHEN e.grade='B' THEN 7.0 ELSE 5.0
    END AS weighted_points
FROM students s
LEFT JOIN enrollments e ON s.student_id = e.student_id
LEFT JOIN courses c ON e.course_id = c.course_id
WHERE s.student_id = 1 ORDER BY c.course_code;
```


---

#### Department Performance Analysis


```sql
WITH CourseStats AS (
    SELECT d.department_id, d.department_name,
        COUNT(DISTINCT c.course_id) AS course_count,
        COUNT(DISTINCT e.student_id) AS student_count,
        AVG(CASE
            WHEN e.grade='A+' THEN 10 WHEN e.grade='A' THEN 9
            WHEN e.grade='B+' THEN 8 WHEN e.grade='B' THEN 7 ELSE 5
        END) AS avg_grade
    FROM departments d
    LEFT JOIN courses c ON d.department_id = c.department_id
    LEFT JOIN enrollments e ON c.course_id = e.course_id
    GROUP BY d.department_id, d.department_name
)
SELECT department_name, course_count, student_count,
    ROUND(avg_grade, 2) AS avg_grade,
    CASE
        WHEN avg_grade >= 9.0 THEN 'Excellent'
        WHEN avg_grade >= 8.0 THEN 'Very Good'
        WHEN avg_grade >= 7.0 THEN 'Good'
        ELSE 'Needs Improvement'
    END AS performance
FROM CourseStats ORDER BY avg_grade DESC;
```


---

#### Course Popularity & Completion Rate


```sql
SELECT c.course_name, d.department_name,
    COUNT(e.enrollment_id) AS enrolled,
    COUNT(CASE WHEN e.grade IS NOT NULL THEN 1 END) AS completed,
    ROUND(COUNT(CASE WHEN e.grade IS NOT NULL THEN 1 END) * 100.0
        / NULLIF(COUNT(e.enrollment_id), 0), 2) AS completion_rate
FROM courses c
LEFT JOIN departments d ON c.department_id = d.department_id
LEFT JOIN enrollments e ON c.course_id = e.course_id
GROUP BY c.course_name, d.department_name
ORDER BY enrolled DESC;
```

#### Pivot: Sales by Category by Month


```sql
SELECT category,
    SUM(CASE WHEN MONTH(sale_date)=1 THEN amount ELSE 0 END) AS January,
    SUM(CASE WHEN MONTH(sale_date)=2 THEN amount ELSE 0 END) AS February,
    SUM(CASE WHEN MONTH(sale_date)=3 THEN amount ELSE 0 END) AS March,
    SUM(amount) AS Total
FROM sales GROUP BY category;

-- PostgreSQL: using crosstab
CREATE EXTENSION IF NOT EXISTS tablefunc;
SELECT * FROM crosstab(
    'SELECT product, quarter, revenue FROM sales ORDER BY 1, 2',
    $$VALUES ('Q1'),('Q2'),('Q3'),('Q4')$$
) AS ct(product TEXT, q1 NUMERIC, q2 NUMERIC, q3 NUMERIC, q4 NUMERIC);
```

### Interview Questions — Worked Solutions

These are the most frequently asked SQL coding questions with complete solutions.

---

#### Q1: Find the Nth highest salary


```sql
-- Method 1: DENSE_RANK
SELECT salary FROM (
    SELECT salary, DENSE_RANK() OVER (ORDER BY salary DESC) AS rn
    FROM employees
) sub WHERE rn = 2;  -- change N here

-- Method 2: LIMIT + OFFSET
SELECT DISTINCT salary FROM employees
ORDER BY salary DESC LIMIT 1 OFFSET 1;  -- 2nd highest

-- Method 3: Subquery
SELECT MAX(salary) FROM employees
WHERE salary < (SELECT MAX(salary) FROM employees);
```


---

#### Q2: Find & delete duplicate records


```sql
-- Find duplicates
SELECT email, COUNT(*) FROM employees
GROUP BY email HAVING COUNT(*) > 1;

-- Delete (keep lowest id)
DELETE FROM employees WHERE id NOT IN (
    SELECT MIN(id) FROM employees GROUP BY email
);

-- Better: CTE + ROW_NUMBER
WITH ranked AS (
    SELECT id, ROW_NUMBER() OVER (PARTITION BY email ORDER BY id) AS rn
    FROM employees
)
DELETE FROM employees WHERE id IN (SELECT id FROM ranked WHERE rn > 1);
```


---

#### Q3: Department with highest average salary


```sql
-- With RANK in case of ties
WITH dept_avg AS (
    SELECT d.name, AVG(e.salary) AS avg_sal,
        RANK() OVER (ORDER BY AVG(e.salary) DESC) AS rn
    FROM employees e JOIN departments d ON e.dept_id = d.id
    GROUP BY d.name
)
SELECT name, ROUND(avg_sal, 2) FROM dept_avg WHERE rn = 1;
```


---

#### Q4: Employees earning more than their manager


```sql
SELECT e.first_name AS employee, e.salary AS emp_sal,
       m.first_name AS manager, m.salary AS mgr_sal
FROM employees e JOIN employees m ON e.manager_id = m.id
WHERE e.salary > m.salary;
```


---

#### Q5: Find consecutive login days (streak)


```sql
-- ROW_NUMBER date trick
WITH login_groups AS (
    SELECT user_id, login_date,
        login_date - (ROW_NUMBER() OVER (
            PARTITION BY user_id ORDER BY login_date
        ))::INT AS grp
    FROM user_logins
)
SELECT user_id, MIN(login_date) AS start, MAX(login_date) AS end,
    COUNT(*) AS days
FROM login_groups GROUP BY user_id, grp
HAVING COUNT(*) >= 3 ORDER BY days DESC;
```


---

#### Q6: Pivot — rows to columns


```sql
SELECT product_name,
    SUM(CASE WHEN quarter='Q1' THEN revenue ELSE 0 END) AS q1,
    SUM(CASE WHEN quarter='Q2' THEN revenue ELSE 0 END) AS q2,
    SUM(CASE WHEN quarter='Q3' THEN revenue ELSE 0 END) AS q3,
    SUM(CASE WHEN quarter='Q4' THEN revenue ELSE 0 END) AS q4,
    SUM(revenue) AS total
FROM quarterly_sales GROUP BY product_name;
```


---

#### Q7: Running total & cumulative %


```sql
SELECT order_date, amount,
    SUM(amount) OVER (ORDER BY order_date) AS running_total,
    ROUND(SUM(amount) OVER (ORDER BY order_date) * 100.0 /
        SUM(amount) OVER (), 2) AS cum_pct
FROM orders ORDER BY order_date;
```


---

#### Q8: Find gaps in a sequence


```sql
SELECT s.id AS missing_id
FROM generate_series(
    (SELECT MIN(id) FROM employees),
    (SELECT MAX(id) FROM employees)
) s(id)
LEFT JOIN employees e ON e.id = s.id
WHERE e.id IS NULL;
```


---

#### Q9: Top N per group


```sql
WITH ranked AS (
    SELECT first_name, dept_id, salary,
        DENSE_RANK() OVER (PARTITION BY dept_id ORDER BY salary DESC) AS rn
    FROM employees
)
SELECT * FROM ranked WHERE rn <= 3;
```


---

#### Q10: Recursive category tree


```sql
WITH RECURSIVE cat_tree AS (
    SELECT id, name, parent_id, name::TEXT AS path, 1 AS depth
    FROM categories WHERE parent_id IS NULL
    UNION ALL
    SELECT c.id, c.name, c.parent_id,
        ct.path || ' > ' || c.name, ct.depth + 1
    FROM categories c JOIN cat_tree ct ON c.parent_id = ct.id
)
SELECT * FROM cat_tree ORDER BY path;
-- Output: Electronics > Phones > iPhone
```

## Section 1: SQL Fundamentals & Core Concepts

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | What is SQL? | SQL, Structured Query Language | • Standard language for managing relational databases<br>• Create tables, insert/update/delete data, retrieve with SELECT<br>• Declarative — you say WHAT you want, not HOW<br>• Used by PostgreSQL, MySQL, Oracle, SQL Server |
| 2 | What is a database vs a schema? | Database, schema | • Database: entire collection of data + metadata<br>• Schema: namespace within a database grouping tables/views<br>• PostgreSQL: one DB can have multiple schemas<br>• Default schema is `public` |
| 3 | Categories of SQL commands (DDL, DML, DCL, TCL)? | DDL, DML, DCL, TCL | • **DDL**: CREATE, ALTER, DROP, TRUNCATE (structure)<br>• **DML**: SELECT, INSERT, UPDATE, DELETE (data)<br>• **DCL**: GRANT, REVOKE (permissions)<br>• **TCL**: COMMIT, ROLLBACK, SAVEPOINT (transactions) |
| 4 | DDL vs DML? | DDL, DML, difference | • DDL changes **structure** (schema) — auto-commits<br>• DML changes **data** (rows) — can be rolled back<br>• DDL: CREATE TABLE, ALTER TABLE, DROP TABLE<br>• DML: INSERT, UPDATE, DELETE, SELECT |
| 5 | DELETE vs TRUNCATE vs DROP? | DELETE, TRUNCATE, DROP | • `DELETE`: removes rows, can use WHERE, logged, fires triggers<br>• `TRUNCATE`: removes ALL rows, faster, resets identity, no triggers<br>• `DROP`: removes entire table (structure + data)<br>• DELETE is rollback-able; TRUNCATE in PG also rollback-able |
| 6 | WHERE vs HAVING? | WHERE, HAVING, filter | • `WHERE`: filters rows BEFORE grouping<br>• `HAVING`: filters groups AFTER GROUP BY<br>• WHERE cannot use aggregate functions<br>• HAVING without GROUP BY treats entire result as one group |
| 7 | CHAR vs VARCHAR? | CHAR, VARCHAR, fixed | • `CHAR(n)`: fixed-length, padded with spaces<br>• `VARCHAR(n)`: variable-length, no padding<br>• Use CHAR for fixed-size codes (country codes, hashes)<br>• In PG, `VARCHAR` without length = unlimited (same as TEXT) |
| 8 | Primary key vs unique key? | PK, unique, constraint | • PK: only one per table, no NULLs, main row identifier<br>• UNIQUE: multiple per table, allows one NULL<br>• Both enforce uniqueness and create an index<br>• PK has stronger semantic meaning |
| 9 | What is a foreign key? | Foreign key, FK, referential | • References a primary key in another table<br>• Enforces referential integrity<br>• Can be NULL (unless constrained), can have multiple FKs<br>• Actions: CASCADE, SET NULL, RESTRICT on delete/update |
| 10 | What is NULL? | NULL, IS NULL, COALESCE | • Represents unknown or missing data — NOT zero, NOT empty string<br>• Any comparison with NULL → NULL (not TRUE/FALSE)<br>• Check with `IS NULL` / `IS NOT NULL`<br>• `COALESCE(col, default)` replaces NULL |
| 11 | NULL vs empty string? | NULL, empty string | • NULL = absence of value; empty string = a value (zero-length text)<br>• `NULL IS NULL` → TRUE; `'' IS NULL` → FALSE<br>• `LENGTH(NULL)` → NULL; `LENGTH('')` → 0<br>• Test: `IS NULL` for NULL, `= ''` for empty string |
| 12 | UNION vs UNION ALL? | UNION, UNION ALL | • `UNION`: combines results, removes duplicates (slower)<br>• `UNION ALL`: combines results, keeps duplicates (faster)<br>• Both require same number and type of columns<br>• Use UNION ALL when duplicates are OK or impossible |
| 13 | SQL logical order of execution? | Execution order, FROM | • FROM → WHERE → GROUP BY → HAVING → SELECT → ORDER BY → LIMIT<br>• Can't use column aliases in WHERE (SELECT runs after)<br>• CAN use aliases in ORDER BY<br>• Understanding this explains many SQL behaviors |
| 14 | Why is execution order important? | Execution order, aliases | • Explains why aliases don't work in WHERE<br>• Explains why HAVING can use aggregates but WHERE can't<br>• Helps debug unexpected results<br>• Essential for writing correct complex queries |
| 15 | What is SQL injection? How to prevent? | SQL injection, security | • Attacker inserts malicious SQL through user input<br>• `' OR 1=1 --` bypasses authentication<br>• **Always use parameterized queries** (prepared statements)<br>• Never concatenate user input into SQL strings |
| 16 | OLTP vs OLAP? | OLTP, OLAP, analytical | • **OLTP**: operational — many short transactions, normalized, write-heavy<br>• **OLAP**: analytical — complex queries, denormalized, read-heavy<br>• OLTP: row-based; OLAP: columnar storage<br>• Different table designs for each |
| 17 | Temporary table vs CTE vs view? | Temp table, CTE, view | • Temp table: persists for session, can be indexed<br>• CTE: exists for one query only, no DDL needed<br>• View: permanent named query, no data stored<br>• Temp table for large intermediate results; CTE for readability |
| 18 | IN vs EXISTS? | IN, EXISTS, subquery | • `IN`: checks if value is in a list or subquery result<br>• `EXISTS`: checks if subquery returns ANY rows<br>• EXISTS stops at first match (often faster for large subqueries)<br>• `NOT IN` fails with NULLs — prefer `NOT EXISTS` |
| 19 | COALESCE vs ISNULL/IFNULL? | COALESCE, ISNULL | • `COALESCE(a, b, c)`: returns first non-NULL (SQL standard, any DB)<br>• `ISNULL(a, b)`: SQL Server specific, only two args<br>• `IFNULL(a, b)`: MySQL specific, only two args<br>• Always prefer COALESCE for portability |
| 20 | NULLIF function? | NULLIF, division by zero | • Returns NULL if two values are equal: `NULLIF(a, b)` → NULL if a=b<br>• Prevents division by zero: `x / NULLIF(y, 0)`<br>• Opposite-ish of COALESCE<br>• Useful for avoiding special-case values |
| 21 | CASE statement — simple vs searched? | CASE, WHEN, conditional | • Simple: `CASE col WHEN 'A' THEN 'Active' END`<br>• Searched: `CASE WHEN salary > 100k THEN 'High' END`<br>• Searched CASE is more flexible (any condition)<br>• Works in SELECT, WHERE, ORDER BY, GROUP BY |
| 22 | Division by zero handling? | Division, zero, NULLIF | • `x / NULLIF(y, 0)` — returns NULL instead of error<br>• `CASE WHEN y != 0 THEN x/y ELSE 0 END`<br>• PostgreSQL raises error on integer division by zero<br>• Always protect division with NULLIF or CASE |
| 23 | CAST vs :: in PostgreSQL? | CAST, ::, type conversion | • `CAST(col AS INTEGER)` — SQL standard, portable<br>• `col::INTEGER` — PostgreSQL shorthand<br>• Both do the same thing<br>• Use CAST for cross-database compatibility |
| 24 | Type coercion — implicit casting problems? | Coercion, implicit cast | • SQL auto-converts types in some cases (implicit cast)<br>• Can prevent index usage: `WHERE int_col = '5'`<br>• Can cause unexpected results with dates/numbers<br>• Always use explicit CAST for clarity |
| 25 | SELECT * vs specific columns? | SELECT *, performance | • SELECT * retrieves unnecessary columns (more I/O)<br>• Prevents covering index (index-only scan)<br>• Schema changes may break application code<br>• Always select only needed columns |
| 26 | DISTINCT vs GROUP BY for deduplication? | DISTINCT, GROUP BY | • Both remove duplicate rows<br>• DISTINCT: simpler syntax for basic dedup<br>• GROUP BY: needed when combining with aggregates<br>• Performance typically identical for simple cases |
| 27 | BETWEEN — inclusive? Gotchas with dates? | BETWEEN, range, dates | • BETWEEN is **inclusive** on both ends<br>• `BETWEEN '2024-01-01' AND '2024-01-31'` includes both<br>• With timestamps: `BETWEEN '2024-01-01' AND '2024-01-31'` misses Jan 31 after midnight<br>• Use `>= AND <` for date ranges |
| 28 | LIKE wildcards and escaping? | LIKE, %, _, wildcard | • `%` = any characters; `_` = single character<br>• `LIKE 'A%'` — starts with A<br>• Escape with `ESCAPE`: `LIKE '10\%' ESCAPE '\'`<br>• PG: `ILIKE` for case-insensitive |
| 29 | LIKE vs ILIKE in PostgreSQL? | LIKE, ILIKE, case | • `LIKE`: case-sensitive pattern matching<br>• `ILIKE`: case-insensitive (PostgreSQL only)<br>• MySQL: LIKE is case-insensitive by default<br>• For indexed searches, consider `pg_trgm` or full-text |
| 30 | LIMIT/OFFSET vs FETCH FIRST? | LIMIT, OFFSET, FETCH FIRST | • `LIMIT n OFFSET m` — MySQL/PostgreSQL syntax<br>• `FETCH FIRST n ROWS ONLY` — SQL standard<br>• Both produce same result<br>• Deep OFFSET is slow — prefer keyset pagination |
| 31 | ORDER BY with NULLs? | ORDER BY, NULLS FIRST, LAST | • PG: NULLs sort last in ASC, first in DESC (default)<br>• Control: `ORDER BY col ASC NULLS FIRST`<br>• `ORDER BY col DESC NULLS LAST`<br>• MySQL: NULLs sort first in ASC |
| 32 | SELECT INTO vs INSERT INTO ... SELECT? | SELECT INTO, INSERT SELECT | • `SELECT INTO new_table FROM ...` — creates new table with results<br>• `INSERT INTO existing_table SELECT ...` — inserts into existing<br>• SELECT INTO fails if table exists<br>• INSERT INTO SELECT preserves table structure |
| 33 | Correlated vs non-correlated subquery? | Correlated, subquery | • Non-correlated: runs once, independent of outer query<br>• Correlated: references outer query, runs per outer row<br>• Correlated is slower (executes N times)<br>• Often rewritable as JOIN for better performance |
| 34 | Generate a sequence of numbers or dates? | generate_series, sequence | • PG: `generate_series(1, 100)` — numbers<br>• PG: `generate_series('2024-01-01'::date, '2024-12-31', '1 day')` — dates<br>• MySQL: recursive CTE or numbers table<br>• Used for gap analysis, calendar tables |
| 35 | ANY, ALL, SOME operators? | ANY, ALL, SOME | • `WHERE x > ANY (subquery)` — greater than at least one<br>• `WHERE x > ALL (subquery)` — greater than every value<br>• `SOME` is synonym for `ANY`<br>• `= ANY(...)` is equivalent to `IN(...)` |


## Section 2: Joins — Types, Performance & Pitfalls

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | INNER, LEFT, RIGHT, FULL OUTER JOIN? | JOIN types, INNER, LEFT | • **INNER**: only matching rows from both tables<br>• **LEFT**: all from left + matching right (NULLs if no match)<br>• **RIGHT**: all from right + matching left<br>• **FULL OUTER**: all from both tables |
| 2 | INNER JOIN vs LEFT JOIN — when to use? | INNER, LEFT, choice | • INNER: when you only want rows with matches in both tables<br>• LEFT: when you want ALL rows from left table regardless<br>• LEFT JOIN + `WHERE right.id IS NULL` → find unmatched rows<br>• INNER is more restrictive |
| 3 | What is a CROSS JOIN? | CROSS JOIN, cartesian | • Produces cartesian product — every row × every row<br>• No join condition needed<br>• 10 rows × 10 rows = 100 result rows<br>• Use for: generating combinations (sizes × colors) |
| 4 | What is a self-join? | SELF JOIN, same table | • Joins a table to itself using aliases<br>• `FROM emp e1 JOIN emp e2 ON e1.mgr_id = e2.id`<br>• Used for hierarchies (employee → manager)<br>• Also for comparing rows within same table |
| 5 | ON vs WHERE in LEFT JOIN? | ON, WHERE, LEFT JOIN | • `ON`: conditions applied DURING join (preserves LEFT rows)<br>• `WHERE`: filters AFTER join (removes LEFT rows with NULLs)<br>• For INNER JOIN: same result either way<br>• For OUTER JOIN: ON vs WHERE gives different results |
| 6 | Implicit join (comma) vs explicit JOIN? | Implicit, explicit, JOIN | • Implicit: `FROM a, b WHERE a.id = b.aid` — old style<br>• Explicit: `FROM a JOIN b ON a.id = b.aid` — modern<br>• Explicit is clearer — separates join from filter<br>• Always use explicit JOIN syntax |
| 7 | USING clause vs ON? | USING, ON, join | • `USING(col)` — shorthand when column name matches in both tables<br>• `ON a.col = b.col` — works with different column names<br>• USING produces one copy of the column in result<br>• ON is more flexible and explicit |
| 8 | What is a NATURAL JOIN? Why dangerous? | NATURAL JOIN, auto | • Automatically joins on all columns with same name<br>• No explicit ON clause needed<br>• Dangerous: columns may match unintentionally<br>• Avoid in production — use explicit JOIN ON |
| 9 | What is a semi-join? | Semi-join, EXISTS | • Returns rows from left where match EXISTS in right<br>• Does NOT return columns from right table<br>• `WHERE EXISTS (SELECT 1 FROM orders WHERE orders.user_id = users.id)`<br>• Also possible with `IN (subquery)` |
| 10 | What is an anti-join? | Anti-join, NOT EXISTS | • Returns rows from left with NO match in right<br>• `WHERE NOT EXISTS (...)` or `LEFT JOIN ... WHERE right.id IS NULL`<br>• Used for finding orphan records<br>• Prefer NOT EXISTS over NOT IN (NULL-safe) |
| 11 | NOT IN with NULL values — gotcha? | NOT IN, NULL, pitfall | • `NOT IN (1, 2, NULL)` returns NO rows (always UNKNOWN)<br>• NULL in subquery poisons the entire NOT IN<br>• Fix: `WHERE col NOT IN (SELECT x FROM t WHERE x IS NOT NULL)`<br>• Better: use `NOT EXISTS` instead |
| 12 | Join on multiple conditions? | Multiple conditions, AND | • `FROM a JOIN b ON a.id = b.aid AND a.type = b.type`<br>• Multiple conditions joined with AND (or OR)<br>• Can mix equality and inequality conditions<br>• More conditions = more selective join |
| 13 | Non-equi join (inequality join)? | Non-equi, theta join | • Join using `<`, `>`, `BETWEEN` instead of `=`<br>• `FROM trans t JOIN tiers r ON t.amount BETWEEN r.low AND r.high`<br>• Used for range lookups, pricing tiers<br>• Equi-join: `=`; Theta-join: any operator |
| 14 | Joining more than two tables? Best order? | Multiple tables, chain | • Chain joins: `FROM a JOIN b ON ... JOIN c ON ...`<br>• Each JOIN relates to any previously joined table<br>• Optimizer usually determines best physical order<br>• Write for readability; let optimizer handle order |
| 15 | Does join order affect performance? | Join order, optimizer | • Logical order (what you write) usually doesn't matter<br>• Optimizer rearranges joins for best plan<br>• Physical join order chosen by planner based on statistics<br>• Ensure table stats are up to date (`ANALYZE`) |
| 16 | Hash join vs nested loop vs merge join? | Hash, nested loop, merge | • **Hash join**: builds hash table on smaller table — O(n+m)<br>• **Nested loop**: for each row in outer, scan inner — good with index<br>• **Merge join**: both sides sorted, merge — O(n log n + m log m)<br>• Optimizer picks based on data size and indexes |
| 17 | Troubleshoot a slow join? | Slow join, EXPLAIN | • Run `EXPLAIN ANALYZE` — check join type and row estimates<br>• Add indexes on join columns<br>• Check for implicit type casting on join conditions<br>• Ensure table statistics are current (`ANALYZE`) |
| 18 | Join on different data types? | Type mismatch, cast | • Implicit casting may prevent index usage<br>• `WHERE int_col = varchar_val` → full scan<br>• Always match types: `CAST(col AS INTEGER)`<br>• Best practice: join columns should be same type |
| 19 | What is a LATERAL join? | LATERAL, correlated FROM | • Allows subquery in FROM to reference earlier tables<br>• `FROM orders o, LATERAL (SELECT ... WHERE user_id = o.user_id) sub`<br>• Like a correlated subquery but in FROM clause<br>• PG-specific; MySQL 8.0+ supports it |
| 20 | Simulate FULL OUTER JOIN in MySQL? | FULL OUTER, simulate | • MySQL doesn't support FULL OUTER JOIN natively<br>• `LEFT JOIN UNION ALL RIGHT JOIN WHERE left IS NULL`<br>• Or: `LEFT JOIN UNION RIGHT JOIN` (with UNION to dedup)<br>• PostgreSQL supports FULL OUTER directly |
| 21 | Join vs subquery — when each preferred? | Join, subquery, choice | • Join: when you need columns from both tables<br>• Subquery: for existence checks or scalar values<br>• JOINs often optimize better (optimizer can rearrange)<br>• Correlated subqueries are often slower than joins |
| 22 | Find rows with no match (orphan records)? | No match, orphan, LEFT JOIN | • `LEFT JOIN b ON ... WHERE b.id IS NULL`<br>• Or `WHERE NOT EXISTS (SELECT 1 FROM b WHERE ...)`<br>• Both find rows in A with no match in B<br>• NOT EXISTS is often cleaner and NULL-safe |
| 23 | Equi-join vs theta-join? | Equi-join, theta-join | • Equi-join: uses `=` operator (most common)<br>• Theta-join: uses any comparison operator (`<`, `>`, `!=`, `BETWEEN`)<br>• INNER/LEFT/RIGHT JOINs are usually equi-joins<br>• Theta-joins used for ranges and intervals |
| 24 | Many-to-many in joins — junction table? | M:N, junction, bridge | • M:N relationship needs a junction table<br>• `student_courses(student_id, course_id)` — both FKs<br>• Join: `students JOIN student_courses JOIN courses`<br>• Junction table can hold extra data (grade, enrolled_at) |
| 25 | Performance impact of joining large tables? | Large join, optimize | • Add indexes on all join columns<br>• Filter early with WHERE to reduce rows before join<br>• Consider partial indexes for common filters<br>• Use EXPLAIN ANALYZE to verify plan |
| 26 | Join on range condition (price tier lookup)? | Range join, BETWEEN | • `FROM trans t JOIN tiers r ON t.amount BETWEEN r.min AND r.max`<br>• Non-equi join — can be slow without proper indexing<br>• Consider a GiST index on range types in PG<br>• Or pre-compute tier assignment |
| 27 | Duplicate keys in join — row explosion? | Duplicates, explosion | • If both sides have duplicate key values, result multiplies<br>• 3 rows × 4 rows with same key = 12 result rows<br>• Fix: deduplicate with DISTINCT or GROUP BY before joining<br>• Or use window function + filter |
| 28 | All combinations of two columns from different tables? | Combinations, CROSS JOIN | • `SELECT a.col, b.col FROM a CROSS JOIN b`<br>• Produces every combination (cartesian product)<br>• Or implicit: `SELECT a.col, b.col FROM a, b`<br>• Use for: size × color, date × category |
| 29 | INNER JOIN vs INTERSECT? | INNER JOIN, INTERSECT | • INNER JOIN: matches rows on condition, can return any columns<br>• INTERSECT: returns rows common to both queries (same columns)<br>• INTERSECT auto-deduplicates; INNER JOIN doesn't<br>• Different tools for different problems |
| 30 | Join across different schemas? | Schema, cross-schema | • `FROM schema1.table1 JOIN schema2.table2 ON ...`<br>• Fully qualify with `schema.table`<br>• Need SELECT privilege on both schemas<br>• Use `search_path` to set default schema |
| 31 | What is join elimination? | Join elimination, optimizer | • Optimizer removes unnecessary joins<br>• If joined table adds no columns or filters, it's removed<br>• Requires FK + UNIQUE constraint for optimizer to trust<br>• Common in views that join more than needed |
| 32 | Recursive self-join (org chart)? | Recursive, self-join, CTE | • Use `WITH RECURSIVE` CTE for hierarchies<br>• Base: `SELECT * FROM emp WHERE manager_id IS NULL`<br>• Recursive: `SELECT e.* FROM emp e JOIN cte ON e.manager_id = cte.id`<br>• Better than repeated self-joins |
| 33 | LEFT JOIN vs LEFT SEMI JOIN? | LEFT, SEMI, difference | • LEFT JOIN: returns ALL left rows + matching right columns<br>• LEFT SEMI JOIN: returns left rows that have a match (no right columns)<br>• Semi-join via `WHERE EXISTS (...)`<br>• Semi-join often faster (stops at first match) |
| 34 | Debug query returning more rows than expected? | Debug, row explosion | • Check for M:N duplicates in join keys<br>• Add `COUNT(*)` before and after join to compare<br>• Look for missing join conditions<br>• Use DISTINCT or GROUP BY to identify duplication |
| 35 | Join table to window function result? | Window function, join | • Use CTE or subquery with window function<br>• `WITH ranked AS (SELECT *, ROW_NUMBER() OVER (...) rn FROM t) SELECT * FROM ranked JOIN other ON ...`<br>• Window functions can't go in JOIN ON directly<br>• Wrap in subquery first |


## Section 3: Views & Materialized Views

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | What is a view? How does it differ from a table? | View, virtual table | • Named query saved in the database<br>• No data stored — query runs each time view is accessed<br>• Simplifies complex queries, provides security layer<br>• `CREATE VIEW v AS SELECT ...` |
| 2 | When should you use a view? | View, benefits | • Simplify complex joins for end users<br>• Restrict column/row access (security)<br>• Provide a stable API when underlying tables change<br>• Abstract common query patterns |
| 3 | Can you update data through a view? | Updatable view, INSERT | • Simple views (single table, no aggregates) are updatable<br>• Complex views (JOINs, GROUP BY, DISTINCT) are NOT updatable<br>• `WITH CHECK OPTION` prevents inserts the view can't see<br>• Use INSTEAD OF triggers for complex updatable views |
| 4 | What is WITH CHECK OPTION? | WITH CHECK OPTION, view | • Prevents INSERT/UPDATE through view if row wouldn't be visible<br>• `CREATE VIEW active_users AS SELECT * FROM users WHERE active = true WITH CHECK OPTION`<br>• Can't insert `active=false` through this view<br>• Enforces view's WHERE condition |
| 5 | What is a materialized view? | Materialized view, snapshot | • Stores query result physically (like a cached table)<br>• Must be refreshed to get updated data<br>• Much faster reads than regular view<br>• Takes storage space (unlike regular view) |
| 6 | Materialized view vs regular view? | Mat view, view, difference | • View: virtual, always current, no storage, slower reads<br>• Mat view: physical, snapshot, uses storage, fast reads<br>• Mat view needs `REFRESH MATERIALIZED VIEW`<br>• Use mat view for expensive aggregations |
| 7 | How to refresh a materialized view? CONCURRENTLY? | REFRESH, CONCURRENTLY | • `REFRESH MATERIALIZED VIEW mv_name` — locks during refresh<br>• `REFRESH ... CONCURRENTLY` — no lock, requires UNIQUE index<br>• Concurrent refresh is slower but doesn't block reads<br>• Schedule refreshes with pg_cron or application |
| 8 | Can you index materialized views? | Index, materialized view | • Yes — create indexes just like on regular tables<br>• `CREATE INDEX idx ON mv_name(col)`<br>• Required for `REFRESH CONCURRENTLY`<br>• Significantly improves query performance |
| 9 | Performance trade-off of materialized views? | Mat view, trade-off | • Faster reads (pre-computed results)<br>• Uses storage space (data duplicated)<br>• Data can be stale (until refreshed)<br>• Refresh is expensive for large datasets |
| 10 | Materialized view vs summary/aggregate table? | Mat view, summary table | • Mat view: simpler, declarative, auto-refresh<br>• Summary table: more control, triggers/ETL to maintain<br>• Mat view: refresh replaces ALL data<br>• Summary table: can be incrementally updated |
| 11 | View joining multiple tables? | View, multi-table | • Yes, views can join any number of tables<br>• `CREATE VIEW v AS SELECT ... FROM a JOIN b ON ... JOIN c ON ...`<br>• Simplifies complex queries for end users<br>• Performance = running the underlying query |
| 12 | What happens if underlying table is dropped? | View, dependency, DROP | • View becomes invalid — queries against it will error<br>• `DROP TABLE ... CASCADE` drops dependent views too<br>• Without CASCADE, DROP fails if views depend on table<br>• Check dependencies before dropping |
| 13 | Can you index regular views? | View, index, no | • No — regular views have no physical data to index<br>• Indexes go on the underlying tables<br>• Materialized views CAN be indexed<br>• Optimize by indexing the tables the view queries |
| 14 | What is a recursive view? | Recursive view, WITH RECURSIVE | • View defined using `WITH RECURSIVE`<br>• `CREATE RECURSIVE VIEW v AS (base UNION ALL recursive)`<br>• Used for hierarchical data traversal<br>• Equivalent to writing recursive CTE in a view |
| 15 | Check definition of an existing view? | View definition, pg_views | • `\d+ view_name` in psql<br>• `SELECT pg_get_viewdef('view_name')` — PG function<br>• `information_schema.views` — standard SQL<br>• `pg_views` catalog table |
| 16 | Nested views (view on a view) — performance? | Nested view, performance | • Each nested level adds query complexity<br>• Optimizer may or may not inline/merge views<br>• Can lead to poor plans if deeply nested<br>• Avoid more than 2-3 levels of nesting |
| 17 | Views for row-level security? | View, RLS, security | • Create view with WHERE clause filtering user's data<br>• Grant access only to view, not underlying table<br>• PG also has native Row-Level Security (RLS) policies<br>• Views are simpler; RLS is more robust |
| 18 | CREATE OR REPLACE VIEW vs DROP + CREATE? | CREATE OR REPLACE, view | • `CREATE OR REPLACE`: updates definition, preserves grants/permissions<br>• `DROP + CREATE`: grants are lost, must re-GRANT<br>• REPLACE can't change column names/types<br>• Prefer OR REPLACE for safe updates |
| 19 | Parameterize a view (workarounds)? | View, parameter, set_config | • Views don't accept parameters directly<br>• Workaround: use session variables `SET myapp.filter = 'value'`<br>• Read in view: `current_setting('myapp.filter')`<br>• Or use a function that returns a table instead |
| 20 | View vs CTE vs temp table? | View, CTE, temp table | • View: permanent, reusable, virtual, no storage<br>• CTE: single-query scope, readable, auto-cleaned<br>• Temp table: session scope, indexable, physical storage<br>• Choose based on scope and persistence needs |


## Section 4: Stored Procedures & Functions

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | Stored procedure vs function in PostgreSQL? | Procedure, function | • Function: returns a value, usable in SELECT<br>• Procedure: may not return, uses CALL, can manage transactions<br>• Functions can't COMMIT/ROLLBACK; procedures can<br>• PG 11+ introduced CREATE PROCEDURE |
| 2 | Function vs application code? | Function, app logic | • Function: runs close to data (no network round-trips)<br>• App code: easier to test, debug, version control<br>• Functions for: data validation, complex calculations<br>• App code for: business logic, external APIs |
| 3 | IN, OUT, INOUT parameter modes? | IN, OUT, INOUT | • `IN`: input only (default) — passed to function<br>• `OUT`: output only — returned by function<br>• `INOUT`: both input and output<br>• Multiple OUT params = function returns record |
| 4 | RETURNS TABLE function? | RETURNS TABLE, set | • Returns a result set: `RETURNS TABLE(id INT, name TEXT)`<br>• Function body uses `RETURN QUERY SELECT ...`<br>• Called like a table: `SELECT * FROM my_func()`<br>• Alternative to RETURNS SETOF |
| 5 | RETURNS SETOF vs RETURNS TABLE? | SETOF, TABLE, difference | • `RETURNS SETOF type`: returns multiple rows of existing type<br>• `RETURNS TABLE(...)`: defines columns inline<br>• TABLE is more flexible (custom column definitions)<br>• SETOF references an existing type/table |
| 6 | SQL vs PL/pgSQL language for functions? | SQL, PL/pgSQL, language | • SQL: simple queries, inline-optimizable, faster for simple cases<br>• PL/pgSQL: control flow, variables, loops, exception handling<br>• SQL functions can be inlined by optimizer<br>• PL/pgSQL for complex procedural logic |
| 7 | When PL/pgSQL over plain SQL? | PL/pgSQL, procedural | • Need variables, loops, or conditionals<br>• Complex error handling with EXCEPTION<br>• Multi-step data processing<br>• Dynamic SQL with EXECUTE |
| 8 | IMMUTABLE, STABLE, VOLATILE? | IMMUTABLE, STABLE, VOLATILE | • **IMMUTABLE**: same input → always same output (cacheable)<br>• **STABLE**: same within single query (reads DB but no writes)<br>• **VOLATILE**: may return different results each call (default)<br>• Affects optimization and indexing |
| 9 | Marking VOLATILE as IMMUTABLE — danger? | IMMUTABLE, wrong, danger | • Optimizer may cache wrong results<br>• Index on expression may return stale data<br>• Prepared statements may reuse wrong plan<br>• Never lie about volatility — causes subtle bugs |
| 10 | Error handling in PL/pgSQL? | EXCEPTION, error handling | • `BEGIN ... EXCEPTION WHEN error_type THEN handler END`<br>• `RAISE NOTICE 'msg'` — print messages<br>• `RAISE EXCEPTION 'msg'` — abort with error<br>• `SQLSTATE` codes for specific error handling |
| 11 | SECURITY DEFINER vs SECURITY INVOKER? | SECURITY DEFINER, INVOKER | • DEFINER: runs with privileges of function OWNER<br>• INVOKER: runs with privileges of CALLER (default)<br>• DEFINER: use for controlled access to restricted tables<br>• DEFINER: security risk if not careful (search_path!) |
| 12 | Function overloading in PostgreSQL? | Overloading, same name | • Multiple functions with same name but different parameter types<br>• PG dispatches based on argument types<br>• `my_func(INTEGER)` vs `my_func(TEXT)` — different functions<br>• Can cause confusion — use sparingly |
| 13 | Function returning multiple output columns? | Multiple outputs, OUT | • Use multiple OUT parameters: `CREATE FUNCTION f(OUT a INT, OUT b TEXT)`<br>• Or `RETURNS TABLE(a INT, b TEXT)`<br>• Call: `SELECT * FROM f()` to get all columns<br>• Or `SELECT (f()).a` for specific column |
| 14 | Inline SQL vs PL/pgSQL performance? | Inline, SQL function, performance | • Simple SQL functions can be inlined by optimizer<br>• Inlined = treated as part of calling query (faster)<br>• PL/pgSQL is never inlined — separate execution<br>• Prefer SQL functions for simple calculations |
| 15 | How optimizer inlines SQL functions? | Inline, optimizer | • Replaces function call with function body in query<br>• Requirements: SQL language, IMMUTABLE/STABLE, no side effects<br>• Check with EXPLAIN — function won't appear as separate node<br>• Huge performance benefit |
| 16 | Variadic parameters? | VARIADIC, variable args | • `CREATE FUNCTION f(VARIADIC args INT[])` — accepts any number of ints<br>• Called: `f(1, 2, 3, 4)` or `f(VARIADIC ARRAY[1,2,3])`<br>• Inside function: args is an array<br>• Useful for flexible APIs |
| 17 | Function that modifies data? | DML, function, INSERT | • Functions CAN do INSERT/UPDATE/DELETE<br>• Must be marked VOLATILE<br>• `PERFORM INSERT INTO ...` in PL/pgSQL<br>• Procedures are preferred for data-modifying operations |
| 18 | What is PERFORM in PL/pgSQL? | PERFORM, discard result | • Executes a query but discards the result<br>• `PERFORM some_function()` — call function, ignore return<br>• Used when you don't need the return value<br>• SELECT without target variable causes error; PERFORM doesn't |
| 19 | Debug a PL/pgSQL function? | Debug, PL/pgSQL | • `RAISE NOTICE 'var = %', var` — print debug output<br>• `plpgsql_check` extension — static analysis<br>• pgAdmin debugger — step through code<br>• Write unit tests for functions |
| 20 | RAISE NOTICE, WARNING, EXCEPTION? | RAISE, NOTICE, EXCEPTION | • `RAISE NOTICE 'info'` — informational message<br>• `RAISE WARNING 'warn'` — warning (doesn't abort)<br>• `RAISE EXCEPTION 'error'` — aborts transaction<br>• `RAISE EXCEPTION 'user % not found', user_id` — with params |
| 21 | EXECUTE for dynamic SQL — injection risks? | EXECUTE, dynamic SQL, injection | • `EXECUTE 'SELECT * FROM ' &#124;&#124; table_name` — DANGEROUS<br>• Safe: `EXECUTE format('SELECT * FROM %I WHERE id = %L', tbl, id)`<br>• `%I` = identifier (quoted), `%L` = literal (escaped)<br>• Always use format() to prevent injection |
| 22 | Dollar quoting ($$)? | $$, dollar quoting | • Alternative to single quotes for function bodies<br>• `CREATE FUNCTION f() RETURNS void AS $$ ... $$ LANGUAGE plpgsql`<br>• No need to escape single quotes inside $$<br>• Can nest: `$outer$ ... $inner$ ... $inner$ ... $outer$` |
| 23 | Drop and replace functions? | DROP FUNCTION, CREATE OR REPLACE | • `DROP FUNCTION func_name(param_types)` — must specify params<br>• `CREATE OR REPLACE FUNCTION` — updates body, keeps grants<br>• Can't change return type with REPLACE<br>• DROP requires exact parameter signature |
| 24 | What is a trigger function? | Trigger function, TG_OP | • Special function returning TRIGGER type<br>• Called automatically on table events<br>• Has access to `NEW`, `OLD`, `TG_OP`, `TG_TABLE_NAME`<br>• Must return NEW (for BEFORE) or NULL (for AFTER) |
| 25 | How to test stored procedures and functions? | Test, function, pgTAP | • `SELECT my_func(args)` — manual testing<br>• `pgTAP` extension — unit testing framework for PG<br>• Assert expected results: `SELECT is(my_func(1), 42, 'test name')`<br>• Test edge cases: NULLs, empty inputs, large data |


## Section 5: Triggers

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | What is a trigger? When to use? | Trigger, event, automatic | • Code that executes automatically on table events<br>• Events: INSERT, UPDATE, DELETE<br>• Use for: audit logs, timestamps, data validation<br>• Defined per-table with `CREATE TRIGGER` |
| 2 | BEFORE vs AFTER trigger? | BEFORE, AFTER, timing | • BEFORE: runs before operation — can modify or reject data<br>• AFTER: runs after operation — for side effects, logging<br>• BEFORE to validate/modify NEW row<br>• AFTER for cascading changes to other tables |
| 3 | What is an INSTEAD OF trigger? | INSTEAD OF, view | • Replaces the triggering operation entirely<br>• Only on views (not tables)<br>• Makes non-updatable views updatable<br>• You write the INSERT/UPDATE/DELETE logic manually |
| 4 | Row-level vs statement-level trigger? | FOR EACH ROW, statement | • Row-level (`FOR EACH ROW`): fires once per affected row<br>• Statement-level (`FOR EACH STATEMENT`): fires once per query<br>• Row-level has access to OLD and NEW row values<br>• Statement-level for aggregate actions |
| 5 | NEW and OLD record variables? | NEW, OLD, trigger | • `OLD`: row values BEFORE the operation (UPDATE, DELETE)<br>• `NEW`: row values AFTER the operation (INSERT, UPDATE)<br>• INSERT: only NEW; DELETE: only OLD; UPDATE: both<br>• Accessed in trigger function body |
| 6 | What should a trigger function return? | RETURN, trigger function | • BEFORE/INSERT: return NEW to allow, NULL to reject<br>• BEFORE/DELETE: return OLD to allow, NULL to reject<br>• AFTER triggers: return value is ignored<br>• Always return the appropriate row or NULL |
| 7 | Trigger to prevent deletion of certain rows? | Prevent delete, RAISE | • BEFORE DELETE trigger function<br>• Check condition: `IF OLD.protected THEN RAISE EXCEPTION 'Cannot delete'; END IF;`<br>• RETURN OLD to allow normal deletes<br>• Cleaner than application-level checks |
| 8 | Audit log trigger? | Audit, log, trigger | • AFTER INSERT/UPDATE/DELETE trigger<br>• Insert into audit table: `INSERT INTO audit_log(table_name, action, old_data, new_data, changed_at)`<br>• Use `TG_OP` for operation type, `TG_TABLE_NAME` for table<br>• Store OLD/NEW as JSONB |
| 9 | Auto-update modified_at timestamp? | modified_at, auto update | • BEFORE UPDATE trigger<br>• `NEW.modified_at = NOW(); RETURN NEW;`<br>• Applies automatically without app code changes<br>• Very common pattern for all updateable tables |
| 10 | Multiple triggers — execution order? | Order, multiple triggers | • Same timing/event: alphabetical order by trigger name<br>• BEFORE triggers run before AFTER triggers<br>• Row-level BEFORE → operation → row-level AFTER<br>• Statement-level BEFORE → all rows → statement-level AFTER |
| 11 | Disable and enable triggers? | DISABLE, ENABLE trigger | • `ALTER TABLE t DISABLE TRIGGER trigger_name` — stops firing<br>• `ALTER TABLE t ENABLE TRIGGER trigger_name` — resumes<br>• `DISABLE TRIGGER ALL` — all triggers on table<br>• Useful during bulk data loads |
| 12 | Performance impact on write-heavy tables? | Trigger, performance, write | • Each trigger adds overhead to every write operation<br>• Row-level BEFORE + AFTER = 2 extra function calls per row<br>• Complex logic in triggers slows INSERT/UPDATE/DELETE<br>• Minimize work inside triggers |
| 13 | Triggers vs application-level logic? | Trigger, app logic | • Triggers: enforce rules at DB level, can't be bypassed<br>• App logic: easier to test, debug, version control<br>• Triggers for: data integrity, audit<br>• App logic for: business rules, notifications |
| 14 | Triggers vs constraints? | Trigger, constraint | • Constraints: declarative, simpler, better optimized<br>• Triggers: procedural, flexible, can do anything<br>• Prefer constraints when possible (CHECK, FK, UNIQUE)<br>• Use triggers when constraints aren't expressive enough |
| 15 | Prevent recursive/infinite trigger loops? | Recursive, infinite loop | • Trigger A modifies table B → trigger on B modifies table A → loop<br>• Use `pg_trigger_depth()` to detect recursion<br>• `IF pg_trigger_depth() > 1 THEN RETURN NEW; END IF;`<br>• Design triggers carefully to avoid cycles |
| 16 | What is an event trigger? | Event trigger, DDL | • Fires on DDL events (CREATE, ALTER, DROP)<br>• Not tied to a specific table — database-wide<br>• `CREATE EVENT TRIGGER my_trigger ON ddl_command_end`<br>• Use for: schema change auditing, preventing drops |
| 17 | Can a trigger modify another table? | Cross-table, trigger | • Yes — trigger function can INSERT/UPDATE/DELETE other tables<br>• Use for: maintaining denormalized data, cascading updates<br>• Be careful of performance implications<br>• Can cause trigger chains (watch for loops) |
| 18 | Debug trigger behavior? | Debug, trigger | • `RAISE NOTICE 'OLD: %, NEW: %', OLD, NEW` inside function<br>• Check `pg_trigger` catalog for trigger definitions<br>• Test with simple INSERT/UPDATE and check results<br>• Temporarily disable other triggers to isolate |
| 19 | Disadvantages of triggers? | Trigger, disadvantages | • Hidden logic — hard to discover and debug<br>• Performance overhead on every write<br>• Difficult to test in isolation<br>• Can create unexpected side effects and circular dependencies |
| 20 | List all triggers on a table? | List triggers, catalog | • `\dS table_name` in psql — shows triggers<br>• `SELECT * FROM information_schema.triggers WHERE event_object_table = 'my_table'`<br>• `pg_trigger` system catalog for low-level details<br>• `pg_get_triggerdef(oid)` for trigger definition |


## Section 6: Data Types

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | INTEGER vs BIGINT vs SMALLINT? | INTEGER, BIGINT, SMALLINT | • SMALLINT: 2 bytes (-32K to 32K)<br>• INTEGER: 4 bytes (-2.1B to 2.1B)<br>• BIGINT: 8 bytes (-9.2 quintillion)<br>• Use smallest type that fits your data |
| 2 | SERIAL vs BIGSERIAL vs IDENTITY? | SERIAL, IDENTITY | • SERIAL: legacy auto-increment (creates sequence + default)<br>• BIGSERIAL: same but BIGINT<br>• `GENERATED ALWAYS AS IDENTITY`: SQL standard (PG 10+)<br>• Prefer IDENTITY for new tables |
| 3 | NUMERIC/DECIMAL vs REAL/DOUBLE PRECISION? | NUMERIC, FLOAT, precision | • NUMERIC: exact, arbitrary precision — use for money<br>• REAL (4 bytes) / DOUBLE PRECISION (8 bytes): approximate floating-point<br>• Floats are faster but have rounding errors<br>• `0.1 + 0.2 != 0.3` with floats |
| 4 | When NUMERIC over FLOAT? Precision pitfalls? | NUMERIC, precision | • Use NUMERIC for: financial, currency, measurements needing precision<br>• Float: `0.1 + 0.2 = 0.30000000000000004`<br>• NUMERIC: `0.1 + 0.2 = 0.3` (exact)<br>• Floats faster, NUMERIC more accurate |
| 5 | VARCHAR(n) vs VARCHAR vs TEXT? | VARCHAR, TEXT, length | • `VARCHAR(n)`: variable-length with max limit<br>• `VARCHAR` (no limit): unlimited in PG (same as TEXT)<br>• `TEXT`: unlimited text (PostgreSQL preferred)<br>• In PG, all three have same storage/performance |
| 6 | Does VARCHAR(n) length improve performance? | VARCHAR, length, performance | • In PostgreSQL: **No** — VARCHAR(100) and TEXT perform identically<br>• Length is only a constraint, not a storage optimization<br>• Use length constraint for data validation (email max 254 chars)<br>• In MySQL, length can affect storage |
| 7 | BOOLEAN and three-value logic? | BOOLEAN, NULL, three-value | • TRUE, FALSE, and NULL (unknown)<br>• `NULL AND TRUE` → NULL; `NULL OR TRUE` → TRUE<br>• `NOT NULL` → NULL<br>• Always handle NULL booleans explicitly |
| 8 | TIMESTAMP WITH vs WITHOUT TIME ZONE? | TIMESTAMPTZ, TIMESTAMP | • `TIMESTAMPTZ`: stores in UTC, converts on display<br>• `TIMESTAMP`: no timezone info, stored as-is<br>• **Always use TIMESTAMPTZ** for global applications<br>• WITHOUT loses context about what timezone it represents |
| 9 | INTERVAL data type? Date arithmetic? | INTERVAL, date arithmetic | • Represents a time span: `INTERVAL '3 days 4 hours'`<br>• `NOW() + INTERVAL '7 days'` — date math<br>• `age('2024-01-01', '2023-01-01')` → `1 year`<br>• Used in WHERE: `WHERE created_at > NOW() - INTERVAL '30 days'` |
| 10 | UUID data type? Pros/cons as PK? | UUID, primary key | • 128-bit universally unique identifier<br>• Pro: globally unique, no sequence contention, merge-friendly<br>• Con: 16 bytes (vs 4 for INT), random → poor index locality<br>• UUIDv7 (time-ordered) solves locality issue |
| 11 | JSONB vs JSON? | JSONB, JSON, binary | • JSON: stored as text, preserves whitespace and key order<br>• JSONB: binary, parsed once, faster queries, indexable<br>• JSONB supports GIN indexing; JSON does not<br>• **Almost always use JSONB** |
| 12 | JSONB column vs normalized tables? | JSONB, normalized, design | • JSONB: flexible schema, dynamic attributes, nested data<br>• Normalized: enforced structure, better query optimization, integrity<br>• JSONB for: configs, metadata, varying attributes<br>• Normalized for: core entities, relationships |
| 13 | ARRAY data type — query, insert, update? | ARRAY, unnest | • `col INTEGER[]` — array column<br>• Insert: `ARRAY[1,2,3]` or `'{1,2,3}'`<br>• Query: `WHERE 3 = ANY(col)` — membership check<br>• `UNNEST(array)` to expand to rows |
| 14 | ENUM type — pros/cons vs lookup table? | ENUM, lookup table | • ENUM: type-safe, readable, stored as 4 bytes<br>• Lookup table: flexible, no ALTER TYPE needed to add values<br>• ENUM: harder to remove or reorder values<br>• Prefer lookup table for values that may change |
| 15 | BYTEA for binary data? | BYTEA, binary | • Stores raw binary data (images, files, encrypted data)<br>• Max 1 GB per value<br>• Comparison and concatenation supported<br>• For large files, consider external storage + path |
| 16 | INET and CIDR for network addresses? | INET, CIDR, network | • `INET`: stores IP address with optional subnet mask<br>• `CIDR`: network address only (subnet required)<br>• Supports comparison, containment (`>>`, `<<`)<br>• `SELECT * FROM t WHERE ip << '192.168.1.0/24'` |
| 17 | tsvector and tsquery for full-text search? | tsvector, tsquery, search | • `tsvector`: tokenized, normalized document representation<br>• `tsquery`: search query with operators (&, or, !)<br>• `@@` operator matches tsvector against tsquery<br>• GIN index for fast full-text search |
| 18 | Range types — use cases? | Range, int4range, daterange | • `int4range`, `daterange`, `tstzrange` — range columns<br>• Operators: `@>` contains, `&&` overlaps<br>• `EXCLUDE USING GiST` prevents overlapping ranges<br>• Use case: booking systems, price tiers |
| 19 | Composite types? | Composite type, record | • User-defined type with multiple fields<br>• `CREATE TYPE address AS (street TEXT, city TEXT, zip TEXT)`<br>• Can be used as column type or function return<br>• Accessed with dot notation: `(col).field` |
| 20 | What is a domain? | Domain, user-defined type | • Named type with CHECK constraint: `CREATE DOMAIN email AS TEXT CHECK (VALUE ~ '@')`<br>• Reusable constraint across multiple columns<br>• Changes domain → changes all columns using it<br>• Like a type alias with validation |
| 21 | ALTER column data type? | ALTER TYPE, column | • `ALTER TABLE t ALTER COLUMN c TYPE new_type USING expression`<br>• USING clause converts existing data<br>• May need to drop dependent indexes/views first<br>• Lock table during ALTER |
| 22 | MONEY type — why avoid? | MONEY, avoid | • Locale-dependent — different formatting in different regions<br>• Loss of precision in calculations<br>• Poor portability across databases<br>• **Use NUMERIC(12,2) instead** for currency |
| 23 | Geospatial data (PostGIS)? | PostGIS, geospatial | • `CREATE EXTENSION postgis` — adds geometry/geography types<br>• `GEOMETRY(Point, 4326)` — point with SRID<br>• `ST_Distance()`, `ST_Contains()`, `ST_Within()` — spatial functions<br>• GiST index for spatial queries |
| 24 | hstore extension vs JSONB? | hstore, JSONB | • hstore: simple key-value (text→text only), older<br>• JSONB: nested, typed, supports arrays/objects<br>• JSONB is more feature-rich and preferred<br>• hstore is slightly faster for simple lookups |
| 25 | How to choose the right data type? | Data type, choice | • Use smallest type that fits (saves storage, faster queries)<br>• NUMERIC for money; INT for counts; TIMESTAMPTZ for times<br>• TEXT over VARCHAR in PG (same performance)<br>• Match types for join columns; avoid implicit casts |


## Section 7: Date & Time Operations

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | NOW() vs CURRENT_TIMESTAMP vs clock_timestamp()? | NOW, CURRENT_TIMESTAMP | • `NOW()` = `CURRENT_TIMESTAMP` — transaction start time<br>• `clock_timestamp()` — actual wall-clock time (changes within transaction)<br>• Use NOW() for consistency within a transaction<br>• clock_timestamp() for benchmarking |
| 2 | EXTRACT vs DATE_PART? | EXTRACT, DATE_PART | • `EXTRACT(YEAR FROM date)` — SQL standard<br>• `DATE_PART('year', date)` — PostgreSQL function<br>• Both return the same result<br>• EXTRACT returns numeric; DATE_PART returns double precision |
| 3 | DATE_TRUNC — truncate to month, week, quarter? | DATE_TRUNC, truncate | • `DATE_TRUNC('month', timestamp)` → first day of month<br>• `DATE_TRUNC('week', timestamp)` → Monday of that week<br>• `DATE_TRUNC('quarter', timestamp)` → first day of quarter<br>• Essential for time-based grouping |
| 4 | Difference between two dates/timestamps? | Date diff, AGE, subtract | • `date2 - date1` → integer (days)<br>• `timestamp2 - timestamp1` → interval<br>• `AGE(date2, date1)` → interval ('1 year 2 months 3 days')<br>• `EXTRACT(EPOCH FROM interval)` → seconds |
| 5 | AGE() function — calculate someone's age? | AGE, birthdate | • `AGE(CURRENT_DATE, birthdate)` → interval<br>• `EXTRACT(YEAR FROM AGE(NOW(), birthdate))` → age in years<br>• Handles leap years and month lengths<br>• Accounts for partial years correctly |
| 6 | Add/subtract intervals from dates? | INTERVAL, add, subtract | • `NOW() + INTERVAL '7 days'` — add 7 days<br>• `NOW() - INTERVAL '3 months'` — subtract 3 months<br>• `date + INTEGER` — add days (integer = days)<br>• Chain: `NOW() + INTERVAL '1 year 6 months'` |
| 7 | Format dates for display (TO_CHAR)? | TO_CHAR, format | • `TO_CHAR(timestamp, 'YYYY-MM-DD HH24:MI:SS')`<br>• `TO_CHAR(date, 'Day, DD Mon YYYY')` → 'Monday, 01 Jan 2024'<br>• `TO_CHAR(12345.67, '99,999.99')` — also for numbers<br>• PG-specific; use client-side formatting when possible |
| 8 | Parse string to date (TO_DATE, TO_TIMESTAMP)? | TO_DATE, TO_TIMESTAMP, parse | • `TO_DATE('2024-01-15', 'YYYY-MM-DD')` → date<br>• `TO_TIMESTAMP('2024-01-15 10:30', 'YYYY-MM-DD HH24:MI')` → timestamp<br>• `'2024-01-15'::date` — simpler cast for ISO format<br>• Strict parsing: rejects invalid dates |
| 9 | Convert between timezones (AT TIME ZONE)? | AT TIME ZONE, convert | • `timestamp AT TIME ZONE 'UTC'` — convert to UTC<br>• `timestamptz AT TIME ZONE 'US/Eastern'` — convert to Eastern<br>• Returns timestamp WITHOUT tz from timestamptz<br>• Returns timestamptz from timestamp WITHOUT tz |
| 10 | First and last day of current month? | Month, first day, last day | • First: `DATE_TRUNC('month', CURRENT_DATE)` → 2024-01-01<br>• Last: `DATE_TRUNC('month', CURRENT_DATE) + INTERVAL '1 month' - INTERVAL '1 day'`<br>• Or: `(DATE_TRUNC('month', CURRENT_DATE + INTERVAL '1 month') - INTERVAL '1 day')::date` |
| 11 | generate_series() for dates? | generate_series, date | • `SELECT generate_series('2024-01-01'::date, '2024-12-31', '1 day')` — every day<br>• `generate_series(..., ..., '1 month')` — every month<br>• Used for: calendar tables, gap filling, reports<br>• PG-specific function |
| 12 | Records from last 30 days, quarter, YTD? | Last 30 days, YTD | • Last 30 days: `WHERE created_at >= NOW() - INTERVAL '30 days'`<br>• This quarter: `WHERE created_at >= DATE_TRUNC('quarter', NOW())`<br>• YTD: `WHERE created_at >= DATE_TRUNC('year', NOW())`<br>• Use TIMESTAMPTZ for accuracy |
| 13 | Group by week, month, quarter? | Group by, DATE_TRUNC | • `GROUP BY DATE_TRUNC('week', created_at)` — weekly<br>• `GROUP BY DATE_TRUNC('month', created_at)` — monthly<br>• `GROUP BY DATE_TRUNC('quarter', created_at)` — quarterly<br>• Combine with COUNT/SUM for time-series reports |
| 14 | Epoch time — convert to/from timestamp? | Epoch, EXTRACT EPOCH | • To epoch: `EXTRACT(EPOCH FROM timestamp)` → seconds<br>• From epoch: `TO_TIMESTAMP(epoch_seconds)` → timestamptz<br>• Epoch = seconds since 1970-01-01 00:00:00 UTC<br>• Useful for JavaScript/Unix interoperability |
| 15 | Daylight saving time issues? | DST, timezone, pitfall | • Use TIMESTAMPTZ to auto-handle DST transitions<br>• `INTERVAL '24 hours'` ≠ `INTERVAL '1 day'` during DST changes<br>• Store in UTC, display in user's timezone<br>• Test queries around DST boundaries |
| 16 | Day of week / week number? | Day of week, EXTRACT DOW | • `EXTRACT(DOW FROM date)` — 0=Sunday to 6=Saturday<br>• `EXTRACT(ISODOW FROM date)` — 1=Monday to 7=Sunday<br>• `EXTRACT(WEEK FROM date)` — ISO week number<br>• `TO_CHAR(date, 'Day')` — day name |
| 17 | Business days between two dates? | Business days, workdays | • No built-in function — use generate_series<br>• `SELECT COUNT(*) FROM generate_series(d1, d2, '1 day') d WHERE EXTRACT(DOW FROM d) NOT IN (0, 6)`<br>• Add holiday table exclusion for accuracy<br>• Or use `isfinite()` and custom function |
| 18 | Date-based partitions? | Partition, date, RANGE | • `CREATE TABLE orders (...) PARTITION BY RANGE (order_date)`<br>• `CREATE TABLE orders_2024 PARTITION OF orders FOR VALUES FROM ('2024-01-01') TO ('2025-01-01')`<br>• Partition pruning: planner skips irrelevant partitions<br>• Essential for time-series data |
| 19 | Common date/time pitfalls? | Date pitfalls, common errors | • BETWEEN with timestamps misses end-of-day records<br>• Mixing DATE and TIMESTAMP comparisons<br>• Using TIMESTAMP WITHOUT TZ for global apps<br>• Not accounting for DST in interval arithmetic |
| 20 | Time-series data — efficient storage and query? | Time-series, storage | • Use TIMESTAMPTZ + BRIN index (tiny, fast for ordered data)<br>• Partition by time range (month/year)<br>• Use TimescaleDB extension for advanced features<br>• Aggregate with DATE_TRUNC + GROUP BY |


## Section 8: String Operations & Pattern Matching

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | Concatenation: &#124;&#124; vs CONCAT()? | Concatenate, string | • `'Hello' &#124;&#124; ' ' &#124;&#124; 'World'` — PG concatenation operator<br>• `CONCAT(a, b, c)` — function form, skips NULLs<br>• `&#124;&#124;` returns NULL if any operand is NULL<br>• CONCAT is safer with NULL values |
| 2 | Extract substrings (SUBSTRING, LEFT, RIGHT)? | SUBSTRING, LEFT, RIGHT | • `SUBSTRING(str FROM 2 FOR 5)` — extract 5 chars starting at position 2<br>• `LEFT(str, 3)` — first 3 characters<br>• `RIGHT(str, 3)` — last 3 characters<br>• 1-indexed in PostgreSQL |
| 3 | SPLIT_PART() — split by delimiter? | SPLIT_PART, delimiter | • `SPLIT_PART('a.b.c', '.', 2)` → 'b' (2nd part)<br>• Returns empty string if part doesn't exist<br>• 1-indexed (first part is 1)<br>• PG-specific; MySQL uses SUBSTRING_INDEX |
| 4 | STRING_AGG() vs GROUP_CONCAT? | STRING_AGG, GROUP_CONCAT | • PG: `STRING_AGG(name, ', ' ORDER BY name)` — concat grouped values<br>• MySQL: `GROUP_CONCAT(name ORDER BY name SEPARATOR ', ')`<br>• Both aggregate strings with a delimiter<br>• Same purpose, different syntax |
| 5 | Regex in PostgreSQL (~, ~*, REGEXP_REPLACE)? | Regex, ~, REGEXP | • `col ~ '^[A-Z]'` — case-sensitive regex match<br>• `col ~* '^[a-z]'` — case-insensitive<br>• `REGEXP_REPLACE(str, pattern, replacement)` — substitute<br>• `REGEXP_MATCHES(str, pattern)` — extract matches |
| 6 | LIKE vs ILIKE vs regex? | LIKE, ILIKE, regex | • LIKE: simple patterns (`%`, `_`), case-sensitive<br>• ILIKE: case-insensitive LIKE (PG only)<br>• Regex: full pattern language (`~`, `~*`)<br>• LIKE is fastest; regex most powerful |
| 7 | Find and replace text in a column? | REPLACE, find, substitute | • `UPDATE t SET col = REPLACE(col, 'old', 'new')`<br>• `REGEXP_REPLACE(col, 'pattern', 'replacement')` — regex<br>• `TRANSLATE(str, from_chars, to_chars)` — char-by-char<br>• REPLACE is for literal strings; REGEXP for patterns |
| 8 | TRIM, LTRIM, RTRIM — stripping characters? | TRIM, LTRIM, whitespace | • `TRIM('  hello  ')` → 'hello' (both sides)<br>• `LTRIM(str)` — left side only<br>• `RTRIM(str)` — right side only<br>• `TRIM(BOTH '*' FROM '**hello**')` → 'hello' |
| 9 | LPAD and RPAD — padding strings? | LPAD, RPAD, padding | • `LPAD('42', 5, '0')` → '00042' (left-pad with zeros)<br>• `RPAD('hello', 10, '.')` → 'hello.....'<br>• Useful for formatting IDs, codes<br>• If string already longer, it's truncated |
| 10 | POSITION() / STRPOS() — find substring? | POSITION, STRPOS | • `POSITION('world' IN 'hello world')` → 7 (SQL standard)<br>• `STRPOS('hello world', 'world')` → 7 (PG function)<br>• Returns 0 if not found<br>• 1-indexed |
| 11 | Reverse a string? | REVERSE, string | • `REVERSE('hello')` → 'olleh'<br>• Works with any string type<br>• Useful for checking palindromes<br>• Simple built-in function in PG |
| 12 | Count occurrences of substring? | Count, substring, occurrences | • No built-in function — use formula<br>• `(LENGTH(str) - LENGTH(REPLACE(str, sub, ''))) / LENGTH(sub)`<br>• Or `REGEXP_MATCHES(str, pattern, 'g')` and count results<br>• Works for any substring |
| 13 | INITCAP, UPPER, LOWER? | INITCAP, UPPER, LOWER | • `UPPER('hello')` → 'HELLO'<br>• `LOWER('HELLO')` → 'hello'<br>• `INITCAP('hello world')` → 'Hello World'<br>• Locale-aware for internationalized text |
| 14 | Character encoding conversion? | Encoding, CONVERT | • `CONVERT(bytes, src_encoding, dest_encoding)`<br>• `ENCODE(data, 'base64')` / `DECODE(text, 'base64')`<br>• Database encoding set at creation (`CREATE DATABASE ... ENCODING 'UTF8'`)<br>• Most modern DBs use UTF-8 by default |
| 15 | LENGTH vs CHAR_LENGTH vs OCTET_LENGTH? | LENGTH, CHAR_LENGTH, bytes | • `CHAR_LENGTH`: counts characters<br>• `LENGTH`: same as CHAR_LENGTH in PG<br>• `OCTET_LENGTH`: counts bytes<br>• 'café': CHAR_LENGTH=4, OCTET_LENGTH=5 (é=2 bytes) |
| 16 | Case-insensitive searches efficiently? | Case insensitive, CITEXT | • `WHERE LOWER(col) = 'value'` — with expression index<br>• `ILIKE '%value%'` — PG specific<br>• `CITEXT` extension — case-insensitive text type<br>• Expression index: `CREATE INDEX idx ON t(LOWER(col))` |
| 17 | Extract email domains, file extensions, URL parts? | Extract, SPLIT_PART | • Domain: `SPLIT_PART(email, '@', 2)` → 'gmail.com'<br>• Extension: `REGEXP_REPLACE(filename, '.*\.', '')` → 'pdf'<br>• URL: `SPLIT_PART(url, '/', 3)` for domain<br>• Or use regex: `REGEXP_MATCHES(url, pattern)` |
| 18 | TRANSLATE() vs REPLACE()? | TRANSLATE, REPLACE | • `REPLACE('hello', 'ell', 'a')` → 'hao' (substring replacement)<br>• `TRANSLATE('hello', 'helo', 'HELO')` → 'HELLO' (char-by-char)<br>• TRANSLATE maps single chars; REPLACE maps substrings<br>• TRANSLATE: `TRANSLATE('abc', 'abc', '123')` → '123' |
| 19 | Multi-byte/Unicode in string operations? | Unicode, UTF-8 | • Use CHAR_LENGTH for character count (not OCTET_LENGTH)<br>• Collation affects sorting: `COLLATE "en_US.utf8"`<br>• `normalize()` for Unicode normalization (PG 13+)<br>• Test with emojis and CJK characters |
| 20 | SIMILAR TO vs LIKE vs regex? | SIMILAR TO, pattern | • `LIKE`: simplest (`%`, `_`), fastest<br>• `SIMILAR TO`: SQL standard regex subset (blends LIKE and regex)<br>• `~` / `~*`: full POSIX regex (most powerful)<br>• Prefer LIKE for simple; `~` for complex patterns |


## Section 9: JSON/JSONB Operations

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | -> vs ->> operators? | ->, ->>, JSON access | • `col->'key'` returns JSON object/array (keeps type)<br>• `col->>'key'` returns text (extracted value)<br>• `->` for chaining: `col->'address'->'city'`<br>• `->>` for final value: `col->>'name'` = text |
| 2 | Access nested values (#> and #>>)? | #>, #>>, nested path | • `col#>'{address,city}'` — navigate nested path, returns JSON<br>• `col#>>'{address,city}'` — returns text<br>• Path is array of keys<br>• Alternative to chaining `->` operators |
| 3 | Query rows by JSONB value? | JSONB, WHERE, query | • `WHERE col->>'status' = 'active'`<br>• `WHERE (col->>'age')::int > 30` — cast text to int<br>• `WHERE col @> '{"status":"active"}'` — containment<br>• Index with GIN for performance |
| 4 | @> containment operator? | @>, containment, JSONB | • `col @> '{"key":"value"}'` — left contains right<br>• Works with nested structures<br>• GIN index supports `@>` efficiently<br>• Best way to filter JSONB by key-value |
| 5 | ? operator for key existence? | ?, key existence, JSONB | • `col ? 'key'` — does top-level key exist?<br>• `col ?& ARRAY['k1','k2']` — ALL keys exist<br>• `col ?&#124; ARRAY['k1','k2']` — ANY key exists<br>• Supported by GIN index |
| 6 | Build JSON from query results? | JSON_BUILD_OBJECT, JSON_AGG | • `JSON_BUILD_OBJECT('name', col1, 'age', col2)` — from columns<br>• `JSON_AGG(row)` — aggregate rows into JSON array<br>• `JSONB_BUILD_ARRAY(1, 2, 3)` — build array<br>• Combine for nested API responses |
| 7 | Update a JSONB key (JSONB_SET)? | JSONB_SET, update key | • `UPDATE t SET col = JSONB_SET(col, '{key}', '"new_value"')`<br>• Path: `'{address,city}'` for nested keys<br>• 4th arg `true` = create if missing<br>• Returns new JSONB (immutable update) |
| 8 | Delete a key from JSONB? | Delete key, - operator | • `col - 'key'` — remove top-level key<br>• `col #- '{address,city}'` — remove nested key<br>• `col - 0` — remove first array element<br>• Returns new JSONB value |
| 9 | Insert new key into JSONB? | Insert, JSONB, add key | • `UPDATE t SET col = col &#124;&#124; '{"new_key":"value"}'`<br>• Or `JSONB_SET(col, '{new_key}', '"value"', true)`<br>• `&#124;&#124;` merges/overwrites at top level<br>• JSONB_SET for nested insertion |
| 10 | Expand JSONB to rows? | JSONB_EACH, ARRAY_ELEMENTS | • `JSONB_EACH(col)` → (key, value) rows from object<br>• `JSONB_ARRAY_ELEMENTS(col)` → one row per array element<br>• `JSONB_EACH_TEXT` → text values instead of JSONB<br>• Used in FROM clause or LATERAL join |
| 11 | Best way to index JSONB? GIN indexes? | GIN, JSONB, index | • `CREATE INDEX idx ON t USING GIN(col)` — indexes all keys/values<br>• `CREATE INDEX idx ON t USING GIN(col jsonb_path_ops)` — containment only<br>• `jsonb_path_ops` is smaller and faster for `@>`<br>• Expression index for specific keys: `CREATE INDEX idx ON t((col->>'status'))` |
| 12 | jsonb_path_query (SQL/JSON path)? | jsonb_path_query, path | • PG 12+: `jsonb_path_query(col, '$.items[*].name')`<br>• SQL/JSON standard path language<br>• Supports filters: `'$.items[*] ? (@.price > 10)'`<br>• More powerful than `->` operators for complex access |
| 13 | Merge two JSONB objects? | Merge, JSONB, concatenate | • `jsonb1 &#124;&#124; jsonb2` — shallow merge (top-level)<br>• Later keys override earlier ones<br>• For deep merge: use `jsonb_deep_merge` custom function<br>• `&#124;&#124;` doesn't recursively merge nested objects |
| 14 | Convert JSONB to relational rows and back? | JSONB, relational, convert | • To rows: `JSONB_EACH`, `JSONB_ARRAY_ELEMENTS`, `JSONB_TO_RECORD`<br>• To JSON: `ROW_TO_JSON(row)`, `JSON_AGG(row)`<br>• `JSONB_POPULATE_RECORD(null::my_type, jsonb)` — map to type<br>• Useful for ETL and API layers |
| 15 | JSONB vs normalized schema — when each? | JSONB, normalized, choice | • JSONB: varying attributes, dynamic schema, prototyping<br>• Normalized: core entities, relationships, integrity<br>• JSONB: metadata, settings, feature flags<br>• Normalized: users, orders, products |
| 16 | Validate JSONB structure? | Validate, JSONB, CHECK | • CHECK constraint: `CHECK (col ? 'required_key')`<br>• PG 17+: `JSON_SCHEMA_VALID` (pending standard)<br>• Application-layer validation (JSON Schema)<br>• Trigger-based validation for complex rules |
| 17 | Performance: -> vs ->> operators? | Performance, ->, ->> | • Both are fast for top-level access<br>• `->>` does an extra text conversion<br>• `->` keeps JSONB type (needed for chaining)<br>• For indexed lookups, `@>` is fastest |
| 18 | Aggregate relational data into nested JSON? | Nested JSON, JSON_AGG | • `SELECT u.name, JSON_AGG(o.*) as orders FROM users u JOIN orders o ON ... GROUP BY u.name`<br>• Builds nested JSON from JOINed data<br>• Use `JSON_BUILD_OBJECT` for custom structure<br>• Perfect for building API responses in SQL |
| 19 | JSONB arrays in WHERE clauses? | JSONB array, WHERE | • `WHERE col @> '["value"]'` — array contains value<br>• `WHERE EXISTS (SELECT 1 FROM JSONB_ARRAY_ELEMENTS_TEXT(col) e WHERE e = 'value')`<br>• `WHERE col ? 'value'` — works for array elements too<br>• GIN index optimizes array containment |
| 20 | Expression index on specific JSONB keys? | Expression index, JSONB | • `CREATE INDEX idx ON t((col->>'status'))` — index specific key<br>• Supports `WHERE col->>'status' = 'active'` queries<br>• `CREATE INDEX idx ON t((col->>'price')::numeric)` — typed<br>• Much smaller than full GIN index |


## Section 10: Aggregation & GROUP BY

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | COUNT(*) vs COUNT(col) vs COUNT(DISTINCT col)? | COUNT, NULL, distinct | • `COUNT(*)`: counts ALL rows (including NULLs)<br>• `COUNT(col)`: counts non-NULL values only<br>• `COUNT(DISTINCT col)`: unique non-NULL values<br>• `COUNT(1)` ≡ `COUNT(*)` |
| 2 | How aggregates handle NULLs? | Aggregate, NULL handling | • SUM, AVG, MIN, MAX all **ignore NULLs**<br>• `AVG(10, 20, NULL)` = 15 (not 10)<br>• Only `COUNT(*)` includes NULLs<br>• Use COALESCE(col, 0) if NULLs should be 0 |
| 3 | WHERE vs HAVING for filtering? | WHERE, HAVING, aggregate | • WHERE: filters rows BEFORE grouping<br>• HAVING: filters groups AFTER aggregation<br>• `HAVING COUNT(*) > 5` — can't do with WHERE<br>• WHERE is faster (reduces data before grouping) |
| 4 | GROUP BY multiple columns? | GROUP BY, multiple | • `GROUP BY dept, year` — one group per unique (dept, year) pair<br>• Single column: one group per unique dept value<br>• More columns = more granular groups<br>• All non-aggregated SELECT columns must be in GROUP BY |
| 5 | ROLLUP for subtotals and grand totals? | ROLLUP, subtotals | • `GROUP BY ROLLUP(year, month)` → (year,month), (year), () groups<br>• Creates hierarchical subtotals<br>• NULL in result = aggregated dimension<br>• () = grand total |
| 6 | CUBE — all combinations? | CUBE, cross-tab | • `GROUP BY CUBE(a, b)` → (a,b), (a), (b), () groups<br>• All possible combinations of grouping<br>• More rows than ROLLUP<br>• Used for cross-tabulation reports |
| 7 | GROUPING SETS? | GROUPING SETS, custom | • `GROUP BY GROUPING SETS ((dept), (year), ())` — specific groups<br>• More control than ROLLUP or CUBE<br>• Replaces multiple UNION ALL queries<br>• Choose exactly which groupings you need |
| 8 | GROUPING() function? | GROUPING, subtotal flag | • `GROUPING(col)` returns 1 if col is aggregated (NULL = subtotal)<br>• 0 if col is a regular group value<br>• Distinguishes real NULLs from subtotal NULLs<br>• Used with ROLLUP/CUBE/GROUPING SETS |
| 9 | STRING_AGG for concatenating grouped strings? | STRING_AGG, concat | • `STRING_AGG(name, ', ' ORDER BY name)` → 'Alice, Bob, Charlie'<br>• Aggregates text values with delimiter<br>• ORDER BY inside controls string order<br>• PG-specific; MySQL uses GROUP_CONCAT |
| 10 | ARRAY_AGG for grouped values? | ARRAY_AGG, array | • `ARRAY_AGG(name ORDER BY name)` → `{Alice,Bob,Charlie}`<br>• Collects values into an array<br>• `ARRAY_AGG(DISTINCT name)` — unique values<br>• Useful for JSON APIs and array operations |
| 11 | JSON_AGG and JSONB_AGG? | JSON_AGG, JSONB_AGG | • `JSON_AGG(row)` — aggregate rows into JSON array<br>• `JSONB_AGG(col)` — values into JSONB array<br>• Combined with JSON_BUILD_OBJECT for nested JSON<br>• Perfect for building API responses |
| 12 | FILTER clause in aggregates? | FILTER, conditional agg | • `COUNT(*) FILTER (WHERE status = 'active')` — count only active<br>• Cleaner than `SUM(CASE WHEN ... THEN 1 END)`<br>• Works with any aggregate: SUM, AVG, COUNT, etc.<br>• PG-specific feature |
| 13 | CASE inside aggregates (conditional aggregation)? | CASE, conditional, SUM | • `SUM(CASE WHEN status='active' THEN 1 ELSE 0 END)`<br>• Cross-database compatible (works everywhere)<br>• Creates pivot-like output in a single query<br>• PG's FILTER is cleaner alternative |
| 14 | Top-N per group (most asked aggregate Q)? | Top N, ROW_NUMBER, group | • `WITH r AS (SELECT *, ROW_NUMBER() OVER (PARTITION BY dept ORDER BY salary DESC) rn FROM emp) SELECT * FROM r WHERE rn <= 3`<br>• Gets top 3 per department<br>• Use DENSE_RANK for ties<br>• Window function + filter pattern |
| 15 | Running totals with aggregates + window? | Running total, SUM OVER | • `SUM(amount) OVER (ORDER BY date)` — running total<br>• `SUM(amount) OVER (PARTITION BY dept ORDER BY date)` — per dept<br>• Cumulative sum without GROUP BY<br>• Window function preserves individual rows |
| 16 | Percentages within groups? | Percentage, group | • `COUNT(*) * 100.0 / SUM(COUNT(*)) OVER ()` — pct of total<br>• `COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (PARTITION BY dept)` — pct within dept<br>• Use window function over aggregate<br>• CAST to numeric for decimal percentages |
| 17 | SUM() vs SUM() OVER — aggregate vs window? | SUM, window, aggregate | • `SUM(x)`: collapses rows via GROUP BY → one row per group<br>• `SUM(x) OVER()`: keeps all rows, adds running total column<br>• Aggregate: reduces rows; Window: preserves rows<br>• Can use both in same query |
| 18 | COUNT DISTINCT across multiple columns? | COUNT DISTINCT, multiple | • No direct `COUNT(DISTINCT a, b)` in standard SQL<br>• Workaround: `COUNT(DISTINCT (a, b))` — PG tuple syntax<br>• Or: `COUNT(DISTINCT a &#124;&#124; ',' &#124;&#124; b)` — concatenate<br>• Or: subquery with SELECT DISTINCT + COUNT |
| 19 | Find median in SQL? | Median, percentile | • `PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY col)` — exact median<br>• `PERCENTILE_DISC(0.5) WITHIN GROUP (ORDER BY col)` — discrete value<br>• PG-specific ordered-set aggregate<br>• Standard SQL doesn't have built-in median |
| 20 | Moving averages? | Moving average, window | • `AVG(price) OVER (ORDER BY date ROWS BETWEEN 6 PRECEDING AND CURRENT ROW)` — 7-day MA<br>• Window frame defines range<br>• ROWS for count-based; RANGE for value-based<br>• Essential for time-series analysis |
| 21 | MODE() aggregate function? | MODE, most frequent | • `MODE() WITHIN GROUP (ORDER BY col)` — most frequent value<br>• PG-specific ordered-set aggregate<br>• Returns one value even if multiple modes exist<br>• Alternative: GROUP BY + COUNT + LIMIT 1 |
| 22 | Aggregate with nested subqueries? | Aggregate, subquery | • `SELECT dept, (SELECT MAX(salary) FROM emp e2 WHERE e2.dept = e1.dept) FROM emp e1`<br>• Scalar subquery in SELECT list<br>• Better: GROUP BY with aggregate function<br>• Correlated subqueries in aggregates are slow |
| 23 | COUNT(DISTINCT) performance on large data? | COUNT DISTINCT, performance | • COUNT(DISTINCT) is expensive — sorts all values<br>• Consider: HyperLogLog extension for approximate counts<br>• Pre-aggregate in materialized views<br>• Batch processing for very large datasets |
| 24 | Handle ties in max/min per group? | Ties, max, group | • `ROW_NUMBER`: picks arbitrary one (no ties)<br>• `RANK`/`DENSE_RANK`: returns all tied rows<br>• Use RANK + filter WHERE rank = 1 to get all top values<br>• Choose based on business requirement |
| 25 | Pivot data with conditional aggregation? | Pivot, cross-tab | • `SELECT dept, SUM(CASE WHEN year=2023 THEN revenue END) AS y2023, SUM(CASE WHEN year=2024 THEN revenue END) AS y2024 FROM ...`<br>• PG: `tablefunc` extension `crosstab()`<br>• Create columns from row values<br>• Dynamic pivots need dynamic SQL |


## Section 11: Subqueries

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | What is a scalar subquery? | Scalar, subquery | • Returns exactly ONE value (one row, one column)<br>• Used in SELECT list or WHERE comparisons<br>• `SELECT name, (SELECT COUNT(*) FROM orders WHERE user_id = u.id) FROM users u`<br>• Errors if subquery returns more than one row |
| 2 | Correlated subquery — why slow? | Correlated, performance | • References outer query — runs once per outer row<br>• N outer rows = N subquery executions<br>• `WHERE salary > (SELECT AVG(salary) FROM emp WHERE dept = e.dept)`<br>• Often rewritable as JOIN for better performance |
| 3 | Rewrite correlated subquery as JOIN? | Rewrite, JOIN, optimize | • Correlated: `WHERE salary > (SELECT AVG(salary) FROM emp e2 WHERE e2.dept = e1.dept)`<br>• JOIN version: `JOIN (SELECT dept, AVG(salary) avg_s FROM emp GROUP BY dept) a ON e.dept = a.dept WHERE e.salary > a.avg_s`<br>• JOIN runs aggregation once |
| 4 | Subquery in WHERE vs JOIN? | Subquery, JOIN, WHERE | • Subquery: when checking existence or comparing to single value<br>• JOIN: when you need columns from both tables<br>• EXISTS/IN subquery for filtering<br>• Optimizer often turns subqueries into joins |
| 5 | Subquery in FROM (derived table)? | Derived table, FROM | • `SELECT * FROM (SELECT dept, AVG(salary) FROM emp GROUP BY dept) AS dept_avg`<br>• Must have an alias<br>• Evaluated once before outer query<br>• Alternative to CTE for simple cases |
| 6 | Subquery in SELECT clause? | SELECT, scalar subquery | • `SELECT name, (SELECT dept_name FROM depts WHERE id = e.dept_id) FROM emp e`<br>• Must return single value (scalar)<br>• Correlated — references outer row<br>• Alternative: LEFT JOIN |
| 7 | IN vs EXISTS with subquery? | IN, EXISTS, subquery | • `IN`: builds list, compares value against it<br>• `EXISTS`: checks if subquery returns any rows<br>• EXISTS stops at first match (faster for large subqueries)<br>• IN can use index on outer table effectively |
| 8 | NOT IN fails with NULL — NOT EXISTS is safe? | NOT IN, NULL, NOT EXISTS | • `NOT IN (1, 2, NULL)` returns NO rows (three-value logic)<br>• NULL poisons entire NOT IN list<br>• `NOT EXISTS` handles NULLs correctly<br>• **Always prefer NOT EXISTS over NOT IN** |
| 9 | ANY operator with subquery vs IN? | ANY, IN, difference | • `= ANY(subquery)` is equivalent to `IN(subquery)`<br>• `> ANY(subquery)` — greater than at least one value<br>• ANY supports all comparison operators<br>• IN only supports equality |
| 10 | ALL operator? Practical example? | ALL, comparison | • `WHERE salary > ALL (SELECT salary FROM managers)` — higher than every manager<br>• `= ALL` — equals every value (rare)<br>• Empty subquery: ALL returns TRUE<br>• Less common than ANY/IN |
| 11 | Second highest salary with subquery? | Second highest, subquery | • `SELECT MAX(salary) FROM emp WHERE salary < (SELECT MAX(salary) FROM emp)`<br>• Or: `SELECT DISTINCT salary FROM emp ORDER BY salary DESC OFFSET 1 LIMIT 1`<br>• Or: `DENSE_RANK() OVER (ORDER BY salary DESC)` then filter |
| 12 | Row-by-row comparison between tables? | Compare, subquery | • `SELECT a.* FROM t1 a WHERE EXISTS (SELECT 1 FROM t2 b WHERE a.id = b.id AND a.val != b.val)`<br>• Finds rows that differ between tables<br>• Also: `EXCEPT` for set difference<br>• Full comparison: also check for rows in t2 not in t1 |
| 13 | Filter based on aggregated data? | Aggregate, filter, subquery | • `WHERE dept_id IN (SELECT dept_id FROM emp GROUP BY dept_id HAVING COUNT(*) > 10)`<br>• Subquery computes aggregate, outer filters by it<br>• Alternative: CTE + JOIN<br>• HAVING in subquery filters groups |
| 14 | Multi-row vs single-row subquery? | Multi-row, single-row | • Single-row: returns one value (use =, >, <)<br>• Multi-row: returns multiple values (use IN, ANY, ALL)<br>• Wrong operator → runtime error<br>• Scalar subquery must return exactly one row |
| 15 | Subqueries with INSERT, UPDATE, DELETE? | DML, subquery | • `INSERT INTO t2 SELECT * FROM t1 WHERE ...`<br>• `UPDATE t1 SET col = (SELECT val FROM t2 WHERE t2.id = t1.id)`<br>• `DELETE FROM t1 WHERE id IN (SELECT id FROM t2 WHERE ...)`<br>• Combine DML with any subquery type |
| 16 | Performance: correlated subquery vs JOIN? | Performance, correlated | • Correlated: O(n²) — runs per outer row<br>• JOIN: O(n log n) or O(n) with hash join<br>• Optimizer may rewrite correlated → JOIN automatically<br>• If not, manually rewrite for better performance |
| 17 | How optimizer handles subqueries? | Optimizer, subquery | • May inline subquery into main query<br>• May convert IN to semi-join<br>• May cache correlated subquery results<br>• Check `EXPLAIN ANALYZE` to see actual execution |
| 18 | Subqueries for running calculations? | Running, subquery | • `SELECT *, (SELECT SUM(amount) FROM orders o2 WHERE o2.date <= o1.date) FROM orders o1`<br>• Correlated subquery for running total<br>• **Better: window function** `SUM(amount) OVER (ORDER BY date)`<br>• Window function is faster and cleaner |
| 19 | Subquery unnesting/flattening? | Unnesting, flattening | • Optimizer converts subquery into JOIN (flattening)<br>• Removes nested execution, allows better join strategies<br>• Not always possible (correlated with side effects)<br>• Check EXPLAIN to verify unnesting happened |
| 20 | When CTE instead of subquery? | CTE, subquery, choice | • CTE: named, readable, reusable within query<br>• Subquery: inline, single-use<br>• CTE for: recursive queries, multiple references, complex logic<br>• Subquery for: simple inline filters |


## Section 12: Common Table Expressions (CTEs)

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | What is a CTE? Readability benefit? | CTE, WITH, readability | • Named temporary result set: `WITH cte AS (SELECT ...)`<br>• Scoped to single query (disappears after)<br>• Makes complex queries more readable than nested subqueries<br>• Think of it as a "named inline view" |
| 2 | CTE syntax with WITH? | WITH, syntax | • `WITH cte_name AS (SELECT ...) SELECT * FROM cte_name`<br>• Can join CTE with other tables<br>• Can chain: `WITH a AS (...), b AS (...) SELECT ...`<br>• Must appear at beginning of query |
| 3 | Multiple CTEs in one query? | Multiple CTEs, chain | • `WITH a AS (...), b AS (...), c AS (...) SELECT * FROM c`<br>• Later CTEs can reference earlier ones<br>• Order matters — reference only preceding CTEs<br>• Great for step-by-step transformations |
| 4 | CTE vs subquery — when each? | CTE, subquery, choice | • CTE: named, reusable, recursive capability<br>• Subquery: inline, single-use, simpler<br>• CTE when referenced multiple times in query<br>• Subquery when used only once and simple |
| 5 | CTE vs temporary table? | CTE, temp table | • CTE: single-query scope, no DDL, auto-cleaned<br>• Temp table: session scope, indexable, physical<br>• CTE for simple transformations<br>• Temp table for large intermediate results needing indexes |
| 6 | What is a recursive CTE? Structure? | Recursive CTE, anchor | • `WITH RECURSIVE cte AS (anchor_query UNION ALL recursive_query)`<br>• Anchor: base case (non-recursive)<br>• Recursive: references CTE itself<br>• Terminates when recursive part returns no rows |
| 7 | Traverse hierarchy with recursive CTE? | Hierarchy, org chart, tree | • Anchor: `SELECT * FROM emp WHERE manager_id IS NULL`<br>• Recursive: `SELECT e.* FROM emp e JOIN cte ON e.manager_id = cte.id`<br>• Builds tree level by level<br>• Add `level` column: `1 AS level` in anchor, `level + 1` in recursive |
| 8 | Generate date series with recursive CTE? | Date series, recursive | • Anchor: `SELECT '2024-01-01'::date AS d`<br>• Recursive: `SELECT d + 1 FROM cte WHERE d < '2024-12-31'`<br>• PG has `generate_series()` which is simpler<br>• Recursive CTE works in all databases |
| 9 | Termination condition — what without one? | Termination, infinite | • Without termination: infinite loop → PG kills after max iterations<br>• Set `max_recursion_depth` or check condition<br>• Always include WHERE clause in recursive part<br>• PG default limit protects against runaway queries |
| 10 | Limit recursion depth? | Recursion depth, limit | • Add counter column: `1 AS depth` → `depth + 1`<br>• Filter: `WHERE depth < 10` in recursive part<br>• Or use `CYCLE` detection (PG 14+)<br>• Prevents infinite loops in cyclic data |
| 11 | UNION ALL requirement — can you use UNION? | UNION ALL, UNION, recursive | • UNION ALL is required in most databases<br>• UNION (dedup) works in PG but adds overhead<br>• UNION ALL is standard for recursive CTEs<br>• Use column-based cycle detection instead of UNION |
| 12 | Detect cycles in recursive CTEs? | Cycle detection, recursive | • PG 14+: `CYCLE id SET is_cycle USING path`<br>• Manual: track visited IDs in array, check with `= ANY(path)`<br>• `WHERE id != ALL(path)` prevents revisiting<br>• Essential for graphs with cycles |
| 13 | Writable CTE (INSERT/UPDATE/DELETE in CTE)? | Writable CTE, DML | • `WITH deleted AS (DELETE FROM old_table RETURNING *) INSERT INTO archive SELECT * FROM deleted`<br>• Move data between tables in one statement<br>• `WITH updated AS (UPDATE ... RETURNING *) SELECT * FROM updated`<br>• Very powerful for data migration |
| 14 | CTEs for data migration/cleanup? | Migration, CTE, cleanup | • Delete duplicates: `WITH dups AS (SELECT id, ROW_NUMBER() OVER (...) rn FROM t) DELETE FROM t WHERE id IN (SELECT id FROM dups WHERE rn > 1)`<br>• Batch updates with RETURNING<br>• Archive + delete in one transaction |
| 15 | CTE materialization (PG 12+)? | MATERIALIZED, NOT MATERIALIZED | • `WITH cte AS MATERIALIZED (query)` — compute once, store result<br>• `NOT MATERIALIZED` — allow optimizer to inline<br>• If CTE used multiple times, MATERIALIZED avoids re-computation<br>• PG 12+ changed default behavior |
| 16 | When materialization hurts performance? | Materialization, performance | • Materialized CTE can't push down WHERE filters<br>• If CTE returns many rows but outer query filters most, inlining is better<br>• `NOT MATERIALIZED` lets optimizer push filters into CTE<br>• Check EXPLAIN for actual behavior |
| 17 | CTEs for multi-step transformations? | Multi-step, pipeline | • `WITH raw AS (...), cleaned AS (SELECT ... FROM raw WHERE ...), aggregated AS (SELECT ... FROM cleaned GROUP BY ...) SELECT * FROM aggregated`<br>• Each CTE = one transformation step<br>• Self-documenting pipeline |
| 18 | Reference CTE multiple times? | Multiple reference, CTE | • Yes: `WITH cte AS (...) SELECT * FROM cte a JOIN cte b ON ...`<br>• Referenced twice → may be materialized automatically<br>• Without materialization, PG may run query twice<br>• Explicit `MATERIALIZED` ensures single computation |
| 19 | Scope and lifetime of a CTE? | Scope, lifetime | • Exists only during the execution of the enclosing query<br>• Not visible to other queries or sessions<br>• No cleanup needed — automatic<br>• Cannot be referenced across statements |
| 20 | Debug complex queries with CTEs? | Debug, CTE, step | • Break query into named CTEs for each step<br>• `SELECT * FROM step1_cte` — inspect intermediate results<br>• Comment out later steps to isolate issues<br>• Each CTE can be tested independently |
| 21 | CTE for top-N per group? | Top N, CTE, ROW_NUMBER | • `WITH ranked AS (SELECT *, ROW_NUMBER() OVER (PARTITION BY dept ORDER BY salary DESC) rn FROM emp) SELECT * FROM ranked WHERE rn <= 3`<br>• Clean, readable pattern<br>• Change ROW_NUMBER to DENSE_RANK for ties |
| 22 | CTEs with EXPLAIN ANALYZE? | EXPLAIN, CTE | • CTEs appear as "CTE Scan" nodes in plan<br>• Materialized CTEs show as separate CTE Scan<br>• Non-materialized may be inlined (no CTE node)<br>• Check actual time and rows in each CTE |
| 23 | Window functions inside a CTE? | Window, CTE | • Yes: `WITH r AS (SELECT *, ROW_NUMBER() OVER (...) FROM t) SELECT * FROM r WHERE ...`<br>• Common pattern: compute window + filter in outer query<br>• Can't filter window results without CTE/subquery<br>• Essential for "partition + filter" queries |
| 24 | CTE vs view? | CTE, view, comparison | • CTE: single-query scope, inline, no DDL<br>• View: permanent, reusable, stored in catalog<br>• CTE for: ad-hoc complex queries<br>• View for: recurring queries shared across users |
| 25 | Limitations of CTEs? | CTE, limitations | • Cannot reference CTE across statements<br>• Cannot add indexes to CTE results<br>• Materialization may prevent filter pushdown<br>• Recursive CTEs can be slow without depth limits |


## Section 13: Window Functions

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | Window functions vs aggregate functions? | Window, aggregate, differ | • Aggregates collapse rows (GROUP BY); windows preserve them<br>• `SUM(x)` → one row per group; `SUM(x) OVER()` → every row<br>• Window adds computed column without reducing result set<br>• Can use both in same query |
| 2 | OVER() — empty vs with PARTITION BY? | OVER, PARTITION BY, empty | • `OVER()`: entire result set is one window<br>• `OVER(PARTITION BY col)`: separate window per partition<br>• `COUNT(*) OVER()` → total count on every row<br>• `COUNT(*) OVER(PARTITION BY dept)` → per-dept count |
| 3 | PARTITION BY vs GROUP BY? | PARTITION BY, GROUP BY | • GROUP BY: collapses into one row per group<br>• PARTITION BY: keeps all rows, segments window calculation<br>• GROUP BY for summaries; PARTITION BY for per-row context<br>• Both create groups, different output shape |
| 4 | ORDER BY inside window function? | ORDER BY, window | • Defines the order for the window calculation<br>• Affects running computations: `SUM() OVER (ORDER BY date)` → running total<br>• Without ORDER BY: function sees entire partition<br>• With ORDER BY: default frame is UNBOUNDED PRECEDING to CURRENT ROW |
| 5 | ROW_NUMBER, RANK, DENSE_RANK — difference? | ROW_NUMBER, RANK, DENSE_RANK | • `ROW_NUMBER()`: unique sequential (1,2,3,4) — no ties<br>• `RANK()`: same rank for ties, skips (1,2,2,4)<br>• `DENSE_RANK()`: same rank for ties, no skip (1,2,2,3)<br>• Choice depends on how to handle ties |
| 6 | ROW_NUMBER for top-N per group? | Top N, ROW_NUMBER | • `ROW_NUMBER() OVER (PARTITION BY dept ORDER BY salary DESC) rn`<br>• Wrap in CTE: `WHERE rn <= 3` → top 3 per dept<br>• Ensures exactly N rows per group<br>• Use DENSE_RANK if ties should all be included |
| 7 | ROW_NUMBER to remove duplicates? | Dedup, ROW_NUMBER | • `WITH dups AS (SELECT *, ROW_NUMBER() OVER (PARTITION BY email ORDER BY id) rn FROM users) DELETE FROM users WHERE id IN (SELECT id FROM dups WHERE rn > 1)`<br>• Keeps first (lowest id), deletes rest<br>• Change ORDER BY to pick which duplicate to keep |
| 8 | NTILE(n) for quartiles/percentiles? | NTILE, quartile | • `NTILE(4) OVER (ORDER BY score)` → quartiles (1-4)<br>• `NTILE(100)` → percentiles<br>• Divides rows into n equal-sized buckets<br>• Uneven rows: first buckets get one extra row |
| 9 | LAG() — compare to previous row? | LAG, previous | • `LAG(salary, 1) OVER (ORDER BY date)` — previous row's salary<br>• `LAG(col, 2, default)` — 2 rows back, with default if null<br>• Use for: month-over-month comparison, change detection<br>• Always specify ORDER BY |
| 10 | LEAD() — row-over-row growth? | LEAD, next row, growth | • `LEAD(revenue, 1) OVER (ORDER BY month)` — next month's revenue<br>• Growth: `(LEAD(val) OVER (...) - val) / val * 100`<br>• `LEAD(col, 1, 0) OVER (...)` — default to 0<br>• Opposite of LAG |
| 11 | FIRST_VALUE, LAST_VALUE — gotcha? | FIRST_VALUE, LAST_VALUE | • `FIRST_VALUE(col)`: first value in frame<br>• `LAST_VALUE(col)`: last value — **gotcha**: default frame ends at CURRENT ROW<br>• Fix: `ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING`<br>• Without fix, LAST_VALUE = current row |
| 12 | NTH_VALUE — when to use? | NTH_VALUE, nth | • `NTH_VALUE(col, 3) OVER (ORDER BY col)` — 3rd value<br>• Need explicit frame for reliable results<br>• Alternative: `ROW_NUMBER()` then filter<br>• Less commonly used than LAG/LEAD |
| 13 | Running totals with SUM() OVER()? | Running total, SUM OVER | • `SUM(amount) OVER (ORDER BY date)` — cumulative sum<br>• `PARTITION BY dept ORDER BY date` — per-dept running total<br>• Frame: UNBOUNDED PRECEDING to CURRENT ROW (default)<br>• Works with SUM, COUNT, AVG, etc. |
| 14 | Moving averages with windows? | Moving average, ROWS BETWEEN | • `AVG(price) OVER (ORDER BY date ROWS BETWEEN 6 PRECEDING AND CURRENT ROW)` — 7-day MA<br>• `ROWS BETWEEN 3 PRECEDING AND 3 FOLLOWING` — centered MA<br>• Fixed-count window frame<br>• Essential for time-series analysis |
| 15 | Cumulative percentages? | Cumulative pct, window | • `SUM(amount) OVER (ORDER BY amount DESC) / SUM(amount) OVER () * 100`<br>• Running sum / total * 100<br>• Or use `CUME_DIST()` for rank-based cumulative distribution<br>• Pareto analysis (80/20 rule) |
| 16 | Window frame — ROWS BETWEEN options? | ROWS BETWEEN, frame | • `UNBOUNDED PRECEDING` — from start of partition<br>• `N PRECEDING` / `N FOLLOWING` — N rows before/after<br>• `CURRENT ROW` — current row<br>• `UNBOUNDED FOLLOWING` — to end of partition |
| 17 | ROWS vs RANGE in window frames? | ROWS, RANGE, frame | • `ROWS`: physical rows (exact count)<br>• `RANGE`: logical range based on ORDER BY value<br>• ROWS is more predictable and commonly used<br>• RANGE groups rows with same ORDER BY value |
| 18 | ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW? | Full frame, UNBOUNDED | • Default frame when ORDER BY is specified<br>• Includes all rows from partition start to current<br>• Used for: running totals, cumulative sums<br>• Explicit declaration makes intent clear |
| 19 | ROWS BETWEEN 3 PRECEDING AND 3 FOLLOWING? | Sliding window, centered | • Centered window of 7 rows (3 before, current, 3 after)<br>• Used for: centered moving averages, smoothing<br>• Edge rows have smaller windows (partial)<br>• Frame size adapts at partition boundaries |
| 20 | Default frame when ORDER BY specified? | Default frame, ORDER BY | • `RANGE BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW`<br>• Note: RANGE not ROWS — groups same-value rows together<br>• Without ORDER BY: entire partition is the frame<br>• Explicit ROWS BETWEEN is recommended for clarity |
| 21 | Multiple window functions in same query? | Multiple windows, query | • `SELECT ROW_NUMBER() OVER w, SUM(x) OVER w FROM t WINDOW w AS (PARTITION BY dept ORDER BY salary)`<br>• Each can have different OVER clause<br>• Named windows (WINDOW clause) reduce repetition<br>• Performance: optimizer shares sort |
| 22 | Named window — WINDOW w AS (...)? | Named window, WINDOW | • `SELECT ... FROM t WINDOW w AS (PARTITION BY dept ORDER BY salary DESC)`<br>• Reference: `ROW_NUMBER() OVER w`, `SUM(x) OVER w`<br>• Can extend: `SUM(x) OVER (w ROWS BETWEEN ...)`<br>• DRY — don't repeat partition/order |
| 23 | PERCENT_RANK and CUME_DIST? | PERCENT_RANK, CUME_DIST | • `PERCENT_RANK()`: (rank - 1) / (total - 1), range 0 to 1<br>• `CUME_DIST()`: fraction of rows ≤ current value<br>• PERCENT_RANK: first row always 0<br>• CUME_DIST: last row always 1 |
| 24 | Find gaps in sequential data? | Gaps, LEAD, sequence | • `SELECT id, LEAD(id) OVER (ORDER BY id) AS next_id FROM t WHERE LEAD(id) OVER (ORDER BY id) - id > 1`<br>• Gap between current and next ID<br>• Or: `generate_series LEFT JOIN` approach<br>• Also: LAG to compare with previous |
| 25 | Find consecutive sequences (islands)? | Islands, consecutive | • Group consecutive rows: `id - ROW_NUMBER() OVER (ORDER BY id)` = constant for each island<br>• Called "gaps and islands" problem<br>• GROUP BY the computed constant to find each island<br>• Classic window function pattern |
| 26 | Time between events with windows? | Time between, LAG | • `event_time - LAG(event_time) OVER (PARTITION BY user ORDER BY event_time)` → interval<br>• Time since previous event per user<br>• AVG for average time between events<br>• NULL for first event (no previous) |
| 27 | FILTER with window functions? | FILTER, window | • `COUNT(*) FILTER (WHERE status='active') OVER (PARTITION BY dept)`<br>• Conditional window aggregation<br>• PG-specific (not widely supported)<br>• Alternative: `SUM(CASE WHEN ... END) OVER (...)` |
| 28 | DISTINCT with window functions? | DISTINCT, window | • No `DISTINCT` directly in window functions<br>• Workaround: CTE with DISTINCT, then apply window<br>• Or: `DENSE_RANK` to assign unique ranks<br>• Use subquery to deduplicate first |
| 29 | Window functions and performance? | Window, performance | • Sort-based: partition + order requires sorting<br>• Multiple windows with same sort share one sort<br>• Indexes on PARTITION BY + ORDER BY columns help<br>• Window functions scan data only once |
| 30 | Window functions in WHERE? Workaround? | WHERE, window, CTE | • Can NOT use window functions in WHERE clause<br>• Workaround: wrap in CTE/subquery, filter in outer<br>• `WITH r AS (SELECT *, ROW_NUMBER() OVER (...) rn FROM t) SELECT * FROM r WHERE rn = 1`<br>• This is why CTEs are essential |
| 31 | Rolling 7-day average of sales? | Rolling average, 7 day | • `AVG(sales) OVER (ORDER BY date ROWS BETWEEN 6 PRECEDING AND CURRENT ROW)`<br>• 7 rows: 6 preceding + current<br>• Or RANGE: `RANGE BETWEEN INTERVAL '7 days' PRECEDING AND CURRENT ROW`<br>• ROWS for fixed count; RANGE for calendar days |
| 32 | Rank within partition, filter top-3 per group? | Rank, partition, top 3 | • `DENSE_RANK() OVER (PARTITION BY category ORDER BY sales DESC) rk`<br>• CTE + `WHERE rk <= 3`<br>• Includes ties (may return >3 per group)<br>• Use ROW_NUMBER for exactly 3 |
| 33 | Year-over-year growth with windows? | YoY, growth, LAG | • `LAG(revenue, 1) OVER (ORDER BY year)` → previous year<br>• Growth: `(revenue - LAG(revenue) OVER (...)) / LAG(revenue) OVER (...) * 100`<br>• PARTITION BY product for per-product YoY<br>• Window approach is cleaner than self-join |
| 34 | SUM(x) OVER() vs SUM(x)? | SUM OVER, aggregate, differ | • `SUM(x)`: requires GROUP BY, returns one row per group<br>• `SUM(x) OVER()`: no GROUP BY, adds total to every row<br>• Window: each row sees the total alongside its own values<br>• Useful for percentage calculations |
| 35 | First and last order per customer? | First, last, per customer | • `FIRST_VALUE(order_date) OVER (PARTITION BY customer ORDER BY order_date ASC)` → first<br>• `LAST_VALUE(...) OVER (... ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING)` → last<br>• Or use `MIN/MAX` aggregate |


## Section 14: Indexes & Query Performance

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | What is an index? How does it speed up queries? | Index, B-Tree, speed | • Data structure for fast row lookup (like a book's index)<br>• Avoids full table scan — jumps to matching rows<br>• Speeds up SELECT, slows INSERT/UPDATE/DELETE<br>• Most common: B-Tree (default) |
| 2 | B-Tree index — how it works? | B-Tree, balanced tree | • Sorted tree structure, O(log n) lookups<br>• Good for: `=`, `<`, `>`, `BETWEEN`, `LIKE 'abc%'`<br>• Bad for: `LIKE '%abc'` (leading wildcard)<br>• Default index type in PostgreSQL |
| 3 | When to create vs NOT create an index? | Index, when, tradeoff | • Create: columns in WHERE, JOIN, ORDER BY<br>• Don't: very small tables, low-cardinality columns<br>• Don't: frequently updated columns, heavy INSERT workloads<br>• Monitor with `pg_stat_user_indexes` |
| 4 | Cost of maintaining indexes? | Index, write cost, storage | • Every INSERT adds entry to all indexes on table<br>• UPDATE old + new index entries (if indexed column changes)<br>• DELETE marks index entries for cleanup<br>• Storage: each index uses disk space |
| 5 | Composite index — column order? | Composite, multi-column, order | • Index on (a, b, c) — column order matters<br>• Supports queries on: (a), (a,b), (a,b,c)<br>• Does NOT help queries on (b) or (c) alone<br>• Put most selective/most filtered column first |
| 6 | Leftmost prefix rule? | Leftmost prefix, composite | • Composite index (a, b, c) is like an index on a, (a,b), and (a,b,c)<br>• Query on (b) alone can't use this index<br>• Must include leftmost column(s) for index usage<br>• Order your composite index based on query patterns |
| 7 | Covering index (index-only scan) with INCLUDE? | Covering, INCLUDE, index-only | • Includes all columns needed → no table access<br>• PG 11+: `CREATE INDEX idx ON t(a) INCLUDE (b, c)`<br>• Index-only scan is fastest possible lookup<br>• INCLUDE columns not in index tree, just stored |
| 8 | Partial index — when useful? | Partial index, filtered | • `CREATE INDEX idx ON orders(status) WHERE status = 'pending'`<br>• Smaller index — only matching rows<br>• Perfect for: columns where you query specific values<br>• PG feature; not available in all databases |
| 9 | Expression (functional) index? | Expression, functional index | • `CREATE INDEX idx ON t(LOWER(email))` — index on expression<br>• `WHERE LOWER(email) = 'user@test.com'` now uses index<br>• Without: functions in WHERE prevent index usage<br>• Any deterministic expression works |
| 10 | Unique index vs unique constraint? | Unique index, constraint | • Unique constraint creates a unique index internally<br>• Semantically: constraint = data rule; index = performance<br>• Both prevent duplicate values<br>• Constraint can be deferred; index cannot |
| 11 | Hash index — when over B-Tree? | Hash index, equality | • Hash index: only supports `=` (equality) lookups<br>• Faster than B-Tree for exact matches<br>• No range queries, no sorting<br>• PG 10+ made hash indexes WAL-logged and crash-safe |
| 12 | GIN index — when to use? | GIN, JSONB, arrays, FTS | • Generalized Inverted Index — for multi-valued columns<br>• JSONB, arrays, full-text search (tsvector)<br>• Maps each element to rows containing it<br>• Slow to update, fast to query |
| 13 | GiST index — what data types? | GiST, spatial, range | • Generalized Search Tree — geometric, range, full-text<br>• PostGIS geometry, inet, range types<br>• Supports containment, overlap, nearest-neighbor<br>• Used with exclusion constraints |
| 14 | BRIN index — when appropriate? | BRIN, large, ordered | • Block Range Index — tiny summary per disk block<br>• Ideal: large, naturally ordered tables (timestamps)<br>• Much smaller than B-Tree (10-100x)<br>• Perfect for append-only time-series |
| 15 | Index selectivity — why low selectivity = useless? | Selectivity, cardinality | • Selectivity = fraction of distinct values in a column<br>• High selectivity (email) → index very useful<br>• Low selectivity (boolean, gender) → index returns most of table<br>• Optimizer prefers seq scan when selectivity is low |
| 16 | Check if index is used? EXPLAIN ANALYZE? | Index usage, EXPLAIN | • `EXPLAIN ANALYZE SELECT ...` — shows actual execution plan<br>• Look for: Index Scan, Index Only Scan (good)<br>• Seq Scan (bad for large tables unless intended)<br>• Compare estimated vs actual rows |
| 17 | Seq scan vs index scan vs bitmap index scan? | Seq scan, index scan, bitmap | • Seq scan: reads every row (good for small tables or most rows)<br>• Index scan: random I/O per row (good for few rows)<br>• Bitmap scan: builds bitmap of matches, then reads (medium selectivity)<br>• Optimizer chooses based on estimated rows |
| 18 | Index bloat — detect and fix? | Bloat, pg_stat, REINDEX | • Dead index entries accumulate after UPDATE/DELETE<br>• Check: `pgstattuple` extension or bloat estimation queries<br>• Fix: `REINDEX INDEX idx_name` or `REINDEX TABLE t`<br>• VACUUM removes dead tuples but not index bloat |
| 19 | REINDEX — when to use? | REINDEX, rebuild | • Rebuilds index from scratch<br>• Use when: index bloated, corrupted, or after many updates<br>• `REINDEX INDEX idx` — specific index<br>• `REINDEX INDEX CONCURRENTLY idx` — no lock (PG 12+) |
| 20 | CLUSTER — physical reorder by index? | CLUSTER, physical order | • `CLUSTER table USING index` — reorders table rows by index<br>• Only lasts until next INSERT/UPDATE<br>• Exclusive lock during operation<br>• Big performance boost for range queries on clustered column |
| 21 | Find unused indexes? | Unused index, pg_stat | • `SELECT * FROM pg_stat_user_indexes WHERE idx_scan = 0`<br>• Zero scans = unused index (wasting space and write performance)<br>• Wait for full workload cycle before dropping<br>• Consider: some indexes exist for constraints |
| 22 | Find missing indexes? | Missing index, slow query | • `EXPLAIN ANALYZE` shows Seq Scans on large tables<br>• `pg_stat_user_tables`: high seq_scan count = missing index<br>• Check WHERE/JOIN/ORDER BY columns in slow queries<br>• `pg_stat_statements` for top slow queries |
| 23 | pg_stat_user_indexes for analysis? | pg_stat_user_indexes | • `idx_scan`: number of times index was used<br>• `idx_tup_read`: rows read from index<br>• `idx_tup_fetch`: rows fetched from table via index<br>• Low idx_scan = candidate for dropping |
| 24 | Choose between multiple indexes? | Index choice, optimizer | • Optimizer considers selectivity, cost, correlation<br>• Only one index per table used in simple queries<br>• Bitmap scan can combine multiple indexes<br>• Hint: check EXPLAIN to see which index was chosen |
| 25 | Indexes and NULL values? | Index, NULL | • B-Tree indexes include NULLs by default<br>• `IS NULL` queries CAN use B-Tree index<br>• Partial index: `WHERE col IS NOT NULL` excludes NULLs<br>• NULLs at the end/start of sorted index depending on ASC/DESC |
| 26 | CREATE INDEX CONCURRENTLY? | CONCURRENTLY, no lock | • `CREATE INDEX CONCURRENTLY idx ON t(col)` — no write lock<br>• Takes longer but doesn't block INSERT/UPDATE/DELETE<br>• Can fail (leaves invalid index — clean up with DROP)<br>• Essential for production databases |
| 27 | Optimize LIKE '%text%' (leading wildcard)? | Leading wildcard, pg_trgm | • Regular B-Tree can't help with leading wildcard<br>• `pg_trgm` extension + GIN index: `CREATE INDEX idx ON t USING GIN(col gin_trgm_ops)`<br>• Supports: `LIKE '%text%'`, `ILIKE`, similarity<br>• Full-text search is alternative for word-based |
| 28 | pg_trgm for trigram-based indexing? | pg_trgm, trigram, similarity | • `CREATE EXTENSION pg_trgm`<br>• Creates trigrams (3-char substrings) of text<br>• `%` operator for similarity matching<br>• `SELECT * FROM t WHERE col % 'search_term'` |
| 29 | Index a JSONB column for key lookups? | JSONB index, GIN | • Full: `CREATE INDEX idx ON t USING GIN(col)` — all keys<br>• Optimized: `USING GIN(col jsonb_path_ops)` — containment only<br>• Expression: `CREATE INDEX idx ON t((col->>'specific_key'))`<br>• Choose based on query patterns |
| 30 | Impact of indexes on INSERT/UPDATE/DELETE? | Index, write, DML | • Each index adds write overhead<br>• INSERT: add entries to every index<br>• UPDATE: delete old + insert new (if indexed column changes)<br>• Unused indexes: pure overhead — drop them |


## Section 15: Database Design & Normalization

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | What is normalization? Why important? | Normalization, redundancy | • Organizes tables to reduce data redundancy<br>• Improves data integrity and consistency<br>• Reduces update anomalies<br>• Progressive forms: 1NF → 2NF → 3NF → BCNF |
| 2 | 1NF, 2NF, 3NF with examples? | 1NF, 2NF, 3NF, normal forms | • 1NF: atomic values (no lists in cells)<br>• 2NF: 1NF + no partial dependencies on composite PK<br>• 3NF: 2NF + no transitive dependencies (non-key → non-key)<br>• Each form eliminates a type of redundancy |
| 3 | BCNF — how differs from 3NF? | BCNF, Boyce-Codd | • Stricter than 3NF: every determinant must be a candidate key<br>• 3NF allows some anomalies with overlapping candidate keys<br>• Most 3NF tables are already BCNF<br>• BCNF handles edge cases that 3NF misses |
| 4 | What is denormalization? When good? | Denormalization, redundancy | • Intentionally adds redundancy for read performance<br>• Reduces JOINs at cost of update complexity<br>• Good for: OLAP, reporting, data warehouses<br>• Bad for: OLTP with frequent updates |
| 5 | Normalization vs denormalization trade-offs? | Normalize, denormalize, trade | • Normalized: consistent, compact, slower reads (more JOINs)<br>• Denormalized: redundant, faster reads, harder updates<br>• OLTP → normalize; OLAP → denormalize<br>• Choose based on read/write ratio |
| 6 | Surrogate key vs natural key — preference? | Surrogate, natural, key | • Surrogate (auto-increment, UUID): stable, compact, no business meaning<br>• Natural (email, SSN): meaningful, may change, format issues<br>• Surrogate preferred for: internal IDs, FKs<br>• Natural good for: lookup codes, standards |
| 7 | Composite key — when to use? | Composite, compound key | • Multiple columns form the primary key<br>• `PRIMARY KEY (student_id, course_id)` in junction tables<br>• When no single column uniquely identifies a row<br>• Common in M:N relationship tables |
| 8 | Model one-to-one relationship? | 1:1, one-to-one | • FK with UNIQUE constraint on child table<br>• `user_profiles.user_id REFERENCES users(id) UNIQUE`<br>• Or merge into single table (if always 1:1)<br>• Separate tables for: optional data, large BLOBs |
| 9 | Model one-to-many relationship? | 1:N, one-to-many | • FK in the "many" side table<br>• `orders.customer_id REFERENCES customers(id)`<br>• Most common relationship type<br>• Child table has FK pointing to parent PK |
| 10 | Model many-to-many with junction table? | M:N, junction, bridge | • Create junction table: `enrollments(student_id, course_id)`<br>• Both columns are FKs to parent tables<br>• Can add extra data: grade, enrolled_at<br>• PK: composite (student_id, course_id) or surrogate |
| 11 | Model hierarchical data (trees)? | Hierarchy, tree, recursive | • Adjacency list: `parent_id` column (simplest)<br>• Nested sets: `lft`, `rgt` columns (fast reads, slow writes)<br>• Materialized path: `path = '1.3.7'` (easy ancestry)<br>• Closure table: all ancestor-descendant pairs |
| 12 | Compare adjacency list, nested sets, materialized path, closure table? | Tree patterns, compare | • Adjacency: simple, recursive CTE for queries<br>• Nested sets: fast subtree queries, slow inserts<br>• Materialized path: easy ancestry, LTREE extension in PG<br>• Closure table: fastest queries, more storage |
| 13 | Schema for tagging system? | Tags, schema, design | • Tags table: `tags(id, name)` — unique tag names<br>• Junction table: `item_tags(item_id, tag_id)` — M:N<br>• Or: JSONB array of tags (simpler, less normalized)<br>• GIN index on tags for fast search |
| 14 | Schema for user permissions (RBAC)? | RBAC, permissions, roles | • `users`, `roles`, `permissions` tables<br>• `user_roles(user_id, role_id)` — M:N<br>• `role_permissions(role_id, permission_id)` — M:N<br>• Check: `WHERE EXISTS (... JOIN user_roles JOIN role_permissions ...)` |
| 15 | Schema for audit log? | Audit log, design | • `audit_log(id, table_name, record_id, action, old_data, new_data, user_id, created_at)`<br>• old_data/new_data as JSONB<br>• Populated by triggers or application code<br>• Append-only — never UPDATE or DELETE |
| 16 | Schema for multi-tenant application? | Multi-tenant, schema | • Shared DB + tenant_id column: simplest, RLS for security<br>• Schema per tenant: good isolation, more complex<br>• DB per tenant: best isolation, hardest to manage<br>• Most common: shared DB + tenant_id |
| 17 | EAV pattern — when appropriate? | EAV, entity-attribute-value | • `entity_id, attribute_name, attribute_value` — flexible schema<br>• Use for: varying attributes per entity, custom fields<br>• Cons: complex queries, no type safety, slow joins<br>• Better alternative: JSONB column |
| 18 | Soft delete vs hard delete design? | Soft delete, hard delete | • Soft: `is_deleted=TRUE, deleted_at=NOW()` (preserve data)<br>• Hard: `DELETE FROM table` (permanent removal)<br>• Soft: add partial index WHERE NOT is_deleted for performance<br>• Consider: legal requirements, GDPR compliance |
| 19 | Temporal data modeling (change history)? | Temporal, history, SCD | • Type 1 SCD: overwrite (no history)<br>• Type 2 SCD: new row with effective_from/to dates<br>• Temporal tables: system-versioned (PG doesn't have built-in)<br>• Use triggers or audit tables for change history |
| 20 | Polymorphic associations? | Polymorphic, association | • One FK referencing multiple parent tables<br>• Approach 1: `parent_type + parent_id` columns<br>• Approach 2: separate FK columns (one per parent type)<br>• Approach 3: table inheritance (PG feature)<br>• Approach 1 can't enforce FK constraint |
| 21 | Table inheritance in PostgreSQL? | Table inheritance, PG | • `CREATE TABLE child () INHERITS (parent)` — PG-specific<br>• Child inherits all parent columns<br>• `SELECT * FROM parent` includes child rows<br>• Limited FK support; consider partitioning instead |
| 22 | Add column vs add table — how to choose? | Add column, add table | • Add column: data always present, same entity<br>• Add table: optional data, separate concern, M:N relationship<br>• Column: simple, fast; Table: normalized, flexible<br>• Consider future growth and query patterns |
| 23 | Vertical vs horizontal partitioning? | Vertical, horizontal, partition | • Vertical: split columns across tables (wide → narrow)<br>• Horizontal: split rows across partitions (range, list, hash)<br>• Vertical for: rarely accessed columns, BLOBs<br>• Horizontal for: time-series, scaling |
| 24 | Schema for event-driven systems? | Event, schema, event sourcing | • Events table: `events(id, type, payload JSONB, created_at)`<br>• Append-only — events are immutable<br>• Current state = replaying all events (event sourcing)<br>• Or: events for audit + separate state table |
| 25 | Naming conventions for tables/columns/constraints? | Naming, conventions | • Tables: plural lowercase (`users`, `orders`)<br>• Columns: snake_case (`created_at`, `user_id`)<br>• Constraints: descriptive (`fk_orders_user_id`, `chk_salary_positive`)<br>• Consistent naming = easier debugging |


## Section 16: Constraints & Data Integrity

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | Types of constraints? | PK, FK, UNIQUE, CHECK, NOT NULL | • PRIMARY KEY: unique + not null, one per table<br>• FOREIGN KEY: references another table's PK/UNIQUE<br>• UNIQUE: no duplicate values (allows NULL)<br>• NOT NULL: disallows NULL<br>• CHECK: custom condition; DEFAULT: fallback value |
| 2 | Primary key vs unique constraint? | PK, UNIQUE, difference | • PK: one per table, no NULLs, primary identifier<br>• UNIQUE: multiple per table, allows one NULL (PG: multiple NULLs)<br>• Both create an index<br>• PK has stronger semantic meaning |
| 3 | ON DELETE CASCADE vs SET NULL vs RESTRICT? | CASCADE, SET NULL, RESTRICT | • CASCADE: auto-delete child rows when parent deleted<br>• SET NULL: set FK to NULL (column must be nullable)<br>• RESTRICT: prevent parent deletion if children exist<br>• Choose based on data relationship semantics |
| 4 | RESTRICT vs NO ACTION for foreign keys? | RESTRICT, NO ACTION | • Both prevent parent deletion when children exist<br>• RESTRICT: checked immediately<br>• NO ACTION: checked at end of statement (can use with deferred constraints)<br>• NO ACTION is the default in PG |
| 5 | ON UPDATE CASCADE — when to use? | ON UPDATE, CASCADE | • Auto-updates FK when parent's PK value changes<br>• Rarely needed if using surrogate keys (PKs don't change)<br>• Useful with natural keys that might change<br>• `ON UPDATE CASCADE` propagates PK changes |
| 6 | CHECK constraint — practical examples? | CHECK, validation | • `CHECK (salary >= 0)` — non-negative salary<br>• `CHECK (end_date > start_date)` — date range validity<br>• `CHECK (status IN ('active', 'inactive'))` — value list<br>• Multiple CHECKs per column allowed |
| 7 | Deferrable constraints? INITIALLY DEFERRED? | Deferrable, deferred | • `DEFERRABLE INITIALLY DEFERRED` — checked at COMMIT<br>• `DEFERRABLE INITIALLY IMMEDIATE` — checked immediately but can be deferred<br>• Normal constraints: always immediate<br>• Useful for circular references during data load |
| 8 | When to use deferrable constraints? | Deferrable, use case | • Loading data with circular FKs<br>• Bulk insert where order matters<br>• Self-referencing tables<br>• `SET CONSTRAINTS ALL DEFERRED` within transaction |
| 9 | Exclusion constraint (EXCLUDE)? | EXCLUDE, exclusion | • `EXCLUDE USING GiST (room WITH =, time_range WITH &&)`<br>• Prevents overlapping ranges for same room<br>• Requires `btree_gist` extension<br>• More powerful than UNIQUE (custom operators) |
| 10 | Add constraint without long lock? | Lock, ADD CONSTRAINT | • `ALTER TABLE t ADD CONSTRAINT c ... NOT VALID` — adds without validating existing data<br>• Then: `ALTER TABLE t VALIDATE CONSTRAINT c` — validates in background<br>• NOT VALID: immediate, no full table lock<br>• VALIDATE: scans table but allows concurrent writes |
| 11 | Validate existing data for NOT NULL? | NOT NULL, existing data | • Step 1: fix NULL values `UPDATE t SET col = default WHERE col IS NULL`<br>• Step 2: `ALTER TABLE t ALTER COLUMN col SET NOT NULL`<br>• Or: add CHECK constraint NOT VALID, then validate<br>• Must fix all NULLs before adding constraint |
| 12 | ADD CONSTRAINT NOT VALID — why? | NOT VALID, constraint | • Adds constraint without scanning existing rows<br>• New inserts/updates are validated immediately<br>• `VALIDATE CONSTRAINT` later to check existing data<br>• Minimizes lock time on large tables |
| 13 | Constraints and insert/update performance? | Constraint, performance | • Each constraint adds check overhead per row<br>• FK constraints: cross-table lookup on insert/update<br>• CHECK: evaluate expression per row<br>• Minor overhead vs. data integrity benefit |
| 14 | Constraints and indexes — relationship? | Constraint, index | • PK and UNIQUE automatically create indexes<br>• FK does NOT create index in PG (create manually!)<br>• CHECK uses no index (evaluated per row)<br>• Always add index on FK columns |
| 15 | FK referencing unique column (not PK)? | FK, UNIQUE, reference | • Yes — FK can reference any UNIQUE column, not just PK<br>• `REFERENCES other_table(unique_column)`<br>• Column must have UNIQUE constraint<br>• Less common but valid |
| 16 | Enforce immutable rows (prevent updates)? | Immutable, prevent update | • Trigger: BEFORE UPDATE → `RAISE EXCEPTION 'immutable'`<br>• Or: REVOKE UPDATE on table from application role<br>• Or: Row-Level Security policy<br>• No built-in immutability constraint |
| 17 | Circular FK dependencies? | Circular FK, dependency | • Table A FK → Table B, Table B FK → Table A<br>• Problem: can't insert into either first<br>• Fix: deferrable constraints (`INITIALLY DEFERRED`)<br>• Or: allow NULL FK, insert, then update |
| 18 | Conditional constraints? | Conditional, constraint | • No direct conditional constraints in SQL standard<br>• Workaround: CHECK with CASE or OR<br>• `CHECK (status != 'shipped' OR tracking_number IS NOT NULL)`<br>• Or: use trigger for complex conditions |
| 19 | DEFAULT with gen_random_uuid()? | DEFAULT, UUID, auto | • `CREATE TABLE t (id UUID DEFAULT gen_random_uuid())`<br>• Auto-generates UUID for each insert<br>• `gen_random_uuid()` — PG 13+ built-in<br>• Or: `uuid-ossp` extension for older versions |
| 20 | Constraint naming — why it matters? | Naming, constraint, debug | • `CONSTRAINT chk_salary_positive CHECK (salary >= 0)`<br>• Named constraints → clear error messages<br>• `ERROR: violating constraint "chk_salary_positive"` vs generic message<br>• Easier to ALTER/DROP specific constraints |


## Section 17: Transactions, ACID & Isolation

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | What is a transaction? Why important? | Transaction, unit of work | • Unit of work that is all-or-nothing<br>• `BEGIN; ...statements...; COMMIT;` or `ROLLBACK;`<br>• Ensures data consistency even on failure<br>• Groups related operations into atomic unit |
| 2 | ACID properties with examples? | ACID, Atomicity, Consistency | • **A**tomicity: all or nothing (transfer = debit + credit)<br>• **C**onsistency: valid state → valid state (constraints hold)<br>• **I**solation: concurrent transactions don't interfere<br>• **D**urability: committed data survives crashes (WAL) |
| 3 | COMMIT vs ROLLBACK? | COMMIT, ROLLBACK | • COMMIT: saves all changes permanently<br>• ROLLBACK: undoes all changes since BEGIN<br>• After COMMIT: durable, can't undo<br>• After ROLLBACK: database returns to prior state |
| 4 | SAVEPOINT — partial rollback? | SAVEPOINT, partial rollback | • Named point within a transaction<br>• `SAVEPOINT sp1; ... ROLLBACK TO sp1;` — partial undo<br>• Transaction continues after partial rollback<br>• Multiple savepoints allowed |
| 5 | Four isolation levels — anomalies prevented? | Isolation levels, anomalies | • **READ UNCOMMITTED**: allows dirty reads (not in PG)<br>• **READ COMMITTED**: prevents dirty reads (PG default)<br>• **REPEATABLE READ**: prevents non-repeatable reads<br>• **SERIALIZABLE**: prevents all anomalies |
| 6 | What is a dirty read? | Dirty read, uncommitted | • Reading data from uncommitted transaction<br>• If that transaction rolls back → you read phantom data<br>• Prevented by READ COMMITTED and above<br>• Not possible in PostgreSQL |
| 7 | Non-repeatable read? | Non-repeatable read, repeat | • Same query returns different data within same transaction<br>• Another transaction modified rows between reads<br>• Prevented by REPEATABLE READ<br>• PG default (READ COMMITTED) allows this |
| 8 | Phantom read? | Phantom read, new rows | • New rows appear when re-running same query in transaction<br>• Another transaction inserted matching rows<br>• Prevented by SERIALIZABLE<br>• REPEATABLE READ in PG also prevents phantoms |
| 9 | Default isolation level in PostgreSQL? | Default isolation, PG | • READ COMMITTED is the default<br>• Each statement sees data committed before it started<br>• Different statements in same transaction may see different snapshots<br>• Change with: `SET TRANSACTION ISOLATION LEVEL ...` |
| 10 | Change isolation level for a transaction? | SET TRANSACTION, isolation | • `BEGIN; SET TRANSACTION ISOLATION LEVEL SERIALIZABLE;`<br>• Must be first command after BEGIN<br>• Or: `BEGIN ISOLATION LEVEL SERIALIZABLE;`<br>• `SET default_transaction_isolation` for session default |
| 11 | Serialization anomaly — SERIALIZABLE prevention? | Serialization, anomaly | • Two transactions whose interleaved execution gives different result than any serial order<br>• SERIALIZABLE detects and aborts one transaction<br>• Application must retry aborted transaction<br>• Uses predicate locking (not actual row locks) |
| 12 | Performance trade-off of higher isolation? | Isolation, performance | • Higher isolation = more locking/checking = slower<br>• READ COMMITTED: fastest, least protection<br>• SERIALIZABLE: safest, most overhead (retries needed)<br>• Default READ COMMITTED is fine for most apps |
| 13 | Concurrent conflicts at SERIALIZABLE? | SERIALIZABLE, conflict | • PG detects potential serialization anomaly<br>• One transaction succeeds, other gets serialization_failure error<br>• Application must catch error and retry<br>• Use retry loop in application code |
| 14 | Implicit vs explicit transaction? | Implicit, explicit, transaction | • Explicit: `BEGIN; ... COMMIT;` — you control boundaries<br>• Implicit: each statement is its own transaction (autocommit)<br>• Autocommit is PG default outside explicit BEGIN<br>• Always use explicit for multi-statement operations |
| 15 | What is autocommit mode? | Autocommit, implicit | • Each SQL statement auto-committed immediately<br>• No manual BEGIN/COMMIT needed<br>• PG default behavior when not in explicit transaction<br>• psql: `\set AUTOCOMMIT off` to disable |
| 16 | Handle transaction failures and retries? | Retry, failure, application | • Catch serialization_failure or deadlock_detected errors<br>• Retry entire transaction from BEGIN<br>• Exponential backoff between retries<br>• Set max retry count to avoid infinite loops |
| 17 | Long-running transaction — why problematic? | Long transaction, problem | • Holds locks → blocks other transactions<br>• Prevents VACUUM from cleaning dead tuples<br>• Causes table bloat over time<br>• Keep transactions as short as possible |
| 18 | Long transactions and VACUUM/bloat? | VACUUM, bloat, long txn | • VACUUM can only clean tuples invisible to ALL transactions<br>• Long transaction holds oldest snapshot → prevents cleanup<br>• Dead tuples accumulate → table bloat<br>• Monitor with `pg_stat_activity` for old transactions |
| 19 | Two-phase commit (2PC)? | 2PC, distributed | • Coordinates transaction across multiple databases<br>• Phase 1: PREPARE — all participants vote<br>• Phase 2: COMMIT — if all voted yes<br>• PG: `PREPARE TRANSACTION`, `COMMIT PREPARED` |
| 20 | Idempotent transactions? | Idempotent, retry safe | • Running transaction multiple times produces same result<br>• Use ON CONFLICT DO NOTHING/UPDATE (UPSERT)<br>• Use conditional updates: `WHERE version = expected_version`<br>• Essential for safe retry logic |
| 21 | Write skew anomaly? | Write skew, serializable | • Two transactions read overlapping data, each writes different row<br>• Combined result violates a constraint neither checked<br>• Only prevented at SERIALIZABLE level<br>• Example: two doctors both take leave because the other is present |
| 22 | Transactions for financial operations? | Financial, money transfer | • `BEGIN; UPDATE accounts SET balance = balance - 100 WHERE id = 1; UPDATE accounts SET balance = balance + 100 WHERE id = 2; COMMIT;`<br>• Use SERIALIZABLE for strict consistency<br>• Always check balance before debit (SELECT FOR UPDATE)<br>• Idempotent with unique transfer_id |
| 23 | Optimistic vs pessimistic concurrency? | Optimistic, pessimistic | • Pessimistic: lock before modify (`SELECT FOR UPDATE`)<br>• Optimistic: no lock, version check on update<br>• `UPDATE t SET ... WHERE id = 1 AND version = 5` — optimistic<br>• Optimistic: better for low-conflict; Pessimistic: high-conflict |
| 24 | Monitor active/blocked transactions? | Monitor, pg_stat_activity | • `SELECT * FROM pg_stat_activity WHERE state != 'idle'`<br>• `wait_event_type` — shows if blocked<br>• `pg_blocking_pids(pid)` — who's blocking whom<br>• `pg_cancel_backend(pid)` to cancel stuck queries |
| 25 | Advisory locks — when to use? | Advisory lock, application | • Application-level locks not tied to tables<br>• `pg_advisory_lock(key)` — session-level<br>• `pg_try_advisory_lock(key)` — non-blocking<br>• Use for: job scheduling, rate limiting, distributed locks |


## Section 18: Locking & Concurrency Control

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | MVCC — how PostgreSQL implements it? | MVCC, multi-version | • Each transaction sees a snapshot of data at a point in time<br>• Rows have xmin/xmax (creation/deletion transaction IDs)<br>• Readers don't block writers; writers don't block readers<br>• Old versions kept until VACUUM cleans them |
| 2 | Why doesn't PG block readers for writers? | MVCC, no reader block | • MVCC gives each reader its own snapshot<br>• Writer creates new version; reader sees old version<br>• No read-write conflict — both proceed independently<br>• Key advantage of MVCC architecture |
| 3 | Shared lock vs exclusive lock? | Shared, exclusive, lock | • Shared: multiple readers can hold simultaneously (read lock)<br>• Exclusive: only one holder, blocks all others (write lock)<br>• Shared + Shared = OK; Shared + Exclusive = blocked<br>• SELECT = shared; UPDATE/DELETE = exclusive (on row) |
| 4 | Row-level vs table-level locking? | Row lock, table lock | • Row-level: locks specific rows (most common in PG)<br>• Table-level: locks entire table (ALTER TABLE, LOCK TABLE)<br>• PG prefers row-level for concurrency<br>• Table locks needed for DDL operations |
| 5 | SELECT FOR UPDATE — when and why? | FOR UPDATE, row lock | • Locks selected rows until COMMIT/ROLLBACK<br>• Prevents other transactions from modifying them<br>• `SELECT * FROM orders WHERE id = 1 FOR UPDATE`<br>• Use for: read-then-update patterns, reservations |
| 6 | SELECT FOR SHARE? | FOR SHARE, shared lock | • Prevents other transactions from updating/deleting<br>• Allows other SELECT FOR SHARE<br>• Weaker than FOR UPDATE (still allows reads)<br>• Use when you need to ensure data doesn't change |
| 7 | NOWAIT — can't acquire lock? | NOWAIT, lock timeout | • `SELECT FOR UPDATE NOWAIT` — error immediately if locked<br>• Instead of waiting indefinitely<br>• Application handles error (retry or fail)<br>• Better UX than hanging on locked rows |
| 8 | SKIP LOCKED — job queues? | SKIP LOCKED, queue | • `SELECT ... FOR UPDATE SKIP LOCKED LIMIT 1` — skip locked rows<br>• Multiple workers can process different jobs concurrently<br>• Locked rows are invisible — worker gets next available<br>• Perfect for: task queues, work distribution |
| 9 | Deadlock — prevention strategies? | Deadlock, prevention | • Two transactions hold locks each other needs<br>• Prevention: always acquire locks in consistent order<br>• Keep transactions short — less chance of conflict<br>• PG auto-detects and kills one transaction |
| 10 | PG deadlock detection and resolution? | Deadlock, detection | • PG checks for deadlocks every `deadlock_timeout` (1s default)<br>• Detects cycle in lock wait graph<br>• Kills one transaction with least work done<br>• Application should retry killed transaction |
| 11 | Lock timeout configuration? | lock_timeout, config | • `SET lock_timeout = '5s'` — fail after 5 seconds<br>• Prevents indefinite waits<br>• `statement_timeout` for overall query timeout<br>• Set per-session or per-transaction |
| 12 | Advisory locks vs regular locks? | Advisory, regular, difference | • Regular: automatic, tied to table/row operations<br>• Advisory: manual, application-controlled<br>• Regular: released at transaction/statement end<br>• Advisory: session-level (or transaction with `_xact` variant) |
| 13 | Diagnose locking issues? | pg_locks, pg_stat_activity | • `pg_locks` — shows all active locks<br>• `pg_stat_activity` — shows waiting queries<br>• `pg_blocking_pids(pid)` — who blocks whom<br>• Join both for complete picture |
| 14 | Optimistic vs pessimistic locking — implementation? | Optimistic, pessimistic, how | • Pessimistic: `SELECT ... FOR UPDATE; UPDATE ...;`<br>• Optimistic: add `version` column, `UPDATE ... WHERE version = expected; CHECK rows_affected > 0`<br>• Pessimistic: simpler, blocks concurrent access<br>• Optimistic: better throughput, retry on conflict |
| 15 | Lock escalation — does PG do it? | Lock escalation, PG | • Lock escalation: upgrading many row locks to table lock<br>• PostgreSQL does NOT do lock escalation<br>• Always row-level locks regardless of count<br>• Other DBs (SQL Server) may escalate |
| 16 | Locking impact on concurrent writes? | Concurrent, write, lock | • Row-level locks: minimal blocking (only same rows)<br>• Many concurrent updates to same row: serialized<br>• Different rows: fully concurrent<br>• Index updates also require brief locks |
| 17 | Job queue with SKIP LOCKED? | Queue, SKIP LOCKED, worker | • `BEGIN; SELECT id FROM jobs WHERE status='pending' FOR UPDATE SKIP LOCKED LIMIT 1; UPDATE jobs SET status='processing' WHERE id = ?; COMMIT;`<br>• Workers automatically get different jobs<br>• No coordinator needed<br>• Simple and effective |
| 18 | Lock contention on hot rows? | Hot row, contention | • Frequently-updated row (counter, balance) creates bottleneck<br>• Solutions: batch updates, advisory locks, partitioning<br>• Use `SELECT FOR UPDATE NOWAIT` to fail fast<br>• Or denormalize — spread updates across rows |
| 19 | Table-level lock modes? | ACCESS SHARE, ROW EXCLUSIVE | • ACCESS SHARE: SELECT (weakest)<br>• ROW SHARE: SELECT FOR UPDATE/SHARE<br>• ROW EXCLUSIVE: INSERT/UPDATE/DELETE<br>• EXCLUSIVE: blocks everything except ACCESS SHARE<br>• ACCESS EXCLUSIVE: strongest (ALTER TABLE, DROP) |
| 20 | Lock wait graph — debug concurrency? | Wait graph, debug | • Graph of transactions waiting for each other<br>• Cycle in graph = deadlock<br>• PG checks periodically for cycles<br>• Tools: `pg_locks` + `pg_stat_activity` join |


## Section 19: Query Optimization & EXPLAIN

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | What is EXPLAIN? | EXPLAIN, query plan | • Shows how the planner will execute a query<br>• Does NOT run the query<br>• Shows: node types, estimated cost, estimated rows<br>• `EXPLAIN SELECT * FROM t WHERE id = 1` |
| 2 | EXPLAIN ANALYZE — how differs? | EXPLAIN ANALYZE, actual | • Actually RUNS the query + shows real timing<br>• Shows: actual time, actual rows, loops<br>• Compare estimated vs actual for stat accuracy<br>• `EXPLAIN ANALYZE SELECT ...` (careful with DML — wraps in transaction) |
| 3 | Key metrics in EXPLAIN ANALYZE? | Metrics, actual time, rows | • Actual time: milliseconds per node<br>• Rows: actual vs estimated (mismatch = bad stats)<br>• Loops: number of times node executed<br>• Buffers: shared/local hit/read/written (with BUFFERS) |
| 4 | Sequential scan — when chosen over index? | Seq scan, full scan | • Small tables: seq scan is faster than index overhead<br>• Query returns most of table (>5-10% of rows)<br>• No suitable index exists<br>• After VACUUM/ANALYZE with accurate stats |
| 5 | Index scan vs bitmap index scan? | Index scan, bitmap scan | • Index scan: look up row by row via index (few rows)<br>• Bitmap scan: build bitmap of matching rows, then read sequentially<br>• Bitmap: multiple indexes can be combined (BitmapAnd/BitmapOr)<br>• Optimizer decides based on selectivity |
| 6 | Index-only scan — why faster? | Index-only scan, covering | • All needed columns are in the index<br>• No table access needed (index has all data)<br>• Requires visibility map to be up-to-date (VACUUM)<br>• Create with INCLUDE: `CREATE INDEX idx ON t(a) INCLUDE (b)` |
| 7 | Nested loop join in EXPLAIN? | Nested loop, plan | • `Nested Loop` node<br>• For each outer row → scan inner (using index or seq scan)<br>• Best when outer has few rows and inner has index<br>• Shows: loops = number of outer rows |
| 8 | Hash join in EXPLAIN? | Hash join, plan | • `Hash Join` node + `Hash` node<br>• Builds hash table on smaller table, probes with larger<br>• Best for medium-to-large joins without useful indexes<br>• Needs memory (work_mem) for hash table |
| 9 | Merge join in EXPLAIN? | Merge join, plan | • `Merge Join` node<br>• Both sides must be sorted (or use index)<br>• Scans both sides in order — efficient for large sorted data<br>• Best when both inputs are pre-sorted |
| 10 | Cost estimate — what does cost=0.00..100.00 mean? | Cost, estimate, startup | • Two numbers: startup cost .. total cost<br>• Startup cost: time before first row returned<br>• Total cost: time for all rows<br>• Arbitrary units — relative comparison only |
| 11 | Query planner/optimizer — how it decides? | Planner, optimizer | • Evaluates multiple possible plans<br>• Estimates cost using table statistics<br>• Picks lowest-cost plan<br>• Accuracy depends on `ANALYZE` being run recently |
| 12 | Table statistics — ANALYZE command? | ANALYZE, statistics | • `ANALYZE table_name` — collects column statistics<br>• Used by planner for cost estimation<br>• Autovacuum runs ANALYZE automatically<br>• Stale stats → bad query plans |
| 13 | VACUUM vs VACUUM FULL? | VACUUM, FULL, dead tuples | • `VACUUM`: marks dead tuples as reusable (no table shrink)<br>• `VACUUM FULL`: rewrites table, reclaims disk space (exclusive lock!)<br>• VACUUM FULL: much slower, locks table<br>• Regular VACUUM is usually sufficient |
| 14 | Autovacuum — how to tune? | Autovacuum, tuning | • Runs VACUUM and ANALYZE automatically<br>• Tune: `autovacuum_vacuum_threshold`, `autovacuum_vacuum_scale_factor`<br>• Hot tables: lower thresholds for more frequent vacuum<br>• Monitor with `pg_stat_user_tables` |
| 15 | Table bloat — how affects performance? | Bloat, dead tuples, space | • Dead rows take up space (MVCC doesn't delete immediately)<br>• More pages to scan → slower queries<br>• VACUUM marks space reusable; VACUUM FULL reclaims<br>• Monitor: `pg_stat_user_tables.n_dead_tup` |
| 16 | Optimize slow COUNT(*) on large table? | COUNT, slow, estimate | • `COUNT(*)` scans entire table (slow for millions of rows)<br>• Approximate: `SELECT reltuples FROM pg_class WHERE relname = 'table'`<br>• Exact: maintain counter table with triggers<br>• Or: count from covering index (faster) |
| 17 | Optimize LIMIT/OFFSET pagination for deep pages? | Pagination, OFFSET, slow | • Deep OFFSET is slow — DB reads and discards N rows<br>• Keyset pagination: `WHERE id > last_seen_id ORDER BY id LIMIT 10`<br>• Uses index, constant performance regardless of page<br>• Requires sortable, unique column |
| 18 | Keyset pagination — why faster? | Keyset, cursor, pagination | • Uses WHERE instead of OFFSET: `WHERE created_at > '2024-01-15'`<br>• Index seek — jumps directly to position<br>• O(1) vs O(n) for deep pages<br>• Client sends last-seen value as cursor |
| 19 | N+1 query problem — detect and fix? | N+1, detect, batch | • 1 query for N items + N queries for related data<br>• Fix: JOINs, batch `WHERE id IN (...)`, eager loading<br>• Detect: query logging, ORM debug mode<br>• ORM: `select_related()` (Django), `joinedload()` (SQLAlchemy) |
| 20 | Optimize multi-join queries? | Multi-join, optimize | • Ensure indexes on all JOIN columns<br>• Filter early with WHERE to reduce rows<br>• Check EXPLAIN for unexpected Seq Scans<br>• Optimizer handles join order (usually) |
| 21 | pg_stat_statements for slow queries? | pg_stat_statements, slow | • `CREATE EXTENSION pg_stat_statements`<br>• Tracks: total time, calls, rows per query<br>• `SELECT * FROM pg_stat_statements ORDER BY total_exec_time DESC`<br>• Top-N slowest queries for optimization |
| 22 | Prepared statements (PREPARE)? | PREPARE, prepared statement | • `PREPARE stmt AS SELECT * FROM t WHERE id = $1`<br>• `EXECUTE stmt(42)` — run with parameter<br>• Plan cached after multiple executions<br>• Prevents SQL injection + reduces planning overhead |
| 23 | Connection pooling (PgBouncer)? | PgBouncer, pool, connections | • Reuses database connections (avoid create/destroy overhead)<br>• PG fork per connection is expensive<br>• PgBouncer: external pooler, lightweight<br>• Modes: session, transaction, statement pooling |
| 24 | Table partitioning for large tables? | Partitioning, large table | • `CREATE TABLE t (...) PARTITION BY RANGE (date)` + child partitions<br>• Partition pruning: only scan relevant partitions<br>• Maintenance: drop old partitions instead of DELETE<br>• Types: range, list, hash |
| 25 | Range vs list vs hash partitioning? | Range, list, hash, partition | • Range: date ranges, numeric ranges (most common)<br>• List: discrete values (country, status)<br>• Hash: even distribution across N partitions<br>• Choose based on query patterns |
| 26 | Identify slow queries in production? | Slow query, log, production | • `log_min_duration_statement = 500ms` — log slow queries<br>• `pg_stat_statements` — aggregate statistics<br>• `auto_explain` extension — log plans for slow queries<br>• APM tools: Datadog, New Relic |
| 27 | Common query anti-patterns? | Anti-pattern, performance | • SELECT * instead of specific columns<br>• Deep OFFSET pagination<br>• Functions in WHERE on indexed columns<br>• N+1 queries; missing indexes on FKs |
| 28 | Optimize queries with OR conditions? | OR, optimize, UNION | • OR can prevent index usage<br>• Rewrite as UNION ALL: `SELECT ... WHERE a=1 UNION ALL SELECT ... WHERE b=2`<br>• Each UNION branch can use its own index<br>• Or: `ANY()` for single-column OR |
| 29 | Optimize DISTINCT on large datasets? | DISTINCT, optimize | • DISTINCT sorts entire result set<br>• Use GROUP BY (same result, sometimes faster plan)<br>• Add index on DISTINCT columns<br>• Consider: do you really need DISTINCT? Fix the JOIN instead |
| 30 | work_mem, shared_buffers, effective_cache_size? | Config, memory, tuning | • `shared_buffers`: PG's page cache (25% of RAM)<br>• `work_mem`: per-operation sort/hash memory (1-4MB default)<br>• `effective_cache_size`: hint for OS cache size (50-75% RAM)<br>• Too high work_mem × many connections = OOM |


## Section 20: Advanced SQL Features

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | UPSERT: ON CONFLICT DO NOTHING / DO UPDATE? | UPSERT, ON CONFLICT | • `INSERT ... ON CONFLICT (key) DO NOTHING` — skip if exists<br>• `DO UPDATE SET col = EXCLUDED.col` — update if exists<br>• `EXCLUDED` refers to the would-be inserted row<br>• Atomic — no race condition |

| 4 | COPY command — why faster than INSERT? | COPY, bulk, performance | • `COPY t FROM '/file.csv' CSV HEADER` — bulk load<br>• Bypasses SQL parsing and planning per row<br>• 10-100x faster than row-by-row INSERT<br>• `\copy` for client-side files (psql) |
| 5 | LISTEN / NOTIFY — real use cases? | LISTEN, NOTIFY, pub-sub | • `LISTEN channel; NOTIFY channel, 'payload';`<br>• Lightweight pub/sub within PostgreSQL<br>• Use for: cache invalidation, real-time updates<br>• Application polls with `SELECT pg_notification_queue_usage()` |
| 6 | Full-text search — tsvector and tsquery? | FTS, tsvector, tsquery | • `to_tsvector('english', text)` — tokenize document<br>• `to_tsquery('english', 'search & terms')` — search query<br>• `@@` operator matches tsvector against tsquery<br>• GIN index for fast full-text search |
| 7 | Rank full-text search results? | ts_rank, ranking, FTS | • `ts_rank(tsvector, tsquery)` — relevance score<br>• `ts_rank_cd` — cover density ranking<br>• `ORDER BY ts_rank(...) DESC` — most relevant first<br>• Combine with `ts_headline()` for highlighted snippets |
| 8 | generate_series() — practical uses? | generate_series, sequence | • Numbers: `generate_series(1, 100)` — 1 to 100<br>• Dates: `generate_series('2024-01-01'::date, '2024-12-31', '1 day')`<br>• Gap filling: LEFT JOIN with generated dates<br>• Calendar tables, test data generation |
| 9 | unnest() — expand arrays to rows? | unnest, array, expand | • `SELECT unnest(ARRAY[1,2,3])` → 3 rows<br>• `SELECT unnest(tags) FROM articles` — expand array column<br>• With ordinality: `unnest(arr) WITH ORDINALITY AS t(elem, idx)`<br>• Useful for: search, aggregation on array elements |
| 10 | DISTINCT ON in PostgreSQL? | DISTINCT ON, first per group | • `SELECT DISTINCT ON (dept) * FROM emp ORDER BY dept, salary DESC`<br>• Returns first row per group (based on ORDER BY)<br>• PG-specific — not standard SQL<br>• Simpler than ROW_NUMBER + filter for first-per-group |
| 11 | TABLESAMPLE — random sample of rows? | TABLESAMPLE, random | • `SELECT * FROM t TABLESAMPLE BERNOULLI(10)` — ~10% of rows<br>• `TABLESAMPLE SYSTEM(10)` — page-based sampling (faster, less random)<br>• BERNOULLI: row-by-row; SYSTEM: block-by-block<br>• No ORDER BY RANDOM() needed |
| 12 | LATERAL join — problems it solves? | LATERAL, correlated FROM | • Allows FROM subquery to reference earlier tables<br>• `FROM orders o, LATERAL (SELECT ... WHERE user_id = o.user_id LIMIT 3) sub`<br>• Can't do LIMIT per row with regular join<br>• Top-N per group without window functions |
| 13 | Generated/computed column? | GENERATED ALWAYS, computed | • `col INTEGER GENERATED ALWAYS AS (price * qty) STORED`<br>• Auto-computed on INSERT/UPDATE<br>• STORED: physically saved; VIRTUAL: computed on read (not in PG yet)<br>• Can't be directly written to |
| 14 | Table partitioning — how to create? | Partition, CREATE TABLE | • `CREATE TABLE orders (...) PARTITION BY RANGE (order_date)`<br>• `CREATE TABLE orders_2024 PARTITION OF orders FOR VALUES FROM ('2024-01-01') TO ('2025-01-01')`<br>• Automatic partition selection on INSERT<br>• Query pruning for SELECT performance |
| 15 | Partition pruning? | Partition pruning, skip | • Planner skips irrelevant partitions based on WHERE<br>• `WHERE order_date = '2024-06-15'` → only scans 2024 partition<br>• Check with EXPLAIN: should show only relevant partitions<br>• `enable_partition_pruning = on` (default) |
| 16 | Logical vs physical replication? | Logical, physical, replication | • Physical: byte-by-byte WAL streaming (exact replica)<br>• Logical: publishes changes as logical events (selective)<br>• Physical: faster, simpler, full database<br>• Logical: table-level, cross-version, cross-platform |
| 17 | pg_dump and pg_restore? | pg_dump, pg_restore, backup | • `pg_dump dbname > backup.sql` — logical backup<br>• `pg_dump -Fc dbname > backup.dump` — custom format (compressed)<br>• `pg_restore -d dbname backup.dump` — restore<br>• `pg_dumpall` for entire cluster |
| 18 | Continuous archiving and PITR? | PITR, point-in-time, recovery | • Archive WAL files + base backup = PITR<br>• Restore to any point in time: `recovery_target_time`<br>• Essential for disaster recovery<br>• Requires `archive_mode = on` and archive_command |
| 19 | WAL — why important? | WAL, write-ahead logging | • All changes written to WAL BEFORE modifying data<br>• Crash recovery: replay WAL to restore consistency<br>• Used for: replication (streaming WAL to replicas)<br>• PostgreSQL's durability mechanism |
| 20 | Zero-downtime schema migrations? | Zero-downtime, migration | • Add column with DEFAULT (PG 11+: instant, no rewrite)<br>• Create index CONCURRENTLY<br>• Add constraint NOT VALID → VALIDATE later<br>• Avoid: ALTER TYPE, adding NOT NULL on large tables |
| 21 | Online DDL — add column without lock? | Online DDL, ADD COLUMN | • PG 11+: `ALTER TABLE ADD COLUMN col TYPE DEFAULT val` — instant<br>• Before PG 11: rewrites entire table (full lock)<br>• DROP column is instant (marks as dropped)<br>• CREATE INDEX CONCURRENTLY for no-lock indexing |
| 22 | PostgreSQL extensions — popular ones? | Extension, pg_trgm, PostGIS | • `pg_trgm`: trigram similarity/indexing<br>• `uuid-ossp` / `pgcrypto`: UUID generation<br>• `PostGIS`: geospatial data<br>• `pg_stat_statements`: query statistics<br>• `CREATE EXTENSION extension_name` |
| 23 | Foreign data wrappers (FDW)? | FDW, external data | • Query external data sources from PostgreSQL<br>• `postgres_fdw`: connect to other PG databases<br>• `mysql_fdw`, `file_fdw`, `oracle_fdw` — other sources<br>• `CREATE FOREIGN TABLE` maps external data |
| 24 | Row-level security (RLS)? | RLS, policy, security | • `ALTER TABLE t ENABLE ROW LEVEL SECURITY`<br>• `CREATE POLICY p ON t USING (tenant_id = current_setting('app.tenant'))`<br>• Users only see rows matching policy<br>• Transparent — no application code changes |
| 25 | EXPLAIN (BUFFERS, FORMAT JSON)? | EXPLAIN, BUFFERS, JSON | • `EXPLAIN (ANALYZE, BUFFERS, FORMAT JSON) SELECT ...`<br>• BUFFERS: shows cache hits vs disk reads<br>• FORMAT JSON: machine-readable output<br>• Paste JSON into explain.dalibo.com for visualization |


## Section 21: Cursors, Loops & Procedural SQL

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | What is a cursor? When to use? | Cursor, row-by-row | • Pointer that traverses query results row by row<br>• Use for: complex row-by-row processing that can't be set-based<br>• Generally discouraged — set-based is faster<br>• Useful for: streaming large result sets |
| 2 | Implicit vs explicit cursor? | Implicit, explicit, cursor | • Implicit: auto-created by SQL engine for DML (INSERT/UPDATE/DELETE)<br>• Explicit: declared by programmer with DECLARE<br>• Implicit: automatic open/close/fetch<br>• Explicit: manual lifecycle control |
| 3 | DECLARE, OPEN, FETCH, CLOSE lifecycle? | DECLARE, OPEN, FETCH, CLOSE | • `DECLARE cur CURSOR FOR SELECT ...` — define query<br>• `OPEN cur` — execute query<br>• `FETCH cur INTO var` — get next row<br>• `CLOSE cur` — release resources |
| 4 | Cursor lifecycle stages? | Lifecycle, cursor | • DECLARE: define cursor and its query<br>• OPEN: execute query, position before first row<br>• FETCH: advance and retrieve rows<br>• CLOSE: release server resources<br>• Must follow this order |
| 5 | SCROLL vs NO SCROLL cursor? | SCROLL, NO SCROLL | • `SCROLL`: can move forward and backward (FETCH PRIOR, FIRST, LAST)<br>• `NO SCROLL`: forward-only (FETCH NEXT only)<br>• NO SCROLL is faster and uses less memory<br>• Default: NO SCROLL |
| 6 | WITH HOLD cursor? | WITH HOLD, transaction | • `DECLARE cur CURSOR WITH HOLD FOR ...`<br>• Survives COMMIT — stays open across transactions<br>• Normal cursors close at transaction end<br>• Useful for: long processing spanning transactions |
| 7 | FETCH NEXT, PRIOR, FIRST, LAST? | FETCH, direction | • `FETCH NEXT FROM cur` — next row<br>• `FETCH PRIOR` — previous row (SCROLL only)<br>• `FETCH FIRST` — first row; `FETCH LAST` — last row<br>• `FETCH ABSOLUTE n` — go to nth row |
| 8 | REFCURSOR — return cursor from function? | REFCURSOR, function | • `DECLARE cur REFCURSOR;` in PL/pgSQL<br>• Function can return REFCURSOR to caller<br>• Caller uses `FETCH ALL FROM cur_name`<br>• Useful for dynamic result sets |
| 9 | Cursor vs set-based query? Why cursors discouraged? | Set-based, discouraged | • Set-based: entire result in one operation (fast, optimized)<br>• Cursor: row-by-row processing (slow, more overhead)<br>• SQL engine optimized for set operations<br>• Use set-based with CTEs, window functions, JOINs instead |
| 10 | Performance of cursors on large result sets? | Cursor, performance, large | • Memory: cursor doesn't load all rows at once<br>• But: row-by-row processing is slow<br>• Network: each FETCH = round trip (remote cursors)<br>• Use FETCH n rows at a time for batch processing |
| 11 | FOR ... IN loop in PL/pgSQL? | FOR, loop, PL/pgSQL | • `FOR rec IN SELECT ... LOOP ... END LOOP;`<br>• Auto-opens and closes cursor<br>• `rec` is a RECORD variable<br>• Simplest way to iterate query results |
| 12 | Cursor FOR loop vs regular FOR loop? | Cursor FOR, regular FOR | • Cursor FOR: iterates query results (`FOR rec IN SELECT ...`)<br>• Regular FOR: iterates integers (`FOR i IN 1..10 LOOP`)<br>• Cursor FOR auto-manages cursor lifecycle<br>• Both are PL/pgSQL constructs |
| 13 | UPDATE WHERE CURRENT OF cursor? | WHERE CURRENT OF, update | • `UPDATE t SET col = val WHERE CURRENT OF cur`<br>• Updates the row cursor is currently pointing at<br>• Cursor must be opened with `FOR UPDATE`<br>• Row-by-row update (usually slower than set-based) |
| 14 | Alternatives to cursors? | Alternatives, set-based | • CTEs for step-by-step transformation<br>• Window functions for row-relative calculations<br>• JOINs for combining data<br>• `generate_series()` for looping patterns<br>• Set-based DML for bulk operations |
| 15 | LOOP, WHILE, FOR loops in PL/pgSQL? | LOOP, WHILE, FOR | • `LOOP ... EXIT WHEN condition; END LOOP;` — basic loop<br>• `WHILE condition LOOP ... END LOOP;` — pre-check loop<br>• `FOR i IN 1..10 LOOP ... END LOOP;` — counter loop<br>• Choose based on termination pattern |
| 16 | EXIT and EXIT WHEN? | EXIT, loop control | • `EXIT;` — immediately exit innermost loop<br>• `EXIT WHEN count > 100;` — conditional exit<br>• `EXIT label WHEN condition;` — exit named loop<br>• Equivalent to BREAK in other languages |
| 17 | CONTINUE to skip iterations? | CONTINUE, skip | • `CONTINUE;` — skip rest of current iteration<br>• `CONTINUE WHEN condition;` — conditional skip<br>• Goes to next iteration immediately<br>• Like `continue` in Python/JavaScript |
| 18 | FOREACH for arrays in PL/pgSQL? | FOREACH, array, loop | • `FOREACH elem IN ARRAY my_array LOOP ... END LOOP;`<br>• Iterates each element of an array<br>• `FOREACH elem IN ARRAY ARRAY[1,2,3] LOOP ...`<br>• Simpler than index-based access |
| 19 | Server-side vs client-side cursors? | Server, client, cursor | • Server-side: cursor lives on DB server (PL/pgSQL DECLARE)<br>• Client-side: driver fetches all rows into client memory<br>• Server-side: better for large results (streaming)<br>• Client-side: simpler but uses more client memory |
| 20 | Cursor for batch processing (N rows at a time)? | Batch, cursor, FETCH N | • `FETCH 1000 FROM cur` — get 1000 rows at a time<br>• Process batch, then fetch next 1000<br>• Limits memory usage for large result sets<br>• Better than loading millions of rows at once |


## Section 22: DML — INSERT, UPDATE, DELETE & MERGE

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | Forms of INSERT (single, multi, from SELECT)? | INSERT, multi-row | • Single: `INSERT INTO t(col) VALUES(val)`<br>• Multi: `INSERT INTO t VALUES(1,'a'), (2,'b')`<br>• From SELECT: `INSERT INTO t2 SELECT * FROM t1`<br>• Multi-row is faster than multiple single INSERTs |
| 2 | Insert from one table to another? | INSERT INTO SELECT | • `INSERT INTO archive SELECT * FROM orders WHERE year < 2023`<br>• Column count and types must match<br>• Can transform with expressions in SELECT<br>• Use with WHERE for selective copy |
| 3 | UPSERT: DO NOTHING vs DO UPDATE? | ON CONFLICT, UPSERT | • `DO NOTHING`: skip conflicting rows silently<br>• `DO UPDATE SET col = EXCLUDED.col`: update on conflict<br>• `EXCLUDED` references the attempted-insert row<br>• Requires conflict target (unique constraint/index) |
| 4 | RETURNING with INSERT/UPDATE/DELETE? | RETURNING, result | • `INSERT ... RETURNING id` → get auto-generated ID<br>• `UPDATE ... RETURNING *` → see all updated rows<br>• `DELETE ... RETURNING id, name` → see what was deleted<br>• Eliminates separate SELECT query |
| 5 | Update multiple columns in one UPDATE? | UPDATE, multiple columns | • `UPDATE t SET a = 1, b = 2, c = 3 WHERE id = 5`<br>• All columns updated atomically<br>• Can reference other columns: `SET total = price * qty`<br>• One WHERE applies to all SET clauses |
| 6 | Update from another table (FROM or subquery)? | UPDATE FROM, subquery | • `UPDATE t1 SET t1.col = t2.val FROM t2 WHERE t1.id = t2.id`<br>• Or: `UPDATE t1 SET col = (SELECT val FROM t2 WHERE t2.id = t1.id)`<br>• FROM join is usually faster<br>• FROM is PG-specific; subquery is standard |
| 7 | DELETE vs TRUNCATE — when each? | DELETE, TRUNCATE | • DELETE: row-by-row, can use WHERE, fires triggers, logged<br>• TRUNCATE: removes ALL rows, no WHERE, no triggers, faster<br>• TRUNCATE resets sequences; DELETE doesn't<br>• TRUNCATE in PG is transactional (rollback-able) |
| 8 | Delete based on condition from joined table? | DELETE, JOIN, subquery | • `DELETE FROM orders WHERE customer_id IN (SELECT id FROM customers WHERE inactive)`<br>• PG: `DELETE FROM orders USING customers WHERE orders.customer_id = customers.id AND customers.inactive`<br>• USING is PG-specific but clean |
| 9 | MERGE statement? PG support? | MERGE, SQL standard | • Standard SQL: `MERGE INTO target USING source ON condition WHEN MATCHED THEN UPDATE WHEN NOT MATCHED THEN INSERT`<br>• PG 15+: supports MERGE<br>• Before PG 15: use INSERT ON CONFLICT<br>• MERGE more flexible: handles match/no-match/delete |
| 10 | Bulk updates efficiently? | Bulk update, batch | • Batch: multiple rows in single UPDATE with CASE<br>• Or: create temp table → UPDATE FROM temp<br>• Row-by-row = N round trips (slow)<br>• Batch = 1 round trip (fast) |
| 11 | COPY vs INSERT for bulk load? | COPY, bulk, INSERT | • COPY: 10-100x faster than INSERT<br>• Bypasses per-row SQL parsing<br>• `COPY t FROM '/file.csv' CSV HEADER`<br>• Disable indexes during large COPY, re-create after |
| 12 | UPDATE SET col = subquery vs UPDATE FROM? | Subquery, FROM, UPDATE | • Subquery: correlated — runs per row (can be slow)<br>• FROM: join-based — set-based (faster)<br>• FROM is PG-specific syntax<br>• Both produce same result; FROM is preferred |
| 13 | Swap values between two columns? | Swap, columns | • `UPDATE t SET a = b, b = a WHERE id = 1`<br>• PG evaluates all SET expressions using OLD values<br>• No temp variable needed (unlike procedural code)<br>• SQL standard guarantees this behavior |
| 14 | Multi-table delete? | Multi-table, DELETE | • PG: `DELETE FROM t1 USING t2 WHERE t1.id = t2.id`<br>• Standard: `DELETE FROM t1 WHERE id IN (SELECT id FROM t2)`<br>• USING clause allows join-based delete<br>• Combine with RETURNING for audit |
| 15 | Cascading delete with FK constraints? | CASCADE, FK, delete | • `ON DELETE CASCADE`: auto-deletes children when parent deleted<br>• `ON DELETE SET NULL`: nullifies FK in children<br>• `ON DELETE RESTRICT`: prevents parent deletion<br>• Define on FK constraint at table creation |


## Section 23: Security & Permissions

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | User vs role in PostgreSQL? | User, role, PG | • In PG, users and roles are the same thing<br>• Role with LOGIN = user<br>• Role without LOGIN = group role<br>• `CREATE ROLE analyst LOGIN PASSWORD '...'` |
| 2 | GRANT and REVOKE? | GRANT, REVOKE | • `GRANT SELECT ON table TO role` — give permission<br>• `REVOKE DELETE ON table FROM role` — remove permission<br>• `GRANT role1 TO role2` — role inheritance<br>• `GRANT ALL ON SCHEMA public TO role` |
| 3 | Principle of least privilege? | Least privilege, security | • Grant only minimum permissions needed<br>• App accounts: only SELECT, INSERT, UPDATE on needed tables<br>• Read-only role for reporting/analytics<br>• Separate admin vs application credentials |
| 4 | Read-only access to a schema? | Read-only, GRANT, schema | • `GRANT USAGE ON SCHEMA myschema TO reader`<br>• `GRANT SELECT ON ALL TABLES IN SCHEMA myschema TO reader`<br>• `ALTER DEFAULT PRIVILEGES IN SCHEMA myschema GRANT SELECT ON TABLES TO reader`<br>• Default privileges apply to future tables |
| 5 | Row-level security (RLS) policies? | RLS, policy, CREATE POLICY | • `ALTER TABLE t ENABLE ROW LEVEL SECURITY`<br>• `CREATE POLICY p ON t USING (tenant_id = current_setting('app.tenant_id')::int)`<br>• Users only see/modify matching rows<br>• Transparent to application queries |
| 6 | pg_hba.conf for authentication? | pg_hba.conf, authentication | • Controls HOW clients authenticate to PG<br>• Format: `host database user address method`<br>• Methods: md5, scram-sha-256, trust, reject, cert<br>• Processed top-to-bottom, first match wins |
| 7 | SECURITY DEFINER vs INVOKER for functions? | SECURITY DEFINER, INVOKER | • DEFINER: runs with function OWNER's privileges<br>• INVOKER: runs with CALLER's privileges (default)<br>• DEFINER: controlled access to restricted tables<br>• Always set `search_path` in DEFINER functions |
| 8 | Rotate database credentials in production? | Rotate, credentials | • Use connection pooler (PgBouncer) — rotate without app restart<br>• Create new role → update app config → drop old role<br>• Secrets manager (AWS Secrets Manager, Vault) for automation<br>• Never hardcode credentials in code |
| 9 | Audit database access and queries? | Audit, logging | • `log_statement = 'all'` — log every query<br>• `pgAudit` extension — structured audit logging<br>• Trigger-based audit tables for data changes<br>• `log_connections = on` for connection auditing |
| 10 | SSL/TLS for database connections? | SSL, TLS, encryption | • `ssl = on` in postgresql.conf<br>• Requires certificate and key files<br>• pg_hba.conf: `hostssl` forces SSL<br>• Encrypts data in transit between client and server |
| 11 | Prevent SQL injection at DB level? | SQL injection, prevention | • Use parameterized queries / prepared statements<br>• In functions: `EXECUTE format('...%I...%L...', ident, literal)`<br>• Never concatenate user input into SQL strings<br>• Review application code and ORM usage |
| 12 | Manage secrets and connection strings? | Secrets, connection string | • Environment variables: `DATABASE_URL`<br>• Secrets managers: AWS Secrets Manager, HashiCorp Vault<br>• `.env` files (not committed to git)<br>• Never put credentials in source code |
| 13 | PUBLIC role — why revoke defaults? | PUBLIC, default privileges | • PUBLIC role: every user has it<br>• By default: PUBLIC can connect to database, use public schema<br>• `REVOKE ALL ON DATABASE db FROM PUBLIC`<br>• `REVOKE CREATE ON SCHEMA public FROM PUBLIC` — prevent schema pollution |
| 14 | Column-level permissions? | Column, permission, GRANT | • `GRANT SELECT(name, email) ON users TO analyst`<br>• Restricts access to specific columns<br>• More granular than table-level GRANT<br>• Or: use views to expose only certain columns |
| 15 | Multi-tenancy at database level? | Multi-tenancy, isolation | • Shared DB + `tenant_id` + RLS: simplest, most common<br>• Schema per tenant: better isolation, more schemas to manage<br>• Database per tenant: full isolation, hardest to manage<br>• Choose based on: isolation needs, number of tenants |


## Section 24: SQL Scenario & Design Questions

| # | Question | Keywords | Answer |
|---|----------|----------|--------|
| 1 | Second highest salary per department? | Second highest, dept | • `WITH r AS (SELECT *, DENSE_RANK() OVER (PARTITION BY dept ORDER BY salary DESC) rk FROM emp) SELECT * FROM r WHERE rk = 2`<br>• DENSE_RANK handles ties<br>• Or: correlated subquery with COUNT(DISTINCT) |
| 2 | Employees earning more than their manager? | Self-join, manager | • `SELECT e.name FROM emp e JOIN emp m ON e.manager_id = m.id WHERE e.salary > m.salary`<br>• Self-join comparing employee to their manager<br>• Can extend to include manager name/salary |
| 3 | Customers who never placed an order? | Anti-join, no orders | • `SELECT c.* FROM customers c LEFT JOIN orders o ON c.id = o.customer_id WHERE o.id IS NULL`<br>• Or: `WHERE NOT EXISTS (SELECT 1 FROM orders WHERE customer_id = c.id)`<br>• Anti-join pattern |
| 4 | Find gaps in sequence of IDs? | Gaps, sequence, LEAD | • `SELECT id + 1 AS gap_start, LEAD(id) OVER (ORDER BY id) - 1 AS gap_end FROM t WHERE LEAD(id) OVER (ORDER BY id) - id > 1`<br>• Or: `generate_series LEFT JOIN` approach<br>• Finds missing IDs in sequence |
| 5 | Pivot rows into columns (cross-tab)? | Pivot, crosstab | • `SELECT dept, SUM(CASE WHEN year = 2023 THEN rev END) y23, SUM(CASE WHEN year = 2024 THEN rev END) y24 FROM sales GROUP BY dept`<br>• Or PG: `tablefunc crosstab()` function<br>• Conditional aggregation pattern |
| 6 | Unpivot columns into rows? | Unpivot, UNNEST, VALUES | • `SELECT id, col_name, col_value FROM t, LATERAL (VALUES ('col1', t.col1), ('col2', t.col2)) AS u(col_name, col_value)`<br>• PG: `UNNEST` + array approach<br>• Turns wide table into narrow |
| 7 | Month-over-month revenue growth? | MoM, growth, LAG | • `SELECT month, revenue, revenue - LAG(revenue) OVER (ORDER BY month) AS growth, ((revenue - LAG(revenue) OVER (ORDER BY month)) / LAG(revenue) OVER (ORDER BY month)) * 100 AS pct_growth FROM monthly_rev`<br>• Window function pattern |
| 8 | Longest consecutive login streak per user? | Streak, consecutive, islands | • `WITH grp AS (SELECT user_id, login_date, login_date - ROW_NUMBER() OVER (PARTITION BY user_id ORDER BY login_date)::int AS island FROM logins) SELECT user_id, COUNT(*) AS streak FROM grp GROUP BY user_id, island ORDER BY streak DESC`<br>• Gaps and islands technique |
| 9 | Identify and keep only one duplicate? | Dedup, ROW_NUMBER, DELETE | • `WITH dups AS (SELECT ctid, ROW_NUMBER() OVER (PARTITION BY email ORDER BY id) rn FROM users) DELETE FROM users WHERE ctid IN (SELECT ctid FROM dups WHERE rn > 1)`<br>• Keeps first, deletes rest |
| 10 | Running total partitioned by category? | Running total, PARTITION | • `SELECT category, date, amount, SUM(amount) OVER (PARTITION BY category ORDER BY date) running_total FROM sales`<br>• Cumulative sum resets per category<br>• Default frame: UNBOUNDED PRECEDING to CURRENT ROW |
| 11 | Median without median function? | Median, PERCENTILE_CONT | • PG: `PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY val)`<br>• Without: `SELECT val FROM (SELECT val, ROW_NUMBER() OVER (ORDER BY val) rn, COUNT(*) OVER () cnt FROM t) WHERE rn IN (cnt/2, cnt/2 + 1)`<br>• Average middle values for even count |
| 12 | Rank products by sales within category? | Rank, category, sales | • `SELECT product, category, sales, RANK() OVER (PARTITION BY category ORDER BY sales DESC) AS rank FROM products`<br>• Use DENSE_RANK if ties should have same rank<br>• Filter with CTE: `WHERE rank <= 5` for top 5 |
| 13 | Users active every day in last week? | Active, every day | • `SELECT user_id FROM activity WHERE login_date >= CURRENT_DATE - 7 GROUP BY user_id HAVING COUNT(DISTINCT login_date) = 7`<br>• COUNT DISTINCT dates = 7 means every day<br>• Adjust date range as needed |
| 14 | Retention rates (Day 1, Day 7, Day 30)? | Retention, cohort | • `WITH cohort AS (SELECT user_id, MIN(signup_date) d FROM users GROUP BY user_id) SELECT c.d, COUNT(DISTINCT CASE WHEN a.date = c.d + 1 THEN a.user_id END)::float / COUNT(DISTINCT c.user_id) AS day1_retention FROM cohort c LEFT JOIN activity a ON c.user_id = a.user_id GROUP BY c.d`<br>• Cohort analysis pattern |
| 15 | Find overlapping date ranges? | Overlap, date range | • `SELECT a.*, b.* FROM bookings a JOIN bookings b ON a.id < b.id AND a.start_date < b.end_date AND b.start_date < a.end_date`<br>• Overlap condition: A.start < B.end AND B.start < A.end<br>• Self-join with `a.id < b.id` avoids duplicates |
| 16 | Merge overlapping intervals? | Merge, intervals | • `WITH sorted AS (SELECT *, LAG(end_date) OVER (ORDER BY start_date) AS prev_end FROM intervals), flagged AS (SELECT *, CASE WHEN start_date <= prev_end THEN 0 ELSE 1 END AS new_group FROM sorted), grouped AS (SELECT *, SUM(new_group) OVER (ORDER BY start_date) AS grp FROM flagged) SELECT MIN(start_date), MAX(end_date) FROM grouped GROUP BY grp`<br>• Multi-step CTE approach |
| 17 | Efficient pagination on large tables? | Pagination, keyset | • Offset-based: `LIMIT 10 OFFSET 10000` — slow for deep pages<br>• Keyset: `WHERE id > last_seen_id ORDER BY id LIMIT 10`<br>• Keyset uses index seek — constant speed<br>• Client sends last_seen cursor |
| 18 | Top 3 products bought together? | Frequently bought, pair | • Self-join on order_id to find product pairs<br>• `SELECT p1, p2, COUNT(*) FROM order_items a JOIN order_items b ON a.order_id = b.order_id AND a.product_id < b.product_id GROUP BY p1, p2 ORDER BY count DESC LIMIT 3`<br>• Market basket analysis |
| 19 | Employees in same department (self-join pairs)? | Self-join, pairs | • `SELECT e1.name, e2.name FROM emp e1 JOIN emp e2 ON e1.dept = e2.dept AND e1.id < e2.id`<br>• `e1.id < e2.id` prevents duplicate pairs (A,B) and (B,A)<br>• Also excludes self-pairs (A,A) |
| 20 | Design schema for social media feed? | Social media, schema | • `users`, `posts`, `comments`, `likes`, `follows`, `media`<br>• `follows(follower_id, following_id)` — self-referencing M:N<br>• `likes(user_id, post_id)` — junction table<br>• Feed: `SELECT p.* FROM posts p JOIN follows f ON p.user_id = f.following_id WHERE f.follower_id = current_user ORDER BY p.created_at DESC` |
| 21 | E-commerce schema (products, orders, inventory)? | E-commerce, schema | • `products(id, name, sku, price)`, `categories`, `product_categories`<br>• `orders(id, user_id, total, status, created_at)`<br>• `order_items(order_id, product_id, qty, unit_price)` — track price at time of order<br>• `inventory(product_id, warehouse_id, qty)` |
| 22 | Booking system with availability slots? | Booking, availability | • `resources(id, name)`, `bookings(id, resource_id, start_time, end_time, user_id)`<br>• Exclusion constraint to prevent overlapping: `EXCLUDE USING gist (resource_id WITH =, tstzrange(start_time, end_time) WITH &&)`<br>• Requires `btree_gist` extension |
| 23 | Messaging/chat application schema? | Chat, messaging, schema | • `conversations(id, type, created_at)`, `conversation_members(conv_id, user_id)`<br>• `messages(id, conv_id, sender_id, body, sent_at)`<br>• Index on (conv_id, sent_at) for message ordering<br>• `read_receipts(message_id, user_id, read_at)` |
| 24 | Optimize query taking 30s on 100M rows? | Optimize, 100M, slow | • Add appropriate indexes (`EXPLAIN ANALYZE` first)<br>• Partition table by date/range<br>• Use covering index for index-only scan<br>• Consider materialized view for aggregations<br>• Check `work_mem` for sort/hash operations |
